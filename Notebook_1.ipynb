<h2><span style="color:lightblue"><b>Project Overview</b></span></h2>


Rossmann Pharmaceuticals operates retail stores across multiple cities and requires an accurate, scalable method to forecast daily sales. The finance team needs reliable sales projections six weeks in advance to support budgeting, inventory planning, staffing, promotion planning, and business-performance monitoring.
Currently, store managers prepare forecasts primarily using personal experience and judgement. While this provides local insight, it can lead to inconsistent forecasts and makes it difficult to plan effectively across all stores.

In a competitive retail environment, reliable sales forecasting transforms
historical information into practical decisions about inventory, staffing,
promotions, and financial planning.

Accurate sales forecasting is a fundamental component of successful retail
management. It allows organizations to anticipate customer demand, maintain
appropriate inventory levels, allocate employees effectively, design targeted
promotional campaigns, and make informed financial decisions. For an
organization such as <strong style="color: #153c59;">Rossmann
Pharmaceuticals</strong>, which manages a large network of stores, even small
improvements in forecasting accuracy can produce meaningful operational and
financial benefits.


<h2><span style="color:lightblue"><b>Business Objective</b></span></h2>


The objective is to develop an end-to-end machine learning solution that predicts daily sales for every Rossmann store up to six weeks ahead. The system should use historical sales data along with relevant business and external factors, including:
- Store promotions and continuing promotion campaigns
- State and school holidays
- Seasonal and calendar patterns
- Store type and assortment level
- Competition distance and competitor opening dates
- Store opening status and day-of-week behaviour
- Local store characteristics


<h2><span style="color:lightblue"><b>Analytical Solution</b></span></h2>


The solution will include data cleaning, exploratory analysis, feature engineering, machine learning and deep-learning models, experiment tracking, and a web-based prediction interface. Finance analysts and store managers will be able to upload future store-date data, receive predicted sales and customer counts, view forecast trends.


<h2><span style="color:lightblue"><b>Expected Business Value</b></span></h2>



A reliable forecasting platform will help Rossmann Pharmaceuticals make more informed operational and financial decisions, reduce forecasting inconsistencies, improve inventory allocation, optimize promotional activity, and support proactive planning across all retail locations.

To explore customer purchasing behaviour across Rossmann stores and generate
reliable, evidence-based insights that can support business decisions and the
development of future sales-forecasting models.

<h2><span style="color:lightblue"><b>Clean and Merge Data</b></span></h2>


```python

if "train" not in globals():
    train = pd.read_csv(RAW_DIR / "train.csv", low_memory=False)

if "test" not in globals():
    test = pd.read_csv(RAW_DIR / "test.csv", low_memory=False)

if "stores" not in globals():
    stores = pd.read_csv(RAW_DIR / "store.csv", low_memory=False)


def clean_and_merge(daily, store_data):
    df = daily.copy(); df['Date'] = pd.to_datetime(df['Date'])
    df = df.merge(store_data, on='Store', how='left', validate='many_to_one')
    for col in ['StoreType', 'Assortment', 'PromoInterval']:
        if col in df: df[col] = df[col].fillna('Unknown').astype(str)
    for col in ['CompetitionDistance','CompetitionOpenSinceMonth','CompetitionOpenSinceYear','Promo2','Promo2SinceWeek','Promo2SinceYear']:
        if col in df: df[col] = pd.to_numeric(df[col], errors='coerce')
    df['StateHoliday'] = df['StateHoliday'].fillna('0').astype(str)
    for col in ['Open', 'Promo', 'SchoolHoliday']:
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)
    return df.sort_values(['Store','Date']).reset_index(drop=True)

train_df, test_df = clean_and_merge(train, stores), clean_and_merge(test, stores)
display(train_df.head())
display(test_df.head())
```

Each variable is examined with respect to its data type, completeness, and

<h2><span style="color:lightblue"><b>Missing-Value Assessment</b></span></h2>

Missing values are quantified for each dataset using both absolute counts and
percentages. This allows high-priority data-quality issues to be distinguished
from low-frequency or structurally meaningful missing information.

```python
def missing_summary(df, dataset_name):
    result = pd.DataFrame({
        "Dataset": dataset_name,
        "Feature": df.columns,
        "Missing": df.isna().sum().values,
        "Missing %": (df.isna().mean().values * 100)
    })

    return result[result["Missing"] > 0]


missing_values = pd.concat(
    [missing_summary(df, name) for name, df in datasets.items()],
    ignore_index=True
).sort_values(["Dataset", "Missing %"], ascending=[True, False])

display(
    missing_values.style
    .format({
        "Missing": "{:,}",
        "Missing %": "{:.2f}%"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold")
            ]
        }
    ])
)

logger.info("Missing-value assessment completed.")
```

<h2><span style="color:lightblue"><b>Business-Rule and Logical Consistency Checks</b></span></h2>



Beyond conventional missing-value and duplicate checks, the dataset is assessed
against basic business rules. These checks identify observations that would be
commercially implausible or internally inconsistent, such as negative sales,
invalid binary indicators, or closed stores reporting active sales.

```python
validation_checks = {
    "Negative Sales": (train["Sales"] < 0).sum(),
    "Negative Customers": (train["Customers"] < 0).sum(),
    "Closed Store with Positive Sales":
        ((train["Open"] == 0) & (train["Sales"] > 0)).sum(),
    "Closed Store with Customers":
        ((train["Open"] == 0) & (train["Customers"] > 0)).sum(),
    "Invalid Train Promo":
        (~train["Promo"].isin([0, 1])).sum(),
    "Invalid Train Open":
        (~train["Open"].dropna().isin([0, 1])).sum(),
    "Invalid School Holiday":
        (~train["SchoolHoliday"].isin([0, 1])).sum(),
    "Negative Competition Distance":
        (stores["CompetitionDistance"].dropna() < 0).sum()
}

validation_summary = pd.DataFrame(
    validation_checks.items(),
    columns=["Validation Rule", "Violations"]
)

validation_summary["Status"] = np.where(
    validation_summary["Violations"] == 0,
    "PASS",
    "REVIEW"
)

display(
    validation_summary.style
    .format({"Violations": "{:,}"})
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold")
            ]
        }
    ])
)

logger.info("Business-rule validation completed.")
```


<h2><span style="color:lightblue"><b>Missing Value Treatment</b></span></h2>


Missing values are treated according to the information represented by each
feature rather than through a single universal imputation method.

```python
train_clean = train.copy()
test_clean = test.copy()
store_clean = stores.copy()

logger.info("Clean working copies created.")
```

```python
for df in [train_clean, test_clean]:
    df["StateHoliday"] = (
        df["StateHoliday"]
        .astype("string")
        .str.strip()
        .str.lower()
        .replace({"0.0": "0"})
    )

valid_holidays = {"0", "a", "b", "c"}

print("Train StateHoliday values:",
      sorted(train_clean["StateHoliday"].dropna().unique()))

print("Test StateHoliday values:",
      sorted(test_clean["StateHoliday"].dropna().unique()))

invalid_train_holiday = (
    ~train_clean["StateHoliday"].isin(valid_holidays)
    & train_clean["StateHoliday"].notna()
).sum()

invalid_test_holiday = (
    ~test_clean["StateHoliday"].isin(valid_holidays)
    & test_clean["StateHoliday"].notna()
).sum()

print("\nInvalid Train StateHoliday values:", invalid_train_holiday)
print("Invalid Test StateHoliday values :", invalid_test_holiday)

logger.info("StateHoliday representations standardised.")
```

```python
open_pattern = (
    train_clean
    .groupby(["Store", "DayOfWeek"])["Open"]
    .median()
)

missing_open = test_clean["Open"].isna()

for idx in test_clean.index[missing_open]:
    key = (
        test_clean.at[idx, "Store"],
        test_clean.at[idx, "DayOfWeek"]
    )

    if key in open_pattern.index:
        test_clean.at[idx, "Open"] = open_pattern.loc[key]

test_clean["Open"] = test_clean["Open"].fillna(1).astype(int)

print("Missing Open values remaining:", test_clean["Open"].isna().sum())

logger.info("Missing test Open values imputed from historical store-day opening patterns.")
```

```python
store_clean["CompetitionDistanceMissing"] = (
    store_clean["CompetitionDistance"].isna().astype(int)
)

store_clean["CompetitionOpenInfoAvailable"] = (
    store_clean["CompetitionOpenSinceMonth"].notna()
    & store_clean["CompetitionOpenSinceYear"].notna()
).astype(int)

competition_medians = (
    store_clean.groupby("StoreType")["CompetitionDistance"]
    .median()
)

store_clean["CompetitionDistance"] = (
    store_clean["CompetitionDistance"]
    .fillna(
        store_clean["StoreType"].map(competition_medians)
    )
    .fillna(store_clean["CompetitionDistance"].median())
)

print(
    "Missing CompetitionDistance:",
    store_clean["CompetitionDistance"].isna().sum()
)

logger.info(
    "Competition distance imputed while preserving a missingness indicator."
)
```

```python
no_promo2 = store_clean["Promo2"].eq(0)

store_clean.loc[
    no_promo2,
    ["Promo2SinceWeek", "Promo2SinceYear"]
] = 0

store_clean.loc[
    no_promo2,
    "PromoInterval"
] = "None"

promo2_missing_after = store_clean[
    [
        "Promo2SinceWeek",
        "Promo2SinceYear",
        "PromoInterval"
    ]
].isna().sum()

display(
    promo2_missing_after.to_frame(
        "Remaining Missing Values"
    )
)

logger.info("Structural Promo2 missing values treated.")
```

<h2><span style="color:lightblue"><b>Duplicate Treatment</b></span></h2>



Exact duplicate observations and repeated business keys are reviewed before
removal. A historical store-date combination should represent a single daily
observation; therefore, duplicate records are removed only where they represent
true replication rather than legitimate business activity.

```python
before = {
    "Train": len(train_clean),
    "Test": len(test_clean),
    "Store": len(store_clean)
}

train_clean = train_clean.drop_duplicates().copy()
test_clean = test_clean.drop_duplicates().copy()
store_clean = store_clean.drop_duplicates().copy()

after = {
    "Train": len(train_clean),
    "Test": len(test_clean),
    "Store": len(store_clean)
}

duplicate_changes = pd.DataFrame({
    "Dataset": before.keys(),
    "Rows Before": before.values(),
    "Rows After": after.values()
})

duplicate_changes["Rows Removed"] = (
    duplicate_changes["Rows Before"]
    - duplicate_changes["Rows After"]
)

display(duplicate_changes)

logger.info("Exact duplicate treatment completed.")
```

<h2><span style="color:lightblue"><b>Outlier Assessment</b></span></h2>

Detecting Unusual Sales and Customer Behaviour

Extreme observations are not automatically treated as errors. Retail sales can
naturally exhibit large peaks during promotions, major holidays, seasonal
periods, and exceptional store activity. Removing such observations without
investigation could eliminate economically meaningful purchasing behaviour.

```python
def iqr_limits(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    return (
        q1 - 1.5 * iqr,
        q3 + 1.5 * iqr
    )


outlier_records = []

for feature in ["Sales", "Customers"]:
    lower, upper = iqr_limits(train_clean[feature])

    count = (
        (train_clean[feature] < lower)
        | (train_clean[feature] > upper)
    ).sum()

    outlier_records.append({
        "Feature": feature,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Potential Outliers": count,
        "Outlier %": count / len(train_clean) * 100
    })

competition_lower, competition_upper = iqr_limits(
    store_clean["CompetitionDistance"]
)

competition_outliers = (
    (store_clean["CompetitionDistance"] < competition_lower)
    | (store_clean["CompetitionDistance"] > competition_upper)
).sum()

outlier_records.append({
    "Feature": "CompetitionDistance",
    "Lower Bound": competition_lower,
    "Upper Bound": competition_upper,
    "Potential Outliers": competition_outliers,
    "Outlier %": competition_outliers / len(store_clean) * 100
})

outlier_summary = pd.DataFrame(outlier_records)

display(
    outlier_summary.style.format({
        "Lower Bound": "{:,.2f}",
        "Upper Bound": "{:,.2f}",
        "Potential Outliers": "{:,}",
        "Outlier %": "{:.2f}%"
    })
)

logger.info("IQR-based potential outlier detection completed.")
```

```python
for feature in ["Sales", "Customers"]:
    lower, upper = iqr_limits(train_clean[feature])

    train_clean[f"{feature}Outlier"] = (
        (train_clean[feature] < lower)
        | (train_clean[feature] > upper)
    ).astype(int)

lower, upper = iqr_limits(
    store_clean["CompetitionDistance"]
)

store_clean["CompetitionDistanceOutlier"] = (
    (store_clean["CompetitionDistance"] < lower)
    | (store_clean["CompetitionDistance"] > upper)
).astype(int)

logger.info("Outlier indicators created without deleting observations.")
```

<h2><span style="color:lightblue"><b>Visualisation of Extreme Observations</b></span></h2>



Distribution and box-plot views are used to understand the magnitude and
frequency of extreme values. Because the dataset contains more than one million
historical observations, the visualisations focus on interpretable aggregate
patterns rather than plotting every individual record.

<h2><span style="color:lightblue">Promotion Intelligence<b></b></span></h2>


The first required investigation compares promotional activity between the
historical training period and the future test period. A forecasting model will
learn the relationship between promotions and sales from historical data;
therefore, a substantial change in promotional exposure between train and test
could indicate a distribution shift and affect future model performance.

```python
promo_distribution = (
    pd.concat(
        [
            train_clean["Promo"]
            .value_counts(normalize=True)
            .mul(100)
            .rename("Training"),

            test_clean["Promo"]
            .value_counts(normalize=True)
            .mul(100)
            .rename("Test")
        ],
        axis=1
    )
    .fillna(0)
    .rename(index={0: "No Promo", 1: "Promo"})
    .T
)

promo_distribution["Total"] = promo_distribution.sum(axis=1)

display(
    promo_distribution[["No Promo", "Promo"]]
    .style
    .format("{:.2f}%")
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold"),
                ("text-align", "center")
            ]
        },
        {
            "selector": "td",
            "props": [
                ("text-align", "center")
            ]
        }
    ])
)

logger.info("Training and test promotion distributions calculated.")
```

```python
promo_plot = promo_distribution[["No Promo", "Promo"]]

fig, ax = plt.subplots(figsize=(11, 5.5))

left = np.zeros(len(promo_plot))

for column, color in zip(
    ["No Promo", "Promo"],
    ["#E8DCC8", "#C99A3D"]
):
    values = promo_plot[column].values

    bars = ax.barh(
        promo_plot.index,
        values,
        left=left,
        color=color,
        height=0.55,
        label=column
    )

    for bar, value, start in zip(bars, values, left):
        if value >= 5:
            ax.text(
                start + value / 2,
                bar.get_y() + bar.get_height() / 2,
                f"{value:.1f}%",
                ha="center",
                va="center",
                fontsize=11,
                fontweight="bold"
            )

    left += values


ax.set_title(
    "Promotion Distribution: Training vs Test",
    loc="left",
    fontsize=12,
    fontweight="bold",
    # color=NAVY,
    pad=16
)

ax.text(
    0,
    1.035,
    "Percentage of store-day observations under promotional and normal conditions",
    transform=ax.transAxes,
    fontsize=10.5,
    # color=SLATE
)

ax.set_xlim(0, 100)
ax.set_xlabel("Share of Observations (%)")
ax.set_ylabel("")

ax.spines[["top", "right", "left"]].set_visible(False)
ax.grid(axis="x", alpha=0.15)

ax.legend(
    frameon=False,
    ncol=2,
    loc="lower center",
    bbox_to_anchor=(0.5, -0.28)
)

plt.tight_layout()
plt.show()
```

```python
train_promo_pct = train_clean["Promo"].mean() * 100
test_promo_pct = test_clean["Promo"].mean() * 100

promo_gap = test_promo_pct - train_promo_pct

promo_comparison = pd.DataFrame({
    "Metric": [
        "Training Promo Share",
        "Test Promo Share",
        "Absolute Difference",
        "Relative Change"
    ],
    "Value": [
        train_promo_pct,
        test_promo_pct,
        abs(promo_gap),
        (
            (test_promo_pct - train_promo_pct)
            / train_promo_pct * 100
        )
    ]
})

display(
    promo_comparison.style
    .format({"Value": "{:.2f}%"})
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold")
            ]
        }
    ])
)

logger.info(
    "Promo distribution comparison completed | Train=%.2f%% | Test=%.2f%%",
    train_promo_pct,
    test_promo_pct
)
```

```python
if abs(promo_gap) < 2:
    similarity = "highly similar"
    implication = (
        "There is little evidence of a meaningful promotion-distribution shift "
        "between the historical and forecasting periods."
    )

elif abs(promo_gap) < 5:
    similarity = "reasonably similar"
    implication = (
        "A modest difference exists, but promotional exposure remains broadly "
        "comparable between the two datasets."
    )

else:
    similarity = "materially different"
    implication = (
        "The future forecasting period contains a noticeably different level "
        "of promotional exposure, which should be considered during modelling "
        "and validation."
    )

print(f"""
PROMOTION DISTRIBUTION INTERPRETATION

Training promo share : {train_promo_pct:.2f}%
Test promo share     : {test_promo_pct:.2f}%
Absolute gap         : {abs(promo_gap):.2f} percentage points

Assessment:
Promotion distributions are {similarity}.

Modelling implication:
{implication}
""")
```

<h2><span style="color:lightblue">Interpretation<b></b></span></h2>


Promotion Exposure Across Historical and Future Periods

The comparison evaluates whether the forecasting period is exposed to a
substantially different promotional environment from the historical data used
for analysis and model development. Similar promotional proportions indicate
that the model will encounter a broadly familiar balance of promotional and
non-promotional store-days when applied to the test period.

<h2><span style="color:lightblue">Holiday Intelligence<b></b></span></h2>



Purchasing Behaviour Before, During and After Holidays

Major holidays can alter purchasing behaviour well before the holiday itself.
Customers may bring purchases forward, stores may operate differently, and
demand may subsequently decline once the event has passed. A simple comparison
between holiday and non-holiday observations would therefore overlook much of
the underlying behavioural pattern.

To capture this effect, each state-holiday occurrence is surrounded by a
<strong>14-day event window</strong>: seven days before the holiday, the
holiday itself, and seven days after. Sales, customer traffic, purchasing value,
and store-opening behaviour are then evaluated relative to the holiday date.


<strong style="color:#102e87;">Event timeline:</strong>
Day −7 → Day −1 &nbsp; | &nbsp;
<strong>Day 0 = Holiday</strong> &nbsp; | &nbsp;
Day +1 → Day +7


```python
holiday_map = {
    "0": "No State Holiday",
    "a": "Public Holiday",
    "b": "Easter Holiday",
    "c": "Christmas"
}

eda_df["StateHoliday"] = (
    eda_df["StateHoliday"]
    .astype("string")
    .str.strip()
    .str.lower()
)

eda_df["HolidayType"] = (
    eda_df["StateHoliday"]
    .map(holiday_map)
)

holiday_events = (
    eda_df.loc[
        eda_df["StateHoliday"].isin(["a", "b", "c"]),
        ["Store", "Date", "HolidayType"]
    ]
    .drop_duplicates()
    .rename(columns={"Date": "HolidayDate"})
    .reset_index(drop=True)
)

holiday_events["EventID"] = np.arange(
    1,
    len(holiday_events) + 1
)

event_source = eda_df[
    [
        "Store",
        "Date",
        "Sales",
        "Customers",
        "Open",
        "Promo",
        "SchoolHoliday"
    ]
].copy()

holiday_windows = []

for relative_day in range(-7, 8):

    lookup = holiday_events.copy()

    lookup["Date"] = (
        lookup["HolidayDate"]
        + pd.to_timedelta(relative_day, unit="D")
    )

    matched = lookup.merge(
        event_source,
        on=["Store", "Date"],
        how="inner"
    )

    matched["RelativeDay"] = relative_day

    holiday_windows.append(matched)

holiday_window = pd.concat(
    holiday_windows,
    ignore_index=True
)

holiday_window["Period"] = pd.cut(
    holiday_window["RelativeDay"],
    bins=[-np.inf, -1, 0, np.inf],
    labels=[
        "Before Holiday",
        "Holiday",
        "After Holiday"
    ],
    include_lowest=True
)

print("HOLIDAY EVENT WINDOW CREATED SUCCESSFULLY")
print("-" * 48)

print(
    f"Holiday events identified : "
    f"{len(holiday_events):,}"
)

print(
    f"Event-window observations : "
    f"{len(holiday_window):,}"
)

print(
    f"Relative-day coverage     : "
    f"{holiday_window['RelativeDay'].min()} "
    f"to "
    f"{holiday_window['RelativeDay'].max()}"
)

print("\nHoliday types:")
display(
    holiday_events["HolidayType"]
    .value_counts()
    .rename_axis("Holiday Type")
    .to_frame("Events")
)

print("\nEvent periods:")
display(
    holiday_window["Period"]
    .value_counts(sort=False)
    .rename_axis("Period")
    .to_frame("Observations")
)

logger.info(
    "Holiday event-window dataset created successfully | "
    "Events=%s | Observations=%s",
    len(holiday_events),
    len(holiday_window)
)
```

<h2><span style="color:lightblue">Holiday Event Study<b></b></span></h2>


How Does Customer Demand Change Around a Holiday?

To understand whether customers shift their purchases around major holidays,
sales and customer traffic are examined across the seven days preceding the
holiday, the holiday itself, and the seven days immediately afterwards.


Both measures are converted into indices using the average activity during the
seven pre-holiday days as a baseline of <strong>100</strong>. Values above 100
indicate stronger-than-baseline demand, while values below 100 indicate weaker
activity.


```python
holiday_daily = (
    holiday_window
    .groupby("RelativeDay", as_index=False)
    .agg(
        AverageSales=("Sales", "mean"),
        MedianSales=("Sales", "median"),
        AverageCustomers=("Customers", "mean"),
        OpenRate=("Open", "mean"),
        PromoRate=("Promo", "mean")
    )
    .sort_values("RelativeDay")
)

baseline = (
    holiday_daily.loc[
        holiday_daily["RelativeDay"].between(-7, -1),
        ["AverageSales", "AverageCustomers"]
    ]
    .mean()
)

holiday_daily["SalesIndex"] = (
    holiday_daily["AverageSales"]
    / baseline["AverageSales"]
    * 100
)

holiday_daily["CustomerIndex"] = (
    holiday_daily["AverageCustomers"]
    / baseline["AverageCustomers"]
    * 100
)

holiday_daily["OpenRate"] *= 100
holiday_daily["PromoRate"] *= 100

display(
    holiday_daily.style
    .format({
        "AverageSales": "{:,.2f}",
        "MedianSales": "{:,.2f}",
        "AverageCustomers": "{:,.2f}",
        "OpenRate": "{:.2f}%",
        "PromoRate": "{:.2f}%",
        "SalesIndex": "{:.2f}",
        "CustomerIndex": "{:.2f}"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold"),
                ("text-align", "center")
            ]
        }
    ])
)

logger.info(
    "Holiday relative-day event-study metrics calculated."
)
```

```python
fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    holiday_daily["RelativeDay"],
    holiday_daily["SalesIndex"],
    marker="o",
    markersize=6,
    linewidth=2.4,
    # color=NAVY,
    label="Sales Index"
)

ax.plot(
    holiday_daily["RelativeDay"],
    holiday_daily["CustomerIndex"],
    marker="o",
    markersize=6,
    linewidth=2.2,
    # color=GOLD,
    label="Customer Index"
)

ax.axhline(
    100,
    # color=SLATE,
    linestyle="--",
    linewidth=1.1,
    alpha=0.65
)

ax.axvline(
    0,
    # color=GOLD,
    linestyle="--",
    linewidth=1.6
)

ax.annotate(
    "Holiday",
    xy=(0, holiday_daily["SalesIndex"].max()),
    xytext=(0.4, holiday_daily["SalesIndex"].max()),
    fontsize=10,
    # color=SLATE
)

ax.set_title(
    "Sales and Customer Behaviour Around State Holidays",
    loc="left",
    fontsize=17,
    fontweight="bold",
    # color=NAVY,
    pad=16
)

ax.text(
    0,
    1.03,
    "Index based on average activity during Days −7 to −1 = 100",
    transform=ax.transAxes,
    fontsize=10.5,
    # color=SLATE
)

ax.set_xlabel("Days Relative to Holiday")
ax.set_ylabel("Demand Index")

ax.set_xticks(range(-7, 8))

ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="both", alpha=0.16)

ax.legend(
    frameon=False,
    ncol=2,
    loc="upper right"
)

plt.tight_layout()
plt.show()

logger.info(
    "Holiday sales and customer event-study chart generated."
)
```

<h2><span style="color:lightblue">Purchasing Behaviour Before, During and After Holidays <b></b></span></h2>


 Holiday Intelligence




Major holidays can alter purchasing behaviour well before the holiday itself.
Customers may bring purchases forward, stores may operate differently, and
demand may subsequently decline once the event has passed. A simple comparison
between holiday and non-holiday observations would therefore overlook much of
the underlying behavioural pattern.



To capture this effect, each state-holiday occurrence is surrounded by a
<strong>14-day event window</strong>: seven days before the holiday, the
holiday itself, and seven days after. Sales, customer traffic, purchasing value,
and store-opening behaviour are then evaluated relative to the holiday date.


<strong style="color:#102e46;">Event timeline:</strong>
Day −7 → Day −1 &nbsp; | &nbsp;
<strong>Day 0 = Holiday</strong> &nbsp; | &nbsp;
Day +1 → Day +7


```python
holiday_map = {
    "0": "No State Holiday",
    "a": "Public Holiday",
    "b": "Easter Holiday",
    "c": "Christmas"
}

eda_df["StateHoliday"] = (
    eda_df["StateHoliday"]
    .astype("string")
    .str.strip()
    .str.lower()
)

eda_df["HolidayType"] = (
    eda_df["StateHoliday"]
    .map(holiday_map)
)

holiday_events = (
    eda_df.loc[
        eda_df["StateHoliday"].isin(["a", "b", "c"]),
        ["Store", "Date", "HolidayType"]
    ]
    .drop_duplicates()
    .rename(columns={"Date": "HolidayDate"})
    .reset_index(drop=True)
)

holiday_events["EventID"] = np.arange(
    1,
    len(holiday_events) + 1
)

event_source = eda_df[
    [
        "Store",
        "Date",
        "Sales",
        "Customers",
        "Open",
        "Promo",
        "SchoolHoliday"
    ]
].copy()

holiday_windows = []

for relative_day in range(-7, 8):

    lookup = holiday_events.copy()

    lookup["Date"] = (
        lookup["HolidayDate"]
        + pd.to_timedelta(relative_day, unit="D")
    )

    matched = lookup.merge(
        event_source,
        on=["Store", "Date"],
        how="inner"
    )

    matched["RelativeDay"] = relative_day

    holiday_windows.append(matched)

holiday_window = pd.concat(
    holiday_windows,
    ignore_index=True
)

holiday_window["Period"] = pd.cut(
    holiday_window["RelativeDay"],
    bins=[-np.inf, -1, 0, np.inf],
    labels=[
        "Before Holiday",
        "Holiday",
        "After Holiday"
    ],
    include_lowest=True
)

print("HOLIDAY EVENT WINDOW CREATED SUCCESSFULLY")
print("-" * 48)

print(
    f"Holiday events identified : "
    f"{len(holiday_events):,}"
)

print(
    f"Event-window observations : "
    f"{len(holiday_window):,}"
)

print(
    f"Relative-day coverage     : "
    f"{holiday_window['RelativeDay'].min()} "
    f"to "
    f"{holiday_window['RelativeDay'].max()}"
)

print("\nHoliday types:")
display(
    holiday_events["HolidayType"]
    .value_counts()
    .rename_axis("Holiday Type")
    .to_frame("Events")
)

print("\nEvent periods:")
display(
    holiday_window["Period"]
    .value_counts(sort=False)
    .rename_axis("Period")
    .to_frame("Observations")
)

logger.info(
    "Holiday event-window dataset created successfully | "
    "Events=%s | Observations=%s",
    len(holiday_events),
    len(holiday_window)
)
```

```python
period_order = [
    "Before Holiday",
    "Holiday",
    "After Holiday"
]

holiday_period_summary = []

for period in period_order:

    part = holiday_window[
        holiday_window["Period"].astype(str).eq(period)
    ].copy()

    open_part = part[
        part["Open"].eq(1)
    ]

    total_customers = open_part["Customers"].sum()

    holiday_period_summary.append({

        "Period": period,

        "Store-Days":
            len(part),

        "Open Rate %":
            part["Open"].mean() * 100,

        "Average Sales":
            part["Sales"].mean(),

        "Median Sales":
            part["Sales"].median(),

        "Average Customers":
            part["Customers"].mean(),

        "Median Customers":
            part["Customers"].median(),

        "Average Open-Store Sales":
            open_part["Sales"].mean(),

        "Average Open-Store Customers":
            open_part["Customers"].mean(),

        "Sales per Customer":
            (
                open_part["Sales"].sum()
                / total_customers
                if total_customers > 0
                else np.nan
            ),

        "Promo Rate %":
            part["Promo"].mean() * 100
    })


holiday_period_summary = pd.DataFrame(
    holiday_period_summary
)


display(
    holiday_period_summary.style

    .format({
        "Store-Days": "{:,.0f}",
        "Open Rate %": "{:.2f}%",
        "Average Sales": "{:,.2f}",
        "Median Sales": "{:,.2f}",
        "Average Customers": "{:,.2f}",
        "Median Customers": "{:,.2f}",
        "Average Open-Store Sales": "{:,.2f}",
        "Average Open-Store Customers": "{:,.2f}",
        "Sales per Customer": "{:,.2f}",
        "Promo Rate %": "{:.2f}%"
    })

    .set_properties(**{
        "text-align": "center"
    })

    .set_table_styles([

        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold"),
                ("text-align", "center"),
                ("padding", "10px")
            ]
        },

        {
            "selector": "td",
            "props": [
                ("padding", "9px"),
                ("border-bottom", "1px solid #e5e1d8")
            ]
        }

    ])
)


logger.info(
    "Holiday purchasing summary completed for before, during and after periods."
)
```

<h2><span style="color:lightblue">Holiday Event Study<b></b></span></h2>

 
How Does Customer Demand Change Around a Holiday?



To understand whether customers shift their purchases around major holidays,
sales and customer traffic are examined across the seven days preceding the
holiday, the holiday itself, and the seven days immediately afterwards.

Both measures are converted into indices using the average activity during the
seven pre-holiday days as a baseline of <strong>100</strong>. Values above 100
indicate stronger-than-baseline demand, while values below 100 indicate weaker
activity.


```python
holiday_daily = (
    holiday_window
    .groupby("RelativeDay", as_index=False)
    .agg(
        AverageSales=("Sales", "mean"),
        MedianSales=("Sales", "median"),
        AverageCustomers=("Customers", "mean"),
        OpenRate=("Open", "mean"),
        PromoRate=("Promo", "mean")
    )
    .sort_values("RelativeDay")
)

baseline = (
    holiday_daily.loc[
        holiday_daily["RelativeDay"].between(-7, -1),
        ["AverageSales", "AverageCustomers"]
    ]
    .mean()
)

holiday_daily["SalesIndex"] = (
    holiday_daily["AverageSales"]
    / baseline["AverageSales"]
    * 100
)

holiday_daily["CustomerIndex"] = (
    holiday_daily["AverageCustomers"]
    / baseline["AverageCustomers"]
    * 100
)

holiday_daily["OpenRate"] *= 100
holiday_daily["PromoRate"] *= 100

display(
    holiday_daily.style
    .format({
        "AverageSales": "{:,.2f}",
        "MedianSales": "{:,.2f}",
        "AverageCustomers": "{:,.2f}",
        "OpenRate": "{:.2f}%",
        "PromoRate": "{:.2f}%",
        "SalesIndex": "{:.2f}",
        "CustomerIndex": "{:.2f}"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold"),
                ("text-align", "center")
            ]
        }
    ])
)

logger.info(
    "Holiday relative-day event-study metrics calculated."
)
```

```python
fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    holiday_daily["RelativeDay"],
    holiday_daily["SalesIndex"],
    marker="o",
    markersize=6,
    linewidth=2.4,
    # color=NAVY,
    label="Sales Index"
)

ax.plot(
    holiday_daily["RelativeDay"],
    holiday_daily["CustomerIndex"],
    marker="o",
    markersize=6,
    linewidth=2.2,
    # color=GOLD,
    label="Customer Index"
)

ax.axhline(
    100,
    # color=SLATE,
    linestyle="--",
    linewidth=1.1,
    alpha=0.65
)

ax.axvline(
    0,
    # color=GOLD,
    linestyle="--",
    linewidth=1.6
)

ax.annotate(
    "Holiday",
    xy=(0, holiday_daily["SalesIndex"].max()),
    xytext=(0.4, holiday_daily["SalesIndex"].max()),
    fontsize=10,
    # color=SLATE
)

ax.set_title(
    "Sales and Customer Behaviour Around State Holidays",
    loc="left",
    fontsize=17,
    fontweight="bold",
    # color=NAVY,
    pad=16
)

ax.text(
    0,
    1.03,
    "Index based on average activity during Days −7 to −1 = 100",
    transform=ax.transAxes,
    fontsize=10.5,
    # color=SLATE
)

ax.set_xlabel("Days Relative to Holiday")
ax.set_ylabel("Demand Index")

ax.set_xticks(range(-7, 8))

ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="both", alpha=0.16)

ax.legend(
    frameon=False,
    ncol=2,
    loc="upper right"
)

plt.tight_layout()
plt.show()

logger.info(
    "Holiday sales and customer event-study chart generated."
)
```

<h2><span style="color:lightblue">Business Insight<b></b></span></h2>




Holiday Effects Reflect Both Customer Behaviour and Store Availability



The event-study framework shows how demand evolves around state holidays rather
than considering the holiday date in isolation. Changes in sales and customer
traffic before Day 0 reveal whether customers anticipate closures or seasonal
events by purchasing earlier.



The holiday-day result must be interpreted alongside the opening-rate analysis.
Where a large proportion of stores are closed, lower network-wide sales reflect
reduced store availability as well as customer behaviour. Open-store metrics
therefore provide a better indication of purchasing intensity among stores that
continue to operate.



From a planning perspective, any consistent pre-holiday acceleration would
support earlier inventory replenishment and staffing adjustments, while a
post-holiday decline would indicate the need to reduce short-term operational
capacity after peak demand passes.

Do Different Holiday Types Produce Different Purchasing Patterns?


The StateHoliday feature distinguishes between public holidays, Easter, and
Christmas. These events are analysed separately because their economic and
customer-behaviour effects may differ substantially. Treating all holiday types
as a single category could therefore conceal meaningful seasonal variation.


```python
holiday_type_summary = []

for holiday_type in [
    "Public Holiday",
    "Easter Holiday",
    "Christmas"
]:

    part = eda_df[
        eda_df["HolidayType"].eq(holiday_type)
    ]

    open_part = part[
        part["Open"].eq(1)
    ]

    total_customers = open_part["Customers"].sum()

    holiday_type_summary.append({

        "Holiday Type":
            holiday_type,

        "Store-Days":
            len(part),

        "Open Rate %":
            part["Open"].mean() * 100,

        "Average Sales":
            part["Sales"].mean(),

        "Median Sales":
            part["Sales"].median(),

        "Average Open-Store Sales":
            open_part["Sales"].mean(),

        "Average Open-Store Customers":
            open_part["Customers"].mean(),

        "Sales per Customer":
            (
                open_part["Sales"].sum()
                / total_customers
                if total_customers > 0
                else np.nan
            ),

        "Promo Rate %":
            part["Promo"].mean() * 100
    })


holiday_type_summary = pd.DataFrame(
    holiday_type_summary
)


display(
    holiday_type_summary.style
    .format({
        "Store-Days": "{:,.0f}",
        "Open Rate %": "{:.2f}%",
        "Average Sales": "{:,.2f}",
        "Median Sales": "{:,.2f}",
        "Average Open-Store Sales": "{:,.2f}",
        "Average Open-Store Customers": "{:,.2f}",
        "Sales per Customer": "{:.2f}",
        "Promo Rate %": "{:.2f}%"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                # ("background-color", NAVY),
                # ("color", WHITE),
                ("font-weight", "bold"),
                ("text-align", "center")
            ]
        }
    ])
)

logger.info(
    "Holiday-type purchasing performance analysed."
)
```

<h2><span style="color:lightblue">Seasonal Event Intelligence<b></b></span></h2>


 

Christmas Purchasing Behaviour



Christmas represents one of the most commercially important periods in the
retail calendar. Customer demand may accelerate before the holiday as households
bring purchases forward, while store closures can cause a sharp reduction in
sales on the holiday itself.



The following mini-dashboard evaluates Christmas through both a demand and an
operational perspective. It identifies the strongest demand day, compares sales
and customer movement around Christmas, measures store availability, and
examines whether purchasing intensity among operating stores changes before,
during, and after the event.




```python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

christmas = (
    holiday_window[
        holiday_window["HolidayType"].eq("Christmas")
    ]
    .copy()
)

christmas_daily = (
    christmas
    .groupby("RelativeDay", as_index=False)
    .agg(
        AverageSales=("Sales", "mean"),
        MedianSales=("Sales", "median"),
        AverageCustomers=("Customers", "mean"),
        OpenRate=("Open", "mean"),
        PromoRate=("Promo", "mean")
    )
    .sort_values("RelativeDay")
)

christmas_daily["OpenRate"] *= 100
christmas_daily["PromoRate"] *= 100

spc_daily = (
    christmas[
        christmas["Open"].eq(1)
        & christmas["Customers"].gt(0)
    ]
    .groupby("RelativeDay")
    .apply(
        lambda x:
        x["Sales"].sum()
        / x["Customers"].sum(),
        include_groups=False
    )
)

christmas_daily["SalesPerCustomer"] = (
    christmas_daily["RelativeDay"]
    .map(spc_daily)
)

baseline = (
    christmas_daily.loc[
        christmas_daily["RelativeDay"]
        .between(-7, -1),
        [
            "AverageSales",
            "AverageCustomers",
            "SalesPerCustomer"
        ]
    ]
    .mean()
)

christmas_daily["SalesIndex"] = (
    christmas_daily["AverageSales"]
    / baseline["AverageSales"]
    * 100
)

christmas_daily["CustomerIndex"] = (
    christmas_daily["AverageCustomers"]
    / baseline["AverageCustomers"]
    * 100
)

christmas_daily["SpendIndex"] = (
    christmas_daily["SalesPerCustomer"]
    / baseline["SalesPerCustomer"]
    * 100
)

period_order = [
    "Before Holiday",
    "Holiday",
    "After Holiday"
]

period_records = []

for period in period_order:

    part = christmas[
        christmas["Period"]
        .astype(str)
        .eq(period)
    ]

    open_part = part[
        part["Open"].eq(1)
    ]

    customers = open_part["Customers"].sum()

    period_records.append({

        "Period":
            period,

        "Store-Days":
            len(part),

        "Open Rate %":
            part["Open"].mean() * 100,

        "Promo Rate %":
            part["Promo"].mean() * 100,

        "Average Sales":
            part["Sales"].mean(),

        "Median Sales":
            part["Sales"].median(),

        "Open-Store Sales":
            open_part["Sales"].mean(),

        "Average Customers":
            part["Customers"].mean(),

        "Open-Store Customers":
            open_part["Customers"].mean(),

        "Sales per Customer":
            (
                open_part["Sales"].sum()
                / customers
                if customers > 0
                else np.nan
            )
    })


christmas_period = pd.DataFrame(
    period_records
)

christmas_period["Period"] = pd.Categorical(
    christmas_period["Period"],
    categories=period_order,
    ordered=True
)

christmas_period = (
    christmas_period
    .sort_values("Period")
    .reset_index(drop=True)
)

peak_sales_row = christmas_daily.loc[
    christmas_daily["SalesIndex"].idxmax()
]

peak_customer_row = christmas_daily.loc[
    christmas_daily["CustomerIndex"].idxmax()
]

peak_sales_day = int(
    peak_sales_row["RelativeDay"]
)

peak_customer_day = int(
    peak_customer_row["RelativeDay"]
)

peak_sales_index = float(
    peak_sales_row["SalesIndex"]
)

peak_customer_index = float(
    peak_customer_row["CustomerIndex"]
)


christmas_day = christmas[
    christmas["RelativeDay"].eq(0)
]

christmas_open = christmas_day[
    christmas_day["Open"].eq(1)
]

christmas_open_rate = (
    christmas_day["Open"].mean() * 100
)

christmas_promo_rate = (
    christmas_day["Promo"].mean() * 100
)

christmas_open_sales = (
    christmas_open["Sales"].mean()
)

christmas_open_customers = (
    christmas_open["Customers"].mean()
)

christmas_spc = (
    christmas_open["Sales"].sum()
    / christmas_open["Customers"].sum()
    if christmas_open["Customers"].sum() > 0
    else np.nan
)

christmas_events = (
    christmas["EventID"].nunique()
)

period_idx = (
    christmas_period
    .set_index("Period")
)

before = period_idx.loc[
    "Before Holiday"
]

during = period_idx.loc[
    "Holiday"
]

after = period_idx.loc[
    "After Holiday"
]

christmas_sales_change = (
    during["Open-Store Sales"]
    / before["Open-Store Sales"] - 1
) * 100

christmas_customer_change = (
    during["Open-Store Customers"]
    / before["Open-Store Customers"] - 1
) * 100

christmas_spc_change = (
    during["Sales per Customer"]
    / before["Sales per Customer"] - 1
) * 100

post_sales_change = (
    after["Open-Store Sales"]
    / before["Open-Store Sales"] - 1
) * 100

view_toggle = widgets.ToggleButtons(
    options=[
        ("📊 Charts", "Charts"),
        ("📋 Table", "Table")
    ],
    value="Charts",
    description="View:",
    button_style="",
    style={
        "button_width": "125px",
        "description_width": "55px"
    }
)

period_filter = widgets.Dropdown(
    options=[
        ("Full Christmas Window", "All"),
        ("Before Christmas", "Before Holiday"),
        ("Christmas Day", "Holiday"),
        ("After Christmas", "After Holiday")
    ],
    value="All",
    description="Period:",
    layout=widgets.Layout(
        width="270px"
    ),
    style={
        "description_width": "60px"
    }
)

metric_filter = widgets.Dropdown(
    options=[
        ("Sales Index", "SalesIndex"),
        ("Customer Index", "CustomerIndex"),
        ("Spend per Customer Index", "SpendIndex"),
        ("Store Open Rate", "OpenRate"),
        ("Promotion Rate", "PromoRate")
    ],
    value="SalesIndex",
    description="Metric:",
    layout=widgets.Layout(
        width="300px"
    ),
    style={
        "description_width": "60px"
    }
)

output = widgets.Output()

def day_text(day):

    if day == 0:
        return "Day 0"

    return f"Day {day:+d}"

def kpi_cards():

    cards = [

        (
            "CHRISTMAS EVENTS",
            f"{christmas_events:,}",
            "Store-level holiday events"
        ),

        (
            "PEAK SALES DAY",
            day_text(peak_sales_day),
            f"Index {peak_sales_index:.1f}"
        ),

        (
            "PEAK CUSTOMER DAY",
            day_text(peak_customer_day),
            f"Index {peak_customer_index:.1f}"
        ),

        (
            "CHRISTMAS OPEN RATE",
            f"{christmas_open_rate:.1f}%",
            "Stores operating on Day 0"
        ),

        (
            "OPEN-STORE SALES",
            f"{christmas_open_sales:,.0f}",
            "Average Christmas sales"
        ),

        (
            "AVG. CUSTOMERS",
            f"{christmas_open_customers:,.0f}",
            "Christmas open stores"
        ),

        (
            "SALES / CUSTOMER",
            f"{christmas_spc:,.2f}",
            f"{christmas_spc_change:+.1f}% vs pre-Christmas"
        )
    ]

    cards_html = "".join([

        f"""
        <div style="
            background:#ffffff;
            border:1px solid #ddd7cc;
            # border-top:4px solid {GOLD};
            border-radius:10px;
            padding:15px 15px 14px;
            min-height:103px;
            box-shadow:0 2px 7px rgba(12,36,57,0.06);
        ">

            <div style="
                # color:{SLATE};
                font-size:9px;
                font-weight:700;
                letter-spacing:1.2px;
                margin-bottom:8px;
            ">
                {title}
            </div>

            <div style="
                # color:{NAVY};
                font-family:Georgia,'Times New Roman',serif;
                font-size:21px;
                font-weight:600;
                margin-bottom:6px;
            ">
                {value}
            </div>

            <div style="
                color:#7c838a;
                font-size:10px;
                line-height:1.35;
            ">
                {subtitle}
            </div>

        </div>
        """

        for title, value, subtitle
        in cards
    ])

    display(
        HTML(
            f"""
            <div style="
                display:grid;
                grid-template-columns:
                    repeat(7,minmax(0,1fr));
                gap:10px;
                margin:5px 0 20px;
                font-family:'Segoe UI',Arial,sans-serif;
            ">
                {cards_html}
            </div>
            """
        )
    )

def show_charts():

    selected_metric = metric_filter.value

    metric_labels = {

        "SalesIndex":
            "Sales Index",

        "CustomerIndex":
            "Customer Index",

        "SpendIndex":
            "Sales per Customer Index",

        "OpenRate":
            "Store Open Rate (%)",

        "PromoRate":
            "Promotion Rate (%)"
    }

    metric_name = metric_labels[
        selected_metric
    ]

    chart_data = christmas_daily.copy()

    if period_filter.value == "Before Holiday":

        chart_data = chart_data[
            chart_data["RelativeDay"] < 0
        ]

    elif period_filter.value == "Holiday":

        chart_data = chart_data[
            chart_data["RelativeDay"] == 0
        ]

    elif period_filter.value == "After Holiday":

        chart_data = chart_data[
            chart_data["RelativeDay"] > 0
        ]

    fig = plt.figure(
        figsize=(16, 8),
        facecolor=WHITE
    )

    gs = fig.add_gridspec(
        2,
        2,
        height_ratios=[2.1, 1.45],
        hspace=0.48,
        wspace=0.30
    )

    ax1 = fig.add_subplot(
        gs[0, :]
    )

    ax1.vlines(
        chart_data["RelativeDay"],
        ymin=0,
        ymax=chart_data[selected_metric],
        color="#c9c5bb",
        linewidth=2
    )

    ax1.scatter(
        chart_data["RelativeDay"],
        chart_data[selected_metric],
        s=95,
        # color=NAVY,
        zorder=3
    )

    if selected_metric in [
        "SalesIndex",
        "CustomerIndex",
        "SpendIndex"
    ]:

        ax1.axhline(
            100,
            linestyle="--",
            linewidth=1.2,
            # color=GOLD,
            alpha=0.8
        )

    ax1.axvline(
        0,
        linestyle=":",
        linewidth=1.4,
        # color=GOLD
    )

    ax1.set_title(
        f"{metric_name} Across the Christmas Window",
        loc="left",
        fontsize=15,
        fontweight="bold",
        # color=NAVY,
        pad=13
    )

    ax1.text(
        0,
        1.025,
        "Select another metric above to change the behavioural perspective",
        transform=ax1.transAxes,
        fontsize=9,
        # color=SLATE
    )

    ax1.set_xlabel(
        "Days Relative to Christmas"
    )

    ax1.set_ylabel(
        metric_name
    )

    ax1.set_xticks(
        chart_data["RelativeDay"]
    )

    ax1.spines[
        ["top", "right"]
    ].set_visible(False)

    ax1.grid(
        axis="y",
        alpha=0.14
    )

    ax2 = fig.add_subplot(
        gs[1, 0]
    )

    comparison = pd.DataFrame({

        "Before": [
            100,
            100,
            100
        ],

        "Christmas": [

            during["Open-Store Sales"]
            / before["Open-Store Sales"]
            * 100,

            during["Open-Store Customers"]
            / before["Open-Store Customers"]
            * 100,

            during["Sales per Customer"]
            / before["Sales per Customer"]
            * 100
        ],

        "After": [

            after["Open-Store Sales"]
            / before["Open-Store Sales"]
            * 100,

            after["Open-Store Customers"]
            / before["Open-Store Customers"]
            * 100,

            after["Sales per Customer"]
            / before["Sales per Customer"]
            * 100
        ]
    },
    index=[
        "Sales",
        "Customers",
        "Spend / Customer"
    ])

    y = np.arange(
        len(comparison)
    )

    height = 0.23

    ax2.barh(
        y - height,
        comparison["Before"],
        height,
        # color=CREAM,
        label="Before"
    )

    ax2.barh(
        y,
        comparison["Christmas"],
        height,
        # color=GOLD,
        label="Christmas"
    )

    ax2.barh(
        y + height,
        comparison["After"],
        height,
        # color=SLATE,
        alpha=0.75,
        label="After"
    )

    ax2.axvline(
        100,
        linestyle="--",
        linewidth=1,
        # color=NAVY,
        alpha=0.45
    )

    ax2.set_yticks(y)

    ax2.set_yticklabels(
        comparison.index
    )

    ax2.set_title(
        "Open-Store Behaviour",
        loc="left",
        fontsize=13,
        fontweight="bold",
        # color=NAVY
    )

    ax2.set_xlabel(
        "Index — Before Christmas = 100"
    )

    ax2.spines[
        ["top", "right"]
    ].set_visible(False)

    ax2.grid(
        axis="x",
        alpha=0.12
    )

    ax2.legend(
        frameon=False,
        fontsize=8,
        ncol=3
    )

    ax3 = fig.add_subplot(
        gs[1, 1]
    )

    operational = pd.DataFrame({

        "Metric": [
            "Open Rate",
            "Promo Rate",
            "Sales Change",
            "Customer Change",
            "Spend Change"
        ],

        "Value": [
            christmas_open_rate,
            christmas_promo_rate,
            christmas_sales_change,
            christmas_customer_change,
            christmas_spc_change
        ]
    })

    bars = ax3.barh(
        operational["Metric"],
        operational["Value"],
        # color=[
            # NAVY,
            # GOLD,
            # SLATE,
            # LIGHT_GOLD,
            # CREAM
        # ],
        # edgecolor=NAVY,
        # linewidth=0.45
    )

    ax3.axvline(
        0,
        # color=SLATE,
        linewidth=0.8
    )

    for bar, value in zip(
        bars,
        operational["Value"]
    ):

        ax3.text(
            value
            + (
                1
                if value >= 0
                else -1
            ),
            bar.get_y()
            + bar.get_height() / 2,
            f"{value:+.1f}%",
            va="center",
            ha=(
                "left"
                if value >= 0
                else "right"
            ),
            fontsize=8.5,
            # color=NAVY
        )

    ax3.set_title(
        "Christmas Operational Signals",
        loc="left",
        fontsize=13,
        fontweight="bold",
        # color=NAVY
    )

    ax3.set_xlabel(
        "Percentage / Change (%)"
    )

    ax3.spines[
        ["top", "right"]
    ].set_visible(False)

    ax3.grid(
        axis="x",
        alpha=0.12
    )

    plt.tight_layout()
    plt.show()

def show_table():

    display(
        HTML(
            f"""
            <div style="
                padding:16px 20px;
                margin-bottom:15px;
                background:#edf3f6;
                # border-left:4px solid {NAVY};
                font-family:'Segoe UI',Arial,sans-serif;
                # color:{SLATE};
                line-height:1.6;
            ">
                # <strong style="color:{NAVY};">
                    Christmas Period Performance Table
                </strong><br>

                The table combines operating conditions,
                sales performance, customer traffic,
                promotional exposure and purchasing intensity.
            </div>
            """
        )
    )

    styled_table = (
        christmas_period.style

        .format({

            "Store-Days":
                "{:,.0f}",

            "Open Rate %":
                "{:.2f}%",

            "Promo Rate %":
                "{:.2f}%",

            "Average Sales":
                "{:,.2f}",

            "Median Sales":
                "{:,.2f}",

            "Open-Store Sales":
                "{:,.2f}",

            "Average Customers":
                "{:,.2f}",

            "Open-Store Customers":
                "{:,.2f}",

            "Sales per Customer":
                "{:.2f}"
        })

        .background_gradient(
            subset=[
                "Open-Store Sales"
            ],
            cmap=sns.light_palette(
                # GOLD,
                as_cmap=True
            )
        )

        .background_gradient(
            subset=[
                "Open-Store Customers"
            ],
            cmap=sns.light_palette(
                # NAVY,
                as_cmap=True
            )
        )

        .set_properties(**{
            "text-align": "center",
            "padding": "9px"
        })

        .set_table_styles([

            {
                "selector": "th",
                "props": [
                    (
                        "background-color",
                        # NAVY
                    ),
                    (
                        "color",
                        WHITE
                    ),
                    (
                        "font-weight",
                        "bold"
                    ),
                    (
                        "text-align",
                        "center"
                    )
                ]
            }
        ])
    )

    display(
        styled_table
    )

def show_insight():

    timing = (

        "before Christmas"
        if peak_sales_day < 0

        else
        "after Christmas"
        if peak_sales_day > 0

        else
        "on Christmas Day"
    )

    traffic_driver = (

        "customer traffic rises more strongly than spend per customer"

        if christmas_customer_change
        > christmas_spc_change

        else
        "purchasing value per customer contributes strongly to the sales movement"
    )

    display(
        HTML(
            f"""
            <div style="
                margin-top:18px;
                padding:18px 21px;
                background:#f2eee5;
                # border-left:4px solid {GOLD};
                border-radius:0 8px 8px 0;
                font-family:'Segoe UI',Arial,sans-serif;
            ">

                <div style="
                    font-size:10px;
                    color:#775d2e;
                    font-weight:700;
                    letter-spacing:1.7px;
                    margin-bottom:8px;
                ">
                    DYNAMIC BUSINESS INTERPRETATION
                </div>

                <div style="
                    color:#4c4538;
                    font-size:13.5px;
                    line-height:1.8;
                ">

                    Peak Christmas-related sales activity occurs
                    <strong>{timing}</strong>,
                    with the strongest observed sales index reaching
                    <strong>{peak_sales_index:.1f}</strong>.

                    Christmas-Day store availability is
                    <strong>{christmas_open_rate:.1f}%</strong>,
                    showing why network-wide holiday sales should not
                    be interpreted independently of operating status.

                    Among stores that remain open, sales change by
                    <strong>{christmas_sales_change:+.1f}%</strong>
                    relative to the pre-Christmas period, while
                    customer traffic changes by
                    <strong>{christmas_customer_change:+.1f}%</strong>
                    and sales per customer changes by
                    <strong>{christmas_spc_change:+.1f}%</strong>.

                    Overall, {traffic_driver}.

                    After Christmas, open-store sales are
                    <strong>{post_sales_change:+.1f}%</strong>
                    relative to the pre-holiday baseline.

                </div>

            </div>
            """
        )
    )

def render_dashboard(change=None):

    with output:

        clear_output(
            wait=True
        )

        kpi_cards()

        if view_toggle.value == "Charts":

            show_charts()

        else:

            show_table()

        show_insight()

view_toggle.observe(
    render_dashboard,
    names="value"
)

period_filter.observe(
    render_dashboard,
    names="value"
)

metric_filter.observe(
    render_dashboard,
    names="value"
)

control_bar = widgets.HBox(
    [
        view_toggle,
        period_filter,
        metric_filter
    ],
    layout=widgets.Layout(
        display="flex",
        flex_flow="row wrap",
        gap="12px",
        align_items="center",
        margin="5px 0 18px 0"
    )
)

display(
    HTML(
        f"""
        <div style="
            font-family:'Segoe UI',Arial,sans-serif;
            margin:5px 0 12px;
        ">

            font-family:Georgia,'Times New Roman',serif;
            font-size:22px;
            font-weight:600;
            margin-bottom:4px;
        ">
                Christmas Intelligence Dashboard
            </div>

        
            ">
                Switch between analytical charts and the detailed
                performance table. Use the metric and period controls
                to explore different dimensions of Christmas
                purchasing behaviour.
            </div>

        </div>
        """
    )
)

display(
    control_bar,
    output
)

render_dashboard()

logger.info(
    "Interactive Christmas intelligence dashboard initialized.")
```

    margin:0;
    color:#4c4538;
    font-size:13.5px;
    line-height:1.82;
">
Christmas-Day sales must be interpreted together with the store-opening rate.
A low network-wide sales value on the holiday itself may primarily reflect
store closures rather than weak consumer demand. For behavioural interpretation,
the open-store metrics therefore provide a more meaningful measure of actual
customer purchasing activity.

<h2><span style="color:lightblue">Easter Purchasing Behaviour<b></b></span></h2>

Seasonal Event Intelligence





Easter is investigated independently because its purchasing cycle may differ
from the Christmas period in terms of customer preparation, promotional
activity, store closures, and post-holiday recovery. Analysing Easter separately
helps determine whether seasonal effects should be represented differently in
the forecasting process.



The dashboard evaluates not only changes in average sales, but also customer
traffic, expenditure per customer, store availability, promotional exposure,
and the underlying distribution of store-level activity. This provides a more
complete assessment of whether Easter-related sales movements are caused by
higher traffic, stronger purchasing intensity, or operational availability.

<strong style="color:#102e89;">Dashboard design:</strong>
The Easter analysis introduces a waterfall view, a multidimensional polar
profile, and a distribution-based violin visual so that chart forms already
used in the Christmas analysis are not repeated.



```python
easter = (
    holiday_window[
        holiday_window["HolidayType"].eq("Easter Holiday")
    ]
    .copy()
)

easter_daily = (
    easter
    .groupby("RelativeDay", as_index=False)
    .agg(
        AverageSales=("Sales", "mean"),
        MedianSales=("Sales", "median"),
        AverageCustomers=("Customers", "mean"),
        MedianCustomers=("Customers", "median"),
        OpenRate=("Open", "mean"),
        PromoRate=("Promo", "mean")
    )
    .sort_values("RelativeDay")
)

easter_open_daily = (
    easter[
        easter["Open"].eq(1)
        & easter["Customers"].gt(0)
    ]
    .groupby("RelativeDay", as_index=False)
    .agg(
        SalesTotal=("Sales", "sum"),
        CustomerTotal=("Customers", "sum")
    )
)

easter_open_daily["SalesPerCustomer"] = (
    easter_open_daily["SalesTotal"]
    / easter_open_daily["CustomerTotal"]
)

easter_daily = easter_daily.merge(
    easter_open_daily[
        [
            "RelativeDay",
            "SalesPerCustomer"
        ]
    ],
    on="RelativeDay",
    how="left"
)

easter_daily["OpenRate"] *= 100
easter_daily["PromoRate"] *= 100

baseline = (
    easter_daily.loc[
        easter_daily["RelativeDay"].between(-7, -1),
        [
            "AverageSales",
            "AverageCustomers",
            "SalesPerCustomer"
        ]
    ]
    .mean()
)

easter_daily["SalesIndex"] = (
    easter_daily["AverageSales"]
    / baseline["AverageSales"]
    * 100
)

easter_daily["CustomerIndex"] = (
    easter_daily["AverageCustomers"]
    / baseline["AverageCustomers"]
    * 100
)

easter_daily["SpendIndex"] = (
    easter_daily["SalesPerCustomer"]
    / baseline["SalesPerCustomer"]
    * 100
)

period_order = [
    "Before Holiday",
    "Holiday",
    "After Holiday"
]

period_records = []

for period in period_order:

    part = easter[
        easter["Period"]
        .astype(str)
        .eq(period)
    ]

    open_part = part[
        part["Open"].eq(1)
    ]

    total_customers = (
        open_part["Customers"].sum()
    )

    period_records.append({

        "Period":
            period,

        "Store-Days":
            len(part),

        "Open Rate %":
            part["Open"].mean() * 100,

        "Promo Rate %":
            part["Promo"].mean() * 100,

        "Average Sales":
            part["Sales"].mean(),

        "Median Sales":
            part["Sales"].median(),

        "Open-Store Sales":
            open_part["Sales"].mean(),

        "Average Customers":
            part["Customers"].mean(),

        "Median Customers":
            part["Customers"].median(),

        "Open-Store Customers":
            open_part["Customers"].mean(),

        "Sales per Customer":
            (
                open_part["Sales"].sum()
                / total_customers
                if total_customers > 0
                else np.nan
            )
    })

easter_period = pd.DataFrame(
    period_records
)

easter_period["Period"] = pd.Categorical(
    easter_period["Period"],
    categories=period_order,
    ordered=True
)

easter_period = (
    easter_period
    .sort_values("Period")
    .reset_index(drop=True)
)

period_idx = (
    easter_period
    .set_index("Period")
)

before_easter = period_idx.loc[
    "Before Holiday"
]

during_easter = period_idx.loc[
    "Holiday"
]

after_easter = period_idx.loc[
    "After Holiday"
]

easter_sales_change = (
    during_easter["Open-Store Sales"]
    / before_easter["Open-Store Sales"]
    - 1
) * 100

easter_customer_change = (
    during_easter["Open-Store Customers"]
    / before_easter["Open-Store Customers"]
    - 1
) * 100

easter_spend_change = (
    during_easter["Sales per Customer"]
    / before_easter["Sales per Customer"]
    - 1
) * 100

post_easter_sales_change = (
    after_easter["Open-Store Sales"]
    / before_easter["Open-Store Sales"]
    - 1
) * 100

peak_sales_row = easter_daily.loc[
    easter_daily["SalesIndex"].idxmax()
]

peak_customer_row = easter_daily.loc[
    easter_daily["CustomerIndex"].idxmax()
]

peak_spend_row = easter_daily.loc[
    easter_daily["SpendIndex"].idxmax()
]

peak_sales_day = int(
    peak_sales_row["RelativeDay"]
)

peak_customer_day = int(
    peak_customer_row["RelativeDay"]
)

peak_spend_day = int(
    peak_spend_row["RelativeDay"]
)

peak_sales_index = float(
    peak_sales_row["SalesIndex"]
)

peak_customer_index = float(
    peak_customer_row["CustomerIndex"]
)

peak_spend_index = float(
    peak_spend_row["SpendIndex"]
)

easter_day = easter[
    easter["RelativeDay"].eq(0)
]

easter_day_open = easter_day[
    easter_day["Open"].eq(1)
]

easter_open_rate = (
    easter_day["Open"].mean() * 100
)

easter_promo_rate = (
    easter_day["Promo"].mean() * 100
)

easter_open_sales = (
    easter_day_open["Sales"].mean()
)

easter_open_customers = (
    easter_day_open["Customers"].mean()
)

easter_spc = (
    easter_day_open["Sales"].sum()
    / easter_day_open["Customers"].sum()
    if easter_day_open["Customers"].sum() > 0
    else np.nan
)

easter_events = (
    easter["EventID"].nunique()
)

easter_view = widgets.ToggleButtons(
    options=[
        ("📊 Charts", "Charts"),
        ("📋 Table", "Table")
    ],
    value="Charts",
    description="View:",
    style={
        "button_width": "125px",
        "description_width": "55px"
    }
)

easter_metric = widgets.Dropdown(
    options=[
        ("Sales", "Sales"),
        ("Customers", "Customers"),
        ("Sales per Customer", "SalesPerCustomer")
    ],
    value="Sales",
    description="Metric:",
    layout=widgets.Layout(
        width="270px"
    ),
    style={
        "description_width": "60px"
    }
)

easter_period_filter = widgets.Dropdown(
    options=[
        ("All Periods", "All"),
        ("Before Easter", "Before Holiday"),
        ("Easter Day", "Holiday"),
        ("After Easter", "After Holiday")
    ],
    value="All",
    description="Period:",
    layout=widgets.Layout(
        width="260px"
    ),
    style={
        "description_width": "60px"
    }
)

easter_table_level = widgets.ToggleButtons(
    options=[
        ("Period Summary", "Period"),
        ("Daily Detail", "Daily")
    ],
    value="Period",
    description="Table:",
    style={
        "button_width": "120px",
        "description_width": "55px"
    }
)

easter_output = widgets.Output()

def easter_day_text(day):

    if day == 0:
        return "Day 0"

    return f"Day {day:+d}"

def easter_kpis():

    cards = [

        (
            "EASTER EVENTS",
            f"{easter_events:,}",
            "Store-level events analysed"
        ),

        (
            "PEAK SALES DAY",
            easter_day_text(
                peak_sales_day
            ),
            f"Sales index {peak_sales_index:.1f}"
        ),

        (
            "PEAK CUSTOMER DAY",
            easter_day_text(
                peak_customer_day
            ),
            f"Traffic index {peak_customer_index:.1f}"
        ),

        (
            "PEAK SPEND DAY",
            easter_day_text(
                peak_spend_day
            ),
            f"Spend index {peak_spend_index:.1f}"
        ),

        (
            "EASTER OPEN RATE",
            f"{easter_open_rate:.1f}%",
            "Stores operating on Day 0"
        ),

        (
            "OPEN-STORE SALES",
            f"{easter_open_sales:,.0f}",
            f"{easter_sales_change:+.1f}% vs before"
        ),

        (
            "AVG. CUSTOMERS",
            f"{easter_open_customers:,.0f}",
            f"{easter_customer_change:+.1f}% vs before"
        ),

        (
            "SALES / CUSTOMER",
            f"{easter_spc:,.2f}",
            f"{easter_spend_change:+.1f}% vs before"
        )
    ]

    cards_html = "".join([

        f"""
        <div style="
            background:#ffffff;
            border:1px solid #ddd7cc;
            # border-top:4px solid {GOLD};
            border-radius:10px;
            padding:14px 14px 13px;
            min-height:102px;
            box-shadow:0 2px 7px rgba(12,36,57,0.06);
        ">

            <div style="
                color:{SLATE};
                font-size:8.5px;
                font-weight:700;
                letter-spacing:1.1px;
                margin-bottom:7px;
            ">
                {title}
            </div>

            <div style="
                # color:{NAVY};
                font-family:Georgia,'Times New Roman',serif;
                font-size:20px;
                font-weight:600;
                margin-bottom:5px;
            ">
                {value}
            </div>

            <div style="
                color:#7c838a;
                font-size:9.5px;
                line-height:1.35;
            ">
                {subtitle}
            </div>

        </div>
        """

        for title, value, subtitle
        in cards
    ])

    display(
        HTML(
            f"""
            <div style="
                display:grid;
                grid-template-columns:
                    repeat(4,minmax(0,1fr));
                gap:11px;
                margin:5px 0 20px;
                font-family:'Segoe UI',Arial,sans-serif;
            ">
                {cards_html}
            </div>
            """
        )
    )

def easter_waterfall(ax):

    selected = easter_metric.value

    metric_map = {

        "Sales": "Open-Store Sales",

        "Customers":
            "Open-Store Customers",

        "SalesPerCustomer":
            "Sales per Customer"
    }

    column = metric_map[
        selected
    ]

    before_value = (
        before_easter[column]
    )

    holiday_index = (
        during_easter[column]
        / before_value
        * 100
    )

    after_index = (
        after_easter[column]
        / before_value
        * 100
    )

    values = [
        100,
        holiday_index - 100,
        after_index - holiday_index
    ]

    labels = [
        "Before\nBaseline",
        "Easter\nChange",
        "Post-Easter\nChange"
    ]

    starts = [
        0,
        100,
        holiday_index
    ]

    colors = [
        NAVY,
        GOLD
        if values[1] >= 0
        else SLATE,
        LIGHT_GOLD
        if values[2] >= 0
        else CREAM
    ]

    for i in range(3):

        ax.bar(
            i,
            values[i],
            bottom=starts[i],
            width=0.58,
            color=colors[i],
            # edgecolor=NAVY,
            linewidth=0.6
        )

    final_levels = [
        100,
        holiday_index,
        after_index
    ]

    for i, level in enumerate(
        final_levels
    ):

        ax.text(
            i,
            level + 3,
            f"{level:.1f}",
            ha="center",
            fontsize=9,
            fontweight="bold",
            # color=NAVY
        )

    ax.plot(
        [0.29, 0.71],
        [100, 100],
        linestyle="--",
        color=SLATE,
        linewidth=0.8
    )

    ax.plot(
        [1.29, 1.71],
        [
            holiday_index,
            holiday_index
        ],
        linestyle="--",
        color=SLATE,
        linewidth=0.8
    )

    ax.set_xticks(
        range(3)
    )

    ax.set_xticklabels(
        labels
    )

    ax.set_title(
        f"{selected.replace('SalesPerCustomer', 'Spend per Customer')} Transition",
        loc="left",
        fontsize=14,
        fontweight="bold",
        # color=NAVY,
        pad=12
    )

    ax.text(
        0,
        1.02,
        "Waterfall view — pre-Easter level indexed to 100",
        transform=ax.transAxes,
        fontsize=8.8,
        color=SLATE
    )

    ax.set_ylabel(
        "Performance Index"
    )

    ax.spines[
        ["top", "right"]
    ].set_visible(False)

    ax.grid(
        axis="y",
        alpha=0.12
    )

def easter_polar(ax):

    metrics = [
        "Sales",
        "Customers",
        "Spend",
        "Open Rate",
        "Promo Rate"
    ]

    base_values = np.array([

        before_easter[
            "Open-Store Sales"
        ],

        before_easter[
            "Open-Store Customers"
        ],

        before_easter[
            "Sales per Customer"
        ],

        before_easter[
            "Open Rate %"
        ],

        before_easter[
            "Promo Rate %"
        ]

    ], dtype=float)

    period_values = {

        "Before": np.array([

            before_easter[
                "Open-Store Sales"
            ],

            before_easter[
                "Open-Store Customers"
            ],

            before_easter[
                "Sales per Customer"
            ],

            before_easter[
                "Open Rate %"
            ],

            before_easter[
                "Promo Rate %"
            ]

        ], dtype=float),

        "Easter": np.array([

            during_easter[
                "Open-Store Sales"
            ],

            during_easter[
                "Open-Store Customers"
            ],

            during_easter[
                "Sales per Customer"
            ],

            during_easter[
                "Open Rate %"
            ],

            during_easter[
                "Promo Rate %"
            ]

        ], dtype=float),

        "After": np.array([

            after_easter[
                "Open-Store Sales"
            ],

            after_easter[
                "Open-Store Customers"
            ],

            after_easter[
                "Sales per Customer"
            ],

            after_easter[
                "Open Rate %"
            ],

            after_easter[
                "Promo Rate %"
            ]

        ], dtype=float)

    }

    base_values = np.where(
        base_values == 0,
        np.nan,
        base_values
    )

    angles = np.linspace(
        0,
        2 * np.pi,
        len(metrics),
        endpoint=False
    )

    angles_closed = np.r_[
        angles,
        angles[0]
    ]

    radar_colors = {
        "Before": NAVY,
        "Easter": GOLD,
        "After": SLATE
    }

    for label, values in period_values.items():

        indexed = (
            values
            / base_values
            * 100
        )

        indexed = np.nan_to_num(
            indexed,
            nan=0,
            posinf=0,
            neginf=0
        )

        closed = np.r_[
            indexed,
            indexed[0]
        ]

        ax.plot(
            angles_closed,
            closed,
            linewidth=2,
            label=label,
            color=radar_colors[label]
        )

        ax.fill(
            angles_closed,
            closed,
            alpha=0.07,
            color=radar_colors[label]
        )

    ax.set_xticks(
        angles
    )

    ax.set_xticklabels(
        metrics,
        fontsize=8.5
    )

    ax.set_title(
        "Multidimensional Easter Profile",
        fontsize=14,
        fontweight="bold",
        # color=NAVY,
        pad=18
    )

    ax.set_rlabel_position(
        25
    )

    ax.tick_params(
        axis="y",
        labelsize=7
    )

    ax.grid(
        alpha=0.20
    )

    ax.legend(
        loc="upper right",
        bbox_to_anchor=(1.28, 1.13),
        frameon=False,
        fontsize=8
    )

def easter_distribution(ax):

    selected = (
        easter_metric.value
    )

    variable_map = {

        "Sales":
            "Sales",

        "Customers":
            "Customers",

        "SalesPerCustomer":
            "SalesPerCustomer"
    }

    label_map = {

        "Sales":
            "Daily Sales",

        "Customers":
            "Customer Traffic",

        "SalesPerCustomer":
            "Sales per Customer"
    }

    plot_data = (
        easter[
            easter["Open"].eq(1)
        ]
        .copy()
    )

    plot_data["PeriodLabel"] = (
        plot_data["Period"]
        .astype(str)
        .map({
            "Before Holiday": "Before",
            "Holiday": "Easter",
            "After Holiday": "After"
        })
    )

    if selected == "SalesPerCustomer":

        plot_data["SalesPerCustomer"] = np.where(
            plot_data["Customers"] > 0,
            plot_data["Sales"]
            / plot_data["Customers"],
            np.nan
        )

    if (
        easter_period_filter.value
        != "All"
    ):

        selected_period = {
            "Before Holiday": "Before",
            "Holiday": "Easter",
            "After Holiday": "After"
        }[
            easter_period_filter.value
        ]

        plot_data = plot_data[
            plot_data[
                "PeriodLabel"
            ].eq(
                selected_period
            )]

    sns.violinplot(
        data=plot_data,
        x="PeriodLabel",
        y=variable_map[selected],
        order=[
            p for p in [
                "Before",
                "Easter",
                "After"
            ]
            if p in plot_data[
                "PeriodLabel"
            ].dropna().unique()
        ],
        inner="quartile",
        cut=0,
        # color=LIGHT_GOLD,
        linewidth=0.8,
        ax=ax
    )

    ax.set_title(
        f"Store-Level {label_map[selected]} Distribution",
        loc="left",
        fontsize=14,
        fontweight="bold",
        # color=NAVY,
        pad=12
    )

    ax.text(
        0,
        1.02,
        "Distribution among operating stores",
        transform=ax.transAxes,
        fontsize=8.8,
        color=SLATE
    )

    ax.set_xlabel("")

    ax.set_ylabel(
        label_map[selected]
    )

    ax.spines[
        ["top", "right"]
    ].set_visible(False)

    ax.grid(
        axis="y",
        alpha=0.12
    )

def show_easter_charts():

    fig = plt.figure(
        figsize=(16, 10),
        facecolor=WHITE
    )

    gs = fig.add_gridspec(
        2,
        2,
        height_ratios=[
            1,
            1.05
        ],
        hspace=0.42,
        wspace=0.30
    )

    ax1 = fig.add_subplot(
        gs[0, 0]
    )

    easter_waterfall(
        ax1
    )

    ax2 = fig.add_subplot(
        gs[0, 1],
        projection="polar"
    )

    easter_polar(
        ax2
    )

    ax3 = fig.add_subplot(
        gs[1, :]
    )

    easter_distribution(
        ax3
    )

    plt.tight_layout()
    plt.show()

def show_easter_table():

    if (
        easter_table_level.value
        == "Period"
    ):

        display(
            HTML(
                f"""
                <div style="
                    padding:15px 19px;
                    margin-bottom:14px;
                    background:#edf3f6;
                    # border-left:4px solid {NAVY};
                    font-family:'Segoe UI',Arial,sans-serif;
                    color:{SLATE};
                    line-height:1.6;
                ">

                    # <strong style="color:{NAVY};">
                        Easter Period Performance
                    </strong><br>

                    Comparison of operational availability,
                    customer traffic, sales performance and
                    purchasing intensity before, during and
                    after Easter.

                </div>
                """
            )
        )

        display(

            easter_period.style

            .format({

                "Store-Days":
                    "{:,.0f}",

                "Open Rate %":
                    "{:.2f}%",

                "Promo Rate %":
                    "{:.2f}%",

                "Average Sales":
                    "{:,.2f}",

                "Median Sales":
                    "{:,.2f}",

                "Open-Store Sales":
                    "{:,.2f}",

                "Average Customers":
                    "{:,.2f}",

                "Median Customers":
                    "{:,.2f}",

                "Open-Store Customers":
                    "{:,.2f}",

                "Sales per Customer":
                    "{:.2f}"
            })

            .background_gradient(
                subset=[
                    "Open-Store Sales"
                ],
                cmap=sns.light_palette(
                    # GOLD,
                    as_cmap=True
                )
            )

            .set_properties(**{
                "text-align": "center",
                "padding": "9px"
            })

            .set_table_styles([
                {
                    "selector": "th",
                    "props": [
                        (
                            "background-color",
                            # NAVY
                        ),
                        (
                            "color",
                            WHITE
                        ),
                        (
                            "font-weight",
                            "bold"
                        ),
                        (
                            "text-align",
                            "center"
                        )
                    ]
                }
            ])
        )

    else:

        daily_table = (
            easter_daily[
                [
                    "RelativeDay",
                    "AverageSales",
                    "MedianSales",
                    "AverageCustomers",
                    "MedianCustomers",
                    "SalesPerCustomer",
                    "OpenRate",
                    "PromoRate",
                    "SalesIndex",
                    "CustomerIndex",
                    "SpendIndex"
                ]
            ]
            .copy()
        )

        display(
            HTML(
                f"""
                <div style="
                    padding:15px 19px;
                    margin-bottom:14px;
                    background:#edf3f6;
                    # border-left:4px solid {NAVY};
                    font-family:'Segoe UI',Arial,sans-serif;
                    color:{SLATE};
                    line-height:1.6;
                ">

                    # <strong style="color:{NAVY};">
                        Daily Easter Event Window
                    </strong><br>

                    Detailed Day −7 to Day +7 performance,
                    including demand indices, customer behaviour,
                    operating rates and promotional exposure.

                </div>
                """
            )
        )

        display(

            daily_table.style

            .format({

                "RelativeDay":
                    "{:+.0f}",

                "AverageSales":
                    "{:,.2f}",

                "MedianSales":
                    "{:,.2f}",

                "AverageCustomers":
                    "{:,.2f}",

                "MedianCustomers":
                    "{:,.2f}",

                "SalesPerCustomer":
                    "{:.2f}",

                "OpenRate":
                    "{:.2f}%",

                "PromoRate":
                    "{:.2f}%",

                "SalesIndex":
                    "{:.1f}",

                "CustomerIndex":
                    "{:.1f}",

                "SpendIndex":
                    "{:.1f}"
            })

            .background_gradient(
                subset=[
                    "SalesIndex"
                ],
                cmap=sns.light_palette(
                    # GOLD,
                    as_cmap=True
                )
            )

            .set_properties(**{
                "text-align": "center",
                "padding": "8px"
            })

            .set_table_styles([
                {
                    "selector": "th",
                    "props": [
                        (
                            "background-color",
                            # NAVY
                        ),
                        (
                            "color",
                            WHITE
                        ),
                        (
                            "font-weight",
                            "bold"
                        )
                    ]
                }
            ])
        )

def easter_insight():

    if peak_sales_day < 0:

        timing_text = (
            "before Easter"
        )

    elif peak_sales_day > 0:

        timing_text = (
            "after Easter"
        )

    else:

        timing_text = (
            "on Easter Day"
        )

    if abs(
        easter_customer_change
    ) > abs(
        easter_spend_change
    ):

        primary_driver = (
            "customer traffic represents the stronger "
            "behavioural movement"
        )

    else:

        primary_driver = (
            "changes in spending per customer represent "
            "the stronger behavioural movement"
        )

    if post_easter_sales_change < 0:

        recovery_text = (
            "sales fall below the pre-Easter level "
            "after the event"
        )

    elif post_easter_sales_change > 0:

        recovery_text = (
            "sales remain above the pre-Easter level "
            "after the event"
        )

    else:

        recovery_text = (
            "sales return approximately to the "
            "pre-Easter level"
        )

    display(
        HTML(
            f"""
            <div style="
                margin-top:18px;
                padding:19px 22px;
                background:#f2eee5;
                # border-left:4px solid {GOLD};
                border-radius:0 8px 8px 0;
                font-family:'Segoe UI',Arial,sans-serif;
            ">

            <div style="
                font-size:10px;
                color:#775d2e;
                font-weight:700;
                letter-spacing:1.7px;
                margin-bottom:8px;
            ">
                DYNAMIC EASTER BUSINESS INTERPRETATION
            </div>

            <div style="
                color:#4c4538;
                font-size:13.5px;
                line-height:1.85;
            ">

                Peak sales activity occurs
                <strong>{timing_text}</strong> on
                <strong>{easter_day_text(peak_sales_day)}</strong>,
                reaching an indexed level of
                <strong>{peak_sales_index:.1f}</strong>
                relative to the pre-Easter baseline.

                On Easter Day,
                <strong>{easter_open_rate:.1f}%</strong>
                of observed stores are operating and promotional
                exposure is
                <strong>{easter_promo_rate:.1f}%</strong>.

                Among open stores, sales change by
                <strong>{easter_sales_change:+.1f}%</strong>,
                customer traffic by
                <strong>{easter_customer_change:+.1f}%</strong>,
                and sales per customer by
                <strong>{easter_spend_change:+.1f}%</strong>
                relative to the pre-Easter period.

                This indicates that
                <strong>{primary_driver}</strong>.

                In the post-Easter period,
                <strong>{recovery_text}</strong>,
                with open-store sales changing by
                <strong>{post_easter_sales_change:+.1f}%</strong>
                relative to the pre-event benchmark.

            </div>

            </div>
            """
        )
    )

def render_easter_dashboard(change=None):

    with easter_output:

        clear_output(
            wait=True
        )

        easter_kpis()

        if (
            easter_view.value
            == "Charts"
        ):

            show_easter_charts()

        else:

            show_easter_table()

        easter_insight()

easter_view.observe(
    render_easter_dashboard,
    names="value"
)

easter_metric.observe(
    render_easter_dashboard,
    names="value"
)

easter_period_filter.observe(
    render_easter_dashboard,
    names="value"
)

easter_table_level.observe(
    render_easter_dashboard,
    names="value"
)

easter_controls = widgets.HBox(

    [
        easter_view,
        easter_metric,
        easter_period_filter,
        easter_table_level
    ],

    layout=widgets.Layout(
        display="flex",
        flex_flow="row wrap",
        gap="12px",
        align_items="center",
        margin="5px 0 18px 0"
    )
)

display(
    HTML(
        f"""
        <div style="
            font-family:'Segoe UI',Arial,sans-serif;
            margin:5px 0 13px;
        ">

        # <div style="
        # color:{NAVY};
        #         font-family:Georgia,'Times New Roman',serif;
        #         font-size:22px;
        #         font-weight:600;
        #         margin-bottom:4px;
        #     ">
                Easter Intelligence Dashboard
            </div>

            <div style="
                color:{SLATE};
                font-size:12px;
                line-height:1.6;
            ">
                Explore demand movement, store-level distributions,
                operational availability and purchasing intensity.
                Switch between visual analysis and detailed numerical
                evidence using the controls below.
            </div>

        </div>
        """
    ))

display(
    easter_controls,
    easter_output)

render_easter_dashboard()

logger.info(
    "Interactive Easter intelligence dashboard initialized.")
```

<h2><span style="color:lightblue">Easter Intelligence · Interpretation<b></b></span></h2>
    


Business Meaning of the Easter Demand Cycle



The Easter dashboard separates changes in overall retail activity from changes
in behaviour among stores that remain operational. This distinction is
important because reduced holiday sales can arise from store closures even when
customer demand at operating stores remains strong.



DEMAND TIMING

The peak relative day identifies when Rossmann should expect the greatest
Easter-related pressure on inventory and store capacity rather than assuming
that the holiday itself represents the commercial peak.

DEMAND DRIVER

Comparing customer traffic with sales per customer distinguishes growth caused
by additional store visits from growth caused by customers spending more during
each visit.



POST-EVENT RECOVERY



The after-Easter period indicates whether purchases were merely shifted forward
or whether elevated demand persists after the holiday, which is important for
post-event inventory and staffing decisions.




MODELLING IMPLICATION



If Easter produces a systematic demand pattern across the surrounding days,
forecasting should not rely only on a binary Easter indicator. Relative timing,
store-opening behaviour, promotional activity, and related calendar effects may
provide more informative signals for predicting seasonal sales.




<h2><span style="color:lightblue">School Holiday Intelligence<b></b></span></


Does School Holiday Status Influence Customer Purchasing Behaviour?

School holidays may alter household routines, travel behaviour, customer
traffic, and shopping frequency even when stores remain operational. To isolate
this relationship more clearly, the analysis focuses on <strong>open stores</strong>
and excludes state-holiday observations, allowing school-holiday behaviour to be
compared against otherwise normal trading days.

```python
NAVY = "#17324D"
GOLD = "#C49A3A"
SLATE = "#64748B"
WHITE = "#FFFFFF"

school = eda_df[
    eda_df["Open"].eq(1)
    & eda_df["StateHoliday"].eq("0")
].copy()

school["HolidayStatus"] = school["SchoolHoliday"].map({
    0: "No School Holiday",
    1: "School Holiday"
})

school["SalesPerCustomer"] = np.where(
    school["Customers"] > 0,
    school["Sales"] / school["Customers"],
    np.nan
)

view = widgets.ToggleButtons(
    options=[("📊 Charts", "Charts"), ("📋 Table", "Table")],
    value="Charts"
)

store_filter = widgets.Dropdown(
    options=["All"] + sorted(school["StoreType"].dropna().unique().tolist()),
    value="All",
    description="Store:"
)

metric = widgets.Dropdown(
    options={
        "Sales": "Sales",
        "Customers": "Customers",
        "Sales / Customer": "SalesPerCustomer"
    },
    value="Sales",
    description="Metric:"
)

out = widgets.Output()


def school_dashboard(*_):

    with out:
        clear_output(wait=True)

        d = school.copy()

        if store_filter.value != "All":
            d = d[d["StoreType"].eq(store_filter.value)]

        summary = (
            d.groupby("HolidayStatus")
            .agg(
                StoreDays=("Sales", "size"),
                AvgSales=("Sales", "mean"),
                MedianSales=("Sales", "median"),
                AvgCustomers=("Customers", "mean"),
                SalesPerCustomer=("SalesPerCustomer", "mean"),
                PromoRate=("Promo", "mean")
            )
        )

        if not {"No School Holiday", "School Holiday"}.issubset(summary.index):
            print("Insufficient observations for the selected filter.")
            return

        normal = summary.loc["No School Holiday"]
        holiday = summary.loc["School Holiday"]

        sales_change = (holiday["AvgSales"] / normal["AvgSales"] - 1) * 100
        customer_change = (holiday["AvgCustomers"] / normal["AvgCustomers"] - 1) * 100
        spend_change = (
            holiday["SalesPerCustomer"] / normal["SalesPerCustomer"] - 1
        ) * 100

        cards = [
            ("SCHOOL-HOLIDAY DAYS", f"{holiday['StoreDays']:,.0f}"),
            ("AVG. SALES", f"{holiday['AvgSales']:,.0f}"),
            ("SALES CHANGE", f"{sales_change:+.1f}%"),
            ("AVG. CUSTOMERS", f"{holiday['AvgCustomers']:,.0f}"),
            ("CUSTOMER CHANGE", f"{customer_change:+.1f}%"),
            ("SPEND / CUSTOMER", f"{holiday['SalesPerCustomer']:.2f}")
        ]

        display(HTML(
            "<div style='display:grid;grid-template-columns:repeat(6,1fr);"
            "gap:10px;margin:8px 0 20px;'>"
            + "".join(
                f"""
                <div style="padding:14px;background:#fff;border:1px solid #ddd7cc;
                border-top:4px solid {GOLD};border-radius:9px;text-align:center;">
                    <div style="font-size:8px;color:{SLATE};font-weight:700;">
                        {a}
                    </div>
                    <div style="font-size:20px;color:{NAVY};font-family:Georgia;
                    margin-top:7px;">
                        {b}
                    </div>
                </div>
                """
                for a, b in cards
            )
            + "</div>"
        ))

        if view.value == "Table":

            table = summary.reset_index()

            table["PromoRate"] *= 100

            display(
                table.style
                .format({
                    "StoreDays": "{:,.0f}",
                    "AvgSales": "{:,.2f}",
                    "MedianSales": "{:,.2f}",
                    "AvgCustomers": "{:,.2f}",
                    "SalesPerCustomer": "{:.2f}",
                    "PromoRate": "{:.2f}%"
                })
                .set_table_styles([{
                    "selector": "th",
                    "props": [
                        ("background-color", NAVY),
                        ("color", WHITE),
                        ("font-weight", "bold")
                    ]
                }])
            )

        else:

            # ----------------------------------------------
            # CHART 1 — SLOPEGRAPH
            # ----------------------------------------------

            metrics = {
                "Sales": (normal["AvgSales"], holiday["AvgSales"]),
                "Customers": (
                    normal["AvgCustomers"],
                    holiday["AvgCustomers"]
                ),
                "Spend / Customer": (
                    normal["SalesPerCustomer"],
                    holiday["SalesPerCustomer"]
                )
            }

            indexed = {
                k: [100, v[1] / v[0] * 100]
                for k, v in metrics.items()
            }

            fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

            for name, values in indexed.items():
                axes[0].plot(
                    [0, 1],
                    values,
                    marker="o",
                    linewidth=2,
                    label=name
                )

                axes[0].text(
                    1.03,
                    values[1],
                    f"{name}  {values[1]:.1f}",
                    va="center",
                    fontsize=9
                )

            axes[0].set_xticks(
                [0, 1],
                ["No School Holiday", "School Holiday"]
            )

            axes[0].axhline(
                100,
                linestyle="--",
                linewidth=1,
                color=SLATE,
                alpha=.5
            )

            axes[0].set_title(
                "Behavioural Shift During School Holidays",
                loc="left",
                fontweight="bold",
                color=NAVY
            )

            axes[0].set_ylabel("Index: Normal Days = 100")
            axes[0].spines[["top", "right"]].set_visible(False)
            axes[0].grid(axis="y", alpha=.15)


            # ----------------------------------------------
            # CHART 2 — ECDF
            # ----------------------------------------------

            variable = metric.value

            for status in [
                "No School Holiday",
                "School Holiday"
            ]:

                values = (
                    d.loc[d["HolidayStatus"].eq(status), variable]
                    .dropna()
                    .sort_values()
                    .values
                )

                y = np.arange(1, len(values) + 1) / len(values)

                axes[1].plot(
                    values,
                    y,
                    linewidth=2,
                    label=status
                )

            axes[1].set_title(
                f"Distribution of {metric.label}",
                loc="left",
                fontweight="bold",
                color=NAVY
            )

            axes[1].set_xlabel(metric.label)
            axes[1].set_ylabel("Cumulative Share of Store-Days")
            axes[1].spines[["top", "right"]].set_visible(False)
            axes[1].grid(alpha=.15)
            axes[1].legend(frameon=False)

            plt.tight_layout()
            plt.show()


        driver = (
            "customer traffic"
            if abs(customer_change) >= abs(spend_change)
            else "spending per customer"
        )

        display(HTML(
            f"""
            <div style="
                margin-top:16px;
                padding:17px 20px;
                background:#f2eee5;
                border-left:4px solid {GOLD};
                font-family:'Segoe UI',Arial,sans-serif;
                color:#4c4538;
                line-height:1.8;
            ">
                <strong style="color:{NAVY};">
                DYNAMIC BUSINESS INSIGHT
                </strong><br>

                During school holidays, average sales change by
                <strong>{sales_change:+.1f}%</strong> and customer traffic by
                <strong>{customer_change:+.1f}%</strong> relative to normal
                school days. Sales per customer changes by
                <strong>{spend_change:+.1f}%</strong>, indicating that
                <strong>{driver}</strong> represents the stronger behavioural
                movement in this comparison.
            </div>
            """
        ))


for w in [view, store_filter, metric]:
    w.observe(school_dashboard, names="value")


display(
    HTML(
        f"""
        <div style="margin:5px 0 12px;font-family:'Segoe UI';">
            <div style="
                color:{NAVY};
                font-family:Georgia;
                font-size:22px;
                font-weight:600;
            ">
                School Holiday Intelligence Dashboard
            </div>

            <div style="color:{SLATE};font-size:12px;margin-top:4px;">
                Compare customer activity and store performance during
                school-holiday and normal trading periods.
            </div>
        </div>
        """
    ),
    widgets.HBox(
        [view, store_filter, metric],
        layout=widgets.Layout(
            gap="12px",
            flex_flow="row wrap"
        )
    ),
    out
)

school_dashboard()

logger.info("School holiday dashboard initialized.")
```

<h2><span style="color:lightblue">BUSINESS INTERPRETATION<b></b></span>



School Holidays Represent a Distinct Trading Environment

The comparison isolates school-holiday behaviour from state holidays and store
closures, providing a clearer indication of how changes in household routines
relate to normal retail activity. The slopegraph identifies whether the overall
effect is driven primarily by customer traffic or by changes in expenditure per
customer.


The distribution view adds an important second perspective. If the school-
holiday curve is consistently shifted relative to normal days, the effect is
broadly distributed across store-days. If differences occur mainly at the upper
end of the distribution, the average effect may instead be concentrated among
a smaller number of high-performing observations.

For forecasting purposes, SchoolHoliday should therefore be evaluated as a
potential behavioural feature rather than treated merely as a calendar label.
Its interaction with store type, weekday, promotions, and seasonal timing can
be examined later during predictive feature engineering.

Recurring Customer Purchasing Patterns



Beyond individual holidays, retail demand may follow recurring calendar cycles.
This section investigates whether sales, customer traffic, and purchasing value
systematically vary across <strong>months, quarters, weekdays, weekends, and
different stages of the month</strong>.

The interactive dashboard also allows the seasonal pattern to be examined by
year and store type. Normal trading days can be isolated from state holidays so
that recurring calendar behaviour is not dominated by exceptional holiday
closures or demand peaks.

<h2><span style="color:lightblue">Seasonality Intelligence<b></b></span>

```python
season = eda_df[eda_df["Open"].eq(1)].copy()

season["SalesPerCustomer"] = np.where(
    season["Customers"] > 0,
    season["Sales"] / season["Customers"],
    np.nan
)

months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

days = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

periods = [
    "Beginning of Month",
    "Middle of Month",
    "End of Month"
]

season["MonthName"] = pd.Categorical(
    season["MonthName"],
    months,
    ordered=True
)

season["DayName"] = pd.Categorical(
    season["DayName"],
    days,
    ordered=True
)

season["MonthPeriod"] = pd.Categorical(
    season["MonthPeriod"],
    periods,
    ordered=True
)

season_view = widgets.ToggleButtons(
    options=[
        ("📊 Charts", "Charts"),
        ("📋 Table", "Table")
    ],
    value="Charts",
    description="View:"
)

season_metric = widgets.Dropdown(
    options={
        "Sales": "Sales",
        "Customers": "Customers",
        "Sales / Customer": "SalesPerCustomer"
    },
    value="Sales",
    description="Metric:"
)

season_year = widgets.Dropdown(
    options=["All"] + sorted(
        season["Year"].dropna().unique().tolist()
    ),
    value="All",
    description="Year:"
)

season_store = widgets.Dropdown(
    options=["All"] + sorted(
        season["StoreType"].dropna().unique().tolist()
    ),
    value="All",
    description="Store:"
)

season_scope = widgets.Dropdown(
    options=[
        ("Normal Trading", "Normal"),
        ("All Open Days", "All")
    ],
    value="Normal",
    description="Scope:"
)

season_table = widgets.ToggleButtons(
    options=[
        ("Month", "MonthName"),
        ("Weekday", "DayName"),
        ("Quarter", "Quarter"),
        ("Month Stage", "MonthPeriod")
    ],
    value="MonthName"
)

season_out = widgets.Output()

def render_seasonality(*_):

    with season_out:

        clear_output(wait=True)

        d = season.copy()

        if season_year.value != "All":
            d = d[d["Year"].eq(season_year.value)]

        if season_store.value != "All":
            d = d[d["StoreType"].eq(season_store.value)]

        if season_scope.value == "Normal":
            d = d[d["StateHoliday"].eq("0")]

        if d.empty:
            display(HTML(
                "<b>No observations available for the selected filters.</b>"
            ))
            return

        m = season_metric.value

        metric_label = {
            "Sales": "Sales",
            "Customers": "Customers",
            "SalesPerCustomer": "Sales / Customer"
        }[m]

        metric_col = {
            "Sales": "AvgSales",
            "Customers": "AvgCustomers",
            "SalesPerCustomer": "SalesPerCustomer"
        }

        overall_avg = d[m].mean()

        def summary(col):

            s = (
                d.groupby(col, observed=True)
                .agg(
                    StoreDays=("Sales", "size"),
                    AvgSales=("Sales", "mean"),
                    AvgCustomers=("Customers", "mean"),
                    SalesPerCustomer=("SalesPerCustomer", "mean"),
                    PromoRate=("Promo", "mean")
                )
                .reset_index()
            )

            s["PromoRate"] *= 100

            s["Index"] = (
                s[metric_col[m]]
                / overall_avg
                * 100
            )

            return s

        monthly = summary("MonthName")
        weekday = summary("DayName")
        quarter = summary("Quarter")
        month_stage = summary("MonthPeriod")

        best_month = monthly.loc[
            monthly["Index"].idxmax()
        ]

        low_month = monthly.loc[
            monthly["Index"].idxmin()
        ]

        best_day = weekday.loc[
            weekday["Index"].idxmax()
        ]

        best_quarter = quarter.loc[
            quarter["Index"].idxmax()
        ]

        weekday_value = d.loc[
            d["IsWeekend"].eq(0),
            m
        ].mean()

        weekend_value = d.loc[
            d["IsWeekend"].eq(1),
            m
        ].mean()

        weekend_change = (
            weekend_value / weekday_value - 1
        ) * 100

        start_index = month_stage.loc[
            month_stage["MonthPeriod"]
            .astype(str)
            .eq("Beginning of Month"),
            "Index"
        ].mean()

        end_index = month_stage.loc[
            month_stage["MonthPeriod"]
            .astype(str)
            .eq("End of Month"),
            "Index"
        ].mean()

        month_end_change = (
            end_index / start_index - 1
        ) * 100

        seasonal_spread = (
            monthly["Index"].max()
            - monthly["Index"].min()
        )

        cards = [

            (
                "BEST MONTH",
                str(best_month["MonthName"]),
                f"Index {best_month['Index']:.1f}"
            ),

            (
                "LOWEST MONTH",
                str(low_month["MonthName"]),
                f"Index {low_month['Index']:.1f}"
            ),

            (
                "BEST WEEKDAY",
                str(best_day["DayName"]),
                f"Index {best_day['Index']:.1f}"
            ),

            (
                "PEAK QUARTER",
                f"Q{int(best_quarter['Quarter'])}",
                f"Index {best_quarter['Index']:.1f}"
            ),

            (
                "WEEKEND EFFECT",
                f"{weekend_change:+.1f}%",
                "vs weekdays"
            ),

            (
                "MONTH-END EFFECT",
                f"{month_end_change:+.1f}%",
                "vs beginning"
            ),

            (
                "SEASONAL SPREAD",
                f"{seasonal_spread:.1f}",
                "index points"
            ),

            (
                "STORE-DAYS",
                f"{len(d):,}",
                "filtered observations"
            )
        ]


        display(
            HTML(
                "<div style='"
                "display:grid;"
                "grid-template-columns:repeat(4,minmax(0,1fr));"
                "gap:11px;"
                "margin:8px 0 20px;"
                "font-family:Segoe UI;'>"

                + "".join(
                    f"""
                    <div style="
                        background:#ffffff;
                        border:1px solid #ddd7cc;
                        border-top:4px solid {GOLD};
                        border-radius:9px;
                        padding:14px 15px;
                        min-height:95px;
                    ">

                        <div style="
                            font-size:8.5px;
                            color:{SLATE};
                            font-weight:700;
                            letter-spacing:1.2px;
                        ">
                            {title}
                        </div>

                        <div style="
                            margin-top:7px;
                            color:{NAVY};
                            font-family:Georgia;
                            font-size:20px;
                            font-weight:600;
                        ">
                            {value}
                        </div>

                        <div style="
                            margin-top:5px;
                            color:#7b8389;
                            font-size:9.5px;
                        ">
                            {subtitle}
                        </div>

                    </div>
                    """
                    for title, value, subtitle in cards
                )

                + "</div>"
            )
        )

        if season_view.value == "Table":

            tables = {
                "MonthName": monthly,
                "DayName": weekday,
                "Quarter": quarter,
                "MonthPeriod": month_stage
            }

            table = tables[
                season_table.value
            ]

            display(
                table.style

                .format({
                    "StoreDays": "{:,.0f}",
                    "AvgSales": "{:,.2f}",
                    "AvgCustomers": "{:,.2f}",
                    "SalesPerCustomer": "{:.2f}",
                    "PromoRate": "{:.2f}%",
                    "Index": "{:.1f}"
                })

                .background_gradient(
                    subset=["Index"],
                    cmap=sns.light_palette(
                        GOLD,
                        as_cmap=True
                    )
                )

                .set_properties(**{
                    "text-align": "center",
                    "padding": "8px"
                })

                .set_table_styles([
                    {
                        "selector": "th",
                        "props": [
                            ("background-color", NAVY),
                            ("color", WHITE),
                            ("font-weight", "bold"),
                            ("text-align", "center")
                        ]
                    }
                ])
            )

        else:

            fig = plt.figure(
                figsize=(15, 9),
                facecolor=WHITE
            )

            gs = fig.add_gridspec(
                2,
                1,
                height_ratios=[1, 1.3],
                hspace=.42
            )

            ax1 = fig.add_subplot(
                gs[0]
            )

            x = np.arange(
                len(monthly)
            )

            ax1.fill_between(
                x,
                100,
                monthly["Index"].values,
                color=GOLD,
                alpha=.22
            )

            ax1.plot(
                x,
                monthly["Index"],
                marker="o",
                linewidth=2.2,
                color=NAVY
            )

            peak_pos = monthly["Index"].values.argmax()

            ax1.scatter(
                peak_pos,
                monthly["Index"].max(),
                s=120,
                color=GOLD,
                edgecolor=NAVY,
                zorder=5
            )

            ax1.axhline(
                100,
                color=SLATE,
                linestyle="--",
                linewidth=1,
                alpha=.55
            )

            ax1.set_xticks(x)

            ax1.set_xticklabels(
                monthly["MonthName"].astype(str),
                rotation=35,
                ha="right"
            )

            ax1.set_title(
                f"Monthly {metric_label} Seasonal Profile",
                loc="left",
                fontsize=15,
                fontweight="bold",
                color=NAVY,
                pad=12
            )

            ax1.text(
                0,
                1.03,
                "Filtered-period average = 100",
                transform=ax1.transAxes,
                fontsize=9,
                color=SLATE
            )

            ax1.set_ylabel(
                "Seasonality Index"
            )

            ax1.spines[
                ["top", "right"]
            ].set_visible(False)

            ax1.grid(
                axis="y",
                alpha=.13
            )

            ax2 = fig.add_subplot(
                gs[1]
            )

            calendar = (
                d.pivot_table(
                    index="DayName",
                    columns="MonthName",
                    values=m,
                    aggfunc="mean",
                    observed=True
                )
                .reindex(
                    index=days,
                    columns=months
                )
            )

            fmt = (
                ".1f"
                if m == "SalesPerCustomer"
                else ".0f"
            )

            sns.heatmap(
                calendar,
                annot=True,
                fmt=fmt,
                linewidths=.7,
                linecolor=WHITE,
                cmap=sns.light_palette(
                    GOLD,
                    as_cmap=True
                ),
                cbar_kws={
                    "label": metric_label
                },
                ax=ax2
            )

            ax2.set_title(
                f"{metric_label}: Month × Weekday Pattern",
                loc="left",
                fontsize=15,
                fontweight="bold",
                color=NAVY,
                pad=12
            )

            ax2.set_xlabel("")
            ax2.set_ylabel("")
            ax2.set_xticklabels(
                ax2.get_xticklabels(),
                rotation=35,
                ha="right"
            )

            ax2.set_yticklabels(
                ax2.get_yticklabels(),
                rotation=0
            )

            plt.tight_layout()
            plt.show()

        weekend_direction = (
            "higher"
            if weekend_change >= 0
            else "lower"
        )

        month_end_direction = (
            "stronger"
            if month_end_change >= 0
            else "weaker"
        )

        seasonality_strength = (
            "pronounced"
            if seasonal_spread >= 20
            else
            "moderate"
            if seasonal_spread >= 10
            else
            "relatively limited"
        )

        display(
            HTML(
                f"""
                <div style="
                    margin-top:17px;
                    padding:19px 22px;
                    background:#f2eee5;
                    border-left:4px solid {GOLD};
                    border-radius:0 8px 8px 0;
                    color:#4c4538;
                    font-family:'Segoe UI',Arial,sans-serif;
                    font-size:13.5px;
                    line-height:1.85;
                ">

                    <div style="
                        color:{NAVY};
                        font-size:10px;
                        font-weight:700;
                        letter-spacing:1.5px;
                        margin-bottom:7px;
                    ">
                        DYNAMIC SEASONALITY INTERPRETATION
                    </div>

                    <strong>{best_month['MonthName']}</strong>
                    records the strongest {metric_label.lower()} pattern,
                    whereas
                    <strong>{low_month['MonthName']}</strong>
                    records the weakest.

                    The difference between these months is
                    <strong>{seasonal_spread:.1f} index points</strong>,
                    indicating
                    <strong>{seasonality_strength} seasonality</strong>
                    under the selected filters.

                    <br><br>

                    <strong>{best_day['DayName']}</strong>
                    is the strongest weekday and
                    <strong>Q{int(best_quarter['Quarter'])}</strong>
                    is the strongest quarter.

                    Weekend activity is
                    <strong>
                    {abs(weekend_change):.1f}% {weekend_direction}
                    </strong>
                    than weekday activity.

                    <br><br>

                    End-of-month activity is
                    <strong>
                    {abs(month_end_change):.1f}% {month_end_direction}
                    </strong>
                    than the beginning of the month.

                    These patterns indicate that recurring calendar
                    structure should be incorporated into later forecasting
                    rather than treating each store-day as temporally
                    independent.

                </div>
                """
            )
        )

for widget in [
    season_view,
    season_metric,
    season_year,
    season_store,
    season_scope,
    season_table
]:
    widget.observe(
        render_seasonality,
        names="value"
    )

display(
    HTML(
        f"""
        <div style="
            margin:5px 0 12px;
            font-family:'Segoe UI',Arial,sans-serif;
        ">

            <div style="
                color:{NAVY};
                font-family:Georgia,'Times New Roman',serif;
                font-size:23px;
                font-weight:600;
            ">
                Seasonality Intelligence Dashboard
            </div>

            <div style="
                color:{SLATE};
                font-size:12px;
                margin-top:4px;
                line-height:1.55;
            ">
                Explore monthly, quarterly, weekday, weekend and
                intra-month customer purchasing patterns using
                interactive business filters.
            </div>

        </div>
        """
    )
)

controls = widgets.HBox(
    [
        season_view,
        season_metric,
        season_year,
        season_store,
        season_scope
    ],
    layout=widgets.Layout(
        gap="10px",
        flex_flow="row wrap",
        margin="0 0 10px 0"
    )
)

display(
    controls,
    season_table,
    season_out
)

render_seasonality()

logger.info(
    "Seasonality intelligence dashboard initialized."
)
```

<h2><span style="color:lightblue">Seasonality Intelligence · Interpretation<b></b></span>

The seasonality dashboard demonstrates that Rossmann purchasing behaviour should
be viewed as a combination of several recurring calendar effects rather than as
a single monthly pattern. Variations across months, quarters, weekdays,
weekends, and different stages of the month indicate that customer demand is
systematically influenced by when a store-day occurs.

MONTHLY & QUARTERLY DEMAND


Differences between the strongest and weakest months reveal the magnitude of
annual seasonality. Quarter-level patterns provide an additional business view
of broader demand cycles and can support inventory planning, staffing, and
sales-target allocation across the year.


WEEKDAY & WEEKEND EFFECT


Weekday-level differences indicate that purchasing activity is not evenly
distributed across the trading week. The weekend comparison further identifies
whether customer traffic and purchasing intensity systematically strengthen or
weaken relative to normal weekdays.


INTRA-MONTH BEHAVIOUR


Comparing the beginning, middle, and end of the month helps identify whether
purchasing behaviour follows recurring household spending cycles. A consistent
month-end or month-start effect may reflect salary timing, household budgeting,
or recurring consumption requirements.

<h2><span style="color:lightblue">Customer Traffic Intelligence<b></b></span>

Customer traffic is expected to represent one of the most important drivers of
daily retail sales. However, a strong relationship does not necessarily imply
that every additional customer contributes the same amount of revenue or that
the relationship remains identical across stores, promotional conditions, and
time periods.

The following analysis therefore examines both <strong>Pearson and Spearman
correlation</strong>, the strength of the linear relationship, sales generated
per customer, customer-volume deciles, and the stability of the relationship
across store types and promotional conditions.

```python
rel = eda_df[
    eda_df["Open"].eq(1)
    & eda_df["Customers"].gt(0)
].copy()

rel["SalesPerCustomer"] = (
    rel["Sales"] / rel["Customers"]
)

rel_view = widgets.ToggleButtons(
    options=[("📊 Charts","Charts"), ("📋 Table","Table")],
    value="Charts"
)

rel_store = widgets.Dropdown(
    options=["All"] + sorted(rel["StoreType"].dropna().unique().tolist()),
    value="All",
    description="Store:"
)

rel_promo = widgets.Dropdown(
    options=[
        ("All Days","All"),
        ("Promo Days",1),
        ("Non-Promo Days",0)
    ],
    value="All",
    description="Promo:"
)

rel_year = widgets.Dropdown(
    options=["All"] + sorted(rel["Year"].unique().tolist()),
    value="All",
    description="Year:"
)

rel_table = widgets.ToggleButtons(
    options=[
        ("Customer Deciles","Decile"),
        ("Store Types","StoreType")
    ],
    value="Decile"
)

rel_out = widgets.Output()

def render_relationship(*_):

    with rel_out:
        clear_output(wait=True)

        d = rel.copy()

        if rel_store.value != "All":
            d = d[d["StoreType"].eq(rel_store.value)]

        if rel_promo.value != "All":
            d = d[d["Promo"].eq(rel_promo.value)]

        if rel_year.value != "All":
            d = d[d["Year"].eq(rel_year.value)]

        if len(d) < 20:
            print("Insufficient observations for the selected filters.")
            return

        pearson = d["Sales"].corr(d["Customers"])
        spearman = d["Sales"].corr(d["Customers"], method="spearman")
        r2 = pearson ** 2 * 100

        customer_var = d["Customers"].var()

        slope = (
            d["Sales"].cov(d["Customers"]) / customer_var
            if customer_var > 0 else np.nan
        )

        spc = d["Sales"].sum() / d["Customers"].sum()
        median_customers = d["Customers"].median()
        p90_customers = d["Customers"].quantile(.90)

        d = d.copy()

        d["CustomerDecile"] = (
            pd.qcut(
                d["Customers"],
                10,
                labels=False,
                duplicates="drop"
            ) + 1
        )

        deciles = (
            d.groupby("CustomerDecile", observed=True)
            .agg(
                StoreDays=("Sales","size"),
                AvgCustomers=("Customers","mean"),
                AvgSales=("Sales","mean"),
                MedianSales=("Sales","median"),
                SalesQ25=("Sales",lambda x:x.quantile(.25)),
                SalesQ75=("Sales",lambda x:x.quantile(.75)),
                TotalSales=("Sales","sum"),
                TotalCustomers=("Customers","sum")
            )
            .reset_index()
        )

        deciles["SalesPerCustomer"] = (
            deciles["TotalSales"] / deciles["TotalCustomers"]
        )

        cards = [
            ("PEARSON CORR.", f"{pearson:.3f}", "Linear association"),
            ("SPEARMAN CORR.", f"{spearman:.3f}", "Rank association"),
            ("LINEAR R²", f"{r2:.1f}%", "Descriptive fit"),
            ("LINEAR SLOPE", f"{slope:.2f}", "Sales per +1 customer"),
            ("SALES / CUSTOMER", f"{spc:.2f}", "Weighted average"),
            ("MEDIAN CUSTOMERS", f"{median_customers:,.0f}", "Per store-day"),
            ("P90 CUSTOMERS", f"{p90_customers:,.0f}", "High-traffic threshold"),
            ("OBSERVATIONS", f"{len(d):,}", "Filtered store-days")
        ]

        display(HTML(
            "<div style='display:grid;grid-template-columns:repeat(4,1fr);"
            "gap:11px;margin:8px 0 20px;'>"
            + "".join(
                f"""
                <div style="
                    background:#fff;
                    border:1px solid #ddd7cc;
                    border-top:4px solid {GOLD};
                    border-radius:9px;
                    padding:14px 15px;
                    min-height:95px;
                ">
                    <div style="
                        font-size:8.5px;
                        color:{SLATE};
                        font-weight:700;
                        letter-spacing:1.1px;
                    ">{a}</div>

                    <div style="
                        margin-top:7px;
                        color:{NAVY};
                        font-family:Georgia;
                        font-size:21px;
                        font-weight:600;
                    ">{b}</div>

                    <div style="
                        margin-top:5px;
                        color:#7b8389;
                        font-size:9.5px;
                    ">{c}</div>
                </div>
                """
                for a,b,c in cards
            )
            + "</div>"
        ))

        if rel_view.value == "Table":

            if rel_table.value == "Decile":

                t = deciles[
                    [
                        "CustomerDecile",
                        "StoreDays",
                        "AvgCustomers",
                        "AvgSales",
                        "MedianSales",
                        "SalesPerCustomer"
                    ]
                ].copy()

                t["CustomerDecile"] = (
                    "D" + t["CustomerDecile"].astype(int).astype(str)
                )

            else:

                t = (
                    d.groupby("StoreType")
                    .agg(
                        StoreDays=("Sales","size"),
                        AvgCustomers=("Customers","mean"),
                        AvgSales=("Sales","mean"),
                        MedianSales=("Sales","median"),
                        SalesPerCustomer=("SalesPerCustomer","mean")
                    )
                    .reset_index()
                )

                corr = (
                    d.groupby("StoreType")
                    .apply(
                        lambda x: x["Sales"].corr(x["Customers"]),
                        include_groups=False
                    )
                )

                t["SalesCustomerCorr"] = (
                    t["StoreType"].map(corr)
                )

            display(
                t.style
                .format({
                    "StoreDays":"{:,.0f}",
                    "AvgCustomers":"{:,.2f}",
                    "AvgSales":"{:,.2f}",
                    "MedianSales":"{:,.2f}",
                    "SalesPerCustomer":"{:.2f}",
                    "SalesCustomerCorr":"{:.3f}"
                })
                .set_properties(**{
                    "text-align":"center",
                    "padding":"8px"
                })
                .set_table_styles([{
                    "selector":"th",
                    "props":[
                        ("background-color",NAVY),
                        ("color",WHITE),
                        ("font-weight","bold"),
                        ("text-align","center")
                    ]
                }])
            )

        else:

            sample = d.sample(
                min(50000, len(d)),
                random_state=42
            )

            fig, axes = plt.subplots(
                1, 2,
                figsize=(16,6)
            )

            hb = axes[0].hexbin(
                sample["Customers"],
                sample["Sales"],
                gridsize=45,
                mincnt=1,
                cmap=sns.light_palette(
                    NAVY,
                    as_cmap=True
                )
            )

            xline = np.linspace(
                sample["Customers"].quantile(.01),
                sample["Customers"].quantile(.99),
                100
            )

            axes[0].plot(
                xline,
                d["Sales"].mean()
                + slope * (xline - d["Customers"].mean()),
                color=GOLD,
                linewidth=2.2,
                label="Linear relationship"
            )

            axes[0].set_xlim(
                0,
                sample["Customers"].quantile(.995)
            )

            axes[0].set_ylim(
                0,
                sample["Sales"].quantile(.995)
            )

            axes[0].set_title(
                "Sales Density by Customer Traffic",
                loc="left",
                fontsize=15,
                fontweight="bold",
                color=NAVY
            )

            axes[0].text(
                0,1.025,
                "Hexagonal density view; plotting sample used for responsiveness",
                transform=axes[0].transAxes,
                fontsize=8.8,
                color=SLATE
            )

            axes[0].set_xlabel("Customers")
            axes[0].set_ylabel("Sales")
            axes[0].spines[["top","right"]].set_visible(False)
            axes[0].legend(frameon=False)

            fig.colorbar(
                hb,
                ax=axes[0],
                label="Observation Density"
            )

            x = np.arange(len(deciles))

            axes[1].vlines(
                x,
                deciles["SalesQ25"],
                deciles["SalesQ75"],
                color=SLATE,
                linewidth=5,
                alpha=.45
            )

            axes[1].scatter(
                x,
                deciles["MedianSales"],
                s=75,
                color=NAVY,
                label="Median Sales",
                zorder=3
            )

            axes[1].scatter(
                x,
                deciles["AvgSales"],
                s=55,
                marker="D",
                color=GOLD,
                label="Average Sales",
                zorder=3
            )

            axes[1].set_xticks(x)

            axes[1].set_xticklabels(
                [
                    f"D{int(i)}"
                    for i in deciles["CustomerDecile"]
                ]
            )

            axes[1].set_title(
                "Sales Range Across Customer-Volume Deciles",
                loc="left",
                fontsize=15,
                fontweight="bold",
                color=NAVY
            )

            axes[1].text(
                0,1.025,
                "Vertical ranges represent the 25th–75th percentile of daily sales",
                transform=axes[1].transAxes,
                fontsize=8.8,
                color=SLATE
            )

            axes[1].set_xlabel(
                "Customer Traffic Decile →"
            )

            axes[1].set_ylabel("Sales")

            axes[1].spines[
                ["top","right"]
            ].set_visible(False)

            axes[1].grid(
                axis="y",
                alpha=.13
            )

            axes[1].legend(
                frameon=False
            )

            plt.tight_layout()
            plt.show()

        strength = (
            "very strong"
            if abs(pearson) >= .8 else
            "strong"
            if abs(pearson) >= .6 else
            "moderate"
            if abs(pearson) >= .4 else
            "weak"
        )

        monotonic = (
            "highly consistent"
            if abs(spearman - pearson) < .05
            else
            "somewhat different from the purely linear pattern"
        )

        high = deciles.iloc[-1]
        low = deciles.iloc[0]
        decile_uplift = (
            high["AvgSales"] / low["AvgSales"] - 1
        ) * 100

        display(HTML(
            f"""
            <div style="
                margin-top:17px;
                padding:19px 22px;
                background:#f2eee5;
                border-left:4px solid {GOLD};
                border-radius:0 8px 8px 0;
                color:#4c4538;
                font-family:'Segoe UI',Arial,sans-serif;
                font-size:13.5px;
                line-height:1.85;
            ">

                <div style="
                    color:{NAVY};
                    font-size:10px;
                    font-weight:700;
                    letter-spacing:1.5px;
                    margin-bottom:7px;
                ">
                    DYNAMIC CUSTOMER TRAFFIC INTERPRETATION
                </div>

                Sales and customer traffic show a
                <strong>{strength} positive relationship</strong>,
                with Pearson correlation of
                <strong>{pearson:.3f}</strong> and Spearman correlation of
                <strong>{spearman:.3f}</strong>.

                The similarity between these measures indicates that the
                monotonic relationship is <strong>{monotonic}</strong>.

                <br><br>

                The simple linear association produces an R² of
                <strong>{r2:.1f}%</strong>. This should be interpreted as a
                descriptive measure of association rather than proof that
                customer traffic alone causes the observed sales variation.

                The fitted relationship corresponds to approximately
                <strong>{slope:.2f} additional sales units per additional
                customer</strong> within the selected sample conditions.

                <br><br>

                Store-days in the highest customer-volume decile generate
                approximately <strong>{decile_uplift:+.1f}%</strong> more
                average sales than those in the lowest customer decile,
                reinforcing the commercial importance of customer traffic.

                Average realised sales per customer across the filtered data is
                <strong>{spc:.2f}</strong>.
            </div>
            """
        ))

for w in [
    rel_view,
    rel_store,
    rel_promo,
    rel_year,
    rel_table
]:
    w.observe(
        render_relationship,
        names="value"
    )

display(
    HTML(
        f"""
        <div style="
            margin:5px 0 12px;
            font-family:'Segoe UI',Arial,sans-serif;
        ">

            <div style="
                color:{NAVY};
                font-family:Georgia,'Times New Roman',serif;
                font-size:23px;
                font-weight:600;
            ">
                Sales–Customer Relationship Dashboard
            </div>

            <div style="
                color:{SLATE};
                font-size:12px;
                margin-top:4px;
                line-height:1.55;
            ">
                Explore the strength, shape and stability of the relationship
                between customer traffic and daily sales across stores,
                promotions and years.
            </div>

        </div>
        """
    )
)

display(
    widgets.HBox(
        [
            rel_view,
            rel_store,
            rel_promo,
            rel_year
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap",
            margin="0 0 10px 0"
        )
    ),
    rel_table,
    rel_out
)

render_relationship()

logger.info(
    "Sales-customer relationship dashboard initialized."
)
```

<h2><span style="color:lightblue"> BUSINESS INTERPRETATION<b></b></span>

The relationship dashboard evaluates both the strength and consistency of the
association between customer traffic and sales. A strong positive relationship
would indicate that store-days with more customers generally generate higher
sales, making customer traffic an important explanatory component of retail
performance.


However, differences in sales per customer demonstrate why customer volume
alone cannot fully explain store performance. Two store-days with similar
customer counts can generate different sales if purchasing intensity, product
mix, promotional activity, or store characteristics differ.

This distinction leads directly to the next business question: whether
promotions create value primarily by increasing customer traffic, by increasing
the amount purchased per customer, or through a combination of both effects.

<h2><span style="color:lightblue"> Promotion Effectiveness Intelligence<b></b></span>

Promotional effectiveness is examined across three complementary dimensions:
<strong>sales uplift, customer uplift, and sales per customer</strong>. This
distinguishes promotions that attract additional store traffic from those that
also encourage customers to purchase more during each visit.


The analysis also evaluates store-level promotional responsiveness. This is
important because a positive network-wide promotion effect does not imply that
every store benefits equally, and promotional resources may create greater
commercial value when concentrated among more responsive stores.

```python
promo_df = eda_df[
    eda_df["Open"].eq(1)
    & eda_df["Customers"].gt(0)
    & eda_df["StateHoliday"].eq("0")
].copy()

promo_df["SalesPerCustomer"] = (
    promo_df["Sales"] / promo_df["Customers"]
)

# ---------------- CONTROLS ----------------

promo_view = widgets.ToggleButtons(
    options=[("📊 Charts","Charts"), ("📋 Table","Table")],
    value="Charts"
)

promo_store = widgets.Dropdown(
    options=["All"] + sorted(promo_df["StoreType"].dropna().unique().tolist()),
    value="All",
    description="Store:"
)

promo_year = widgets.Dropdown(
    options=["All"] + sorted(promo_df["Year"].unique().tolist()),
    value="All",
    description="Year:"
)

promo_day = widgets.Dropdown(
    options=[
        ("All Days","All"),
        ("Weekdays","Weekday"),
        ("Weekends","Weekend")
    ],
    value="All",
    description="Day:"
)

promo_table = widgets.ToggleButtons(
    options=[
        ("Overall Effect","Overall"),
        ("Store Response","Stores")
    ],
    value="Overall"
)

promo_out = widgets.Output()


# ============================================================
# DASHBOARD
# ============================================================

def render_promo(*_):

    with promo_out:
        clear_output(wait=True)

        d = promo_df.copy()

        if promo_store.value != "All":
            d = d[d["StoreType"].eq(promo_store.value)]

        if promo_year.value != "All":
            d = d[d["Year"].eq(promo_year.value)]

        if promo_day.value == "Weekday":
            d = d[d["IsWeekend"].eq(0)]

        elif promo_day.value == "Weekend":
            d = d[d["IsWeekend"].eq(1)]

        if d["Promo"].nunique() < 2:
            print("Both promo and non-promo observations are required.")
            return

        # ----------------------------------------------------
        # OVERALL EFFECT
        # ----------------------------------------------------

        overall = (
            d.groupby("Promo")
            .agg(
                StoreDays=("Sales","size"),
                AvgSales=("Sales","mean"),
                MedianSales=("Sales","median"),
                AvgCustomers=("Customers","mean"),
                TotalSales=("Sales","sum"),
                TotalCustomers=("Customers","sum")
            )
        )

        overall["SalesPerCustomer"] = (
            overall["TotalSales"] / overall["TotalCustomers"]
        )

        normal, promoted = overall.loc[0], overall.loc[1]

        sales_up = (promoted["AvgSales"] / normal["AvgSales"] - 1) * 100
        cust_up = (promoted["AvgCustomers"] / normal["AvgCustomers"] - 1) * 100
        spend_up = (
            promoted["SalesPerCustomer"]
            / normal["SalesPerCustomer"] - 1
        ) * 100

        promo_share = d["Promo"].mean() * 100
        incremental_sales = promoted["AvgSales"] - normal["AvgSales"]

        # ----------------------------------------------------
        # STORE RESPONSE
        # ----------------------------------------------------

        s = (
            d.groupby(["Store","Promo"])
            .agg(
                Days=("Sales","size"),
                AvgSales=("Sales","mean"),
                AvgCustomers=("Customers","mean"),
                TotalSales=("Sales","sum"),
                TotalCustomers=("Customers","sum")
            )
            .reset_index()
        )

        s["SalesPerCustomer"] = (
            s["TotalSales"] / s["TotalCustomers"]
        )

        wide = s.pivot(
            index="Store",
            columns="Promo",
            values=[
                "Days","AvgSales",
                "AvgCustomers","SalesPerCustomer"
            ]
        )

        wide.columns = [
            f"{a}_{int(b)}"
            for a,b in wide.columns
        ]

        wide = (
            wide.dropna()
            .query("Days_0 >= 5 and Days_1 >= 5")
            .reset_index()
        )

        wide["SalesUplift"] = (
            wide["AvgSales_1"] / wide["AvgSales_0"] - 1
        ) * 100

        wide["CustomerUplift"] = (
            wide["AvgCustomers_1"] / wide["AvgCustomers_0"] - 1
        ) * 100

        wide["SpendUplift"] = (
            wide["SalesPerCustomer_1"]
            / wide["SalesPerCustomer_0"] - 1
        ) * 100

        store_map = (
            d[["Store","StoreType"]]
            .drop_duplicates("Store")
            .set_index("Store")["StoreType"]
        )

        wide["StoreType"] = wide["Store"].map(store_map)

        positive = (
            (wide["SalesUplift"] > 0).mean() * 100
            if len(wide) else np.nan
        )

        median_uplift = (
            wide["SalesUplift"].median()
            if len(wide) else np.nan
        )

        top_store = (
            wide.loc[wide["SalesUplift"].idxmax(), "Store"]
            if len(wide) else "—"
        )

        top_uplift = (
            wide["SalesUplift"].max()
            if len(wide) else np.nan
        )

        # ====================================================
        # KPIs
        # ====================================================

        cards = [
            ("SALES UPLIFT", f"{sales_up:+.1f}%", "Promo vs non-promo"),
            ("CUSTOMER UPLIFT", f"{cust_up:+.1f}%", "Traffic effect"),
            ("SPEND UPLIFT", f"{spend_up:+.1f}%", "Per customer"),
            ("INCREMENTAL SALES", f"{incremental_sales:,.0f}", "Per store-day"),
            ("PROMO SHARE", f"{promo_share:.1f}%", "Filtered store-days"),
            ("POSITIVE STORES", f"{positive:.1f}%", "Stores with sales uplift"),
            ("MEDIAN STORE UPLIFT", f"{median_uplift:+.1f}%", "Across valid stores"),
            ("TOP RESPONSIVE STORE", f"{top_store}", f"Uplift {top_uplift:+.1f}%")
        ]

        display(HTML(
            "<div style='display:grid;grid-template-columns:repeat(4,1fr);"
            "gap:11px;margin:8px 0 20px;'>"
            + "".join(
                f"""
                <div style="
                    background:#fff;
                    border:1px solid #ddd7cc;
                    border-top:4px solid {GOLD};
                    border-radius:9px;
                    padding:14px 15px;
                    min-height:95px;
                ">
                    <div style="
                        font-size:8.5px;color:{SLATE};
                        font-weight:700;letter-spacing:1.1px;
                    ">{a}</div>

                    <div style="
                        margin-top:7px;color:{NAVY};
                        font-family:Georgia;font-size:20px;
                        font-weight:600;
                    ">{b}</div>

                    <div style="
                        margin-top:5px;color:#7b8389;
                        font-size:9.5px;
                    ">{c}</div>
                </div>
                """
                for a,b,c in cards
            )
            + "</div>"
        ))

        # ====================================================
        # TABLE VIEW
        # ====================================================

        if promo_view.value == "Table":

            if promo_table.value == "Overall":

                t = overall[
                    [
                        "StoreDays","AvgSales","MedianSales",
                        "AvgCustomers","SalesPerCustomer"
                    ]
                ].copy()

                t.index = [
                    "No Promotion",
                    "Promotion"
                ]

            else:

                t = (
                    wide[
                        [
                            "Store","StoreType",
                            "SalesUplift",
                            "CustomerUplift",
                            "SpendUplift",
                            "Days_0","Days_1"
                        ]
                    ]
                    .sort_values("SalesUplift", ascending=False)
                    .head(30)
                )

            display(
                t.style
                .format({
                    "StoreDays":"{:,.0f}",
                    "AvgSales":"{:,.2f}",
                    "MedianSales":"{:,.2f}",
                    "AvgCustomers":"{:,.2f}",
                    "SalesPerCustomer":"{:.2f}",
                    "SalesUplift":"{:+.2f}%",
                    "CustomerUplift":"{:+.2f}%",
                    "SpendUplift":"{:+.2f}%",
                    "Days_0":"{:,.0f}",
                    "Days_1":"{:,.0f}"
                })
                .set_properties(**{
                    "text-align":"center",
                    "padding":"8px"
                })
                .set_table_styles([{
                    "selector":"th",
                    "props":[
                        ("background-color",NAVY),
                        ("color",WHITE),
                        ("font-weight","bold"),
                        ("text-align","center")
                    ]
                }])
            )

        # ====================================================
        # CHART VIEW
        # ====================================================

        else:

            fig, axes = plt.subplots(
                1,2,
                figsize=(16,6)
            )

            # -----------------------------------------------
            # STORE PROMOTION UPLIFT DISTRIBUTION
            # -----------------------------------------------

            axes[0].hist(
                wide["SalesUplift"].clip(-50,150),
                bins=28,
                color=CREAM,
                edgecolor=NAVY,
                linewidth=.7
            )

            axes[0].axvline(
                0,
                color=SLATE,
                linestyle="--",
                linewidth=1.2
            )

            axes[0].axvline(
                median_uplift,
                color=GOLD,
                linewidth=2,
                label=f"Median {median_uplift:+.1f}%"
            )

            axes[0].set_title(
                "Distribution of Store-Level Promotion Uplift",
                loc="left",
                fontsize=15,
                fontweight="bold",
                color=NAVY
            )

            axes[0].text(
                0,1.025,
                "Sales uplift for stores with sufficient promo and non-promo history",
                transform=axes[0].transAxes,
                fontsize=8.8,
                color=SLATE
            )

            axes[0].set_xlabel("Promotion Sales Uplift (%)")
            axes[0].set_ylabel("Number of Stores")
            axes[0].spines[["top","right"]].set_visible(False)
            axes[0].grid(axis="y", alpha=.12)
            axes[0].legend(frameon=False)

            # -----------------------------------------------
            # PROMOTION RESPONSE QUADRANT
            # -----------------------------------------------

            plot = wide.copy()

            size = (
                np.clip(plot["SalesUplift"].abs(), 2, 100)
                * 3
            )

            axes[1].scatter(
                plot["CustomerUplift"],
                plot["SpendUplift"],
                s=size,
                alpha=.55,
                color=NAVY,
                edgecolor=WHITE,
                linewidth=.4
            )

            axes[1].axvline(
                0,
                color=SLATE,
                linestyle="--",
                linewidth=1
            )

            axes[1].axhline(
                0,
                color=SLATE,
                linestyle="--",
                linewidth=1
            )

            axes[1].text(
                .98,.96,
                "Traffic ↑\nSpend ↑",
                transform=axes[1].transAxes,
                ha="right",
                va="top",
                fontsize=9,
                color=NAVY
            )

            axes[1].text(
                .02,.96,
                "Traffic ↓\nSpend ↑",
                transform=axes[1].transAxes,
                va="top",
                fontsize=9,
                color=SLATE
            )

            axes[1].text(
                .98,.05,
                "Traffic ↑\nSpend ↓",
                transform=axes[1].transAxes,
                ha="right",
                fontsize=9,
                color=SLATE
            )

            axes[1].set_title(
                "How Promotions Create Value Across Stores",
                loc="left",
                fontsize=15,
                fontweight="bold",
                color=NAVY
            )

            axes[1].text(
                0,1.025,
                "Bubble size represents magnitude of store-level sales uplift",
                transform=axes[1].transAxes,
                fontsize=8.8,
                color=SLATE
            )

            axes[1].set_xlabel("Customer Uplift (%)")
            axes[1].set_ylabel("Sales per Customer Uplift (%)")
            axes[1].spines[["top","right"]].set_visible(False)
            axes[1].grid(alpha=.12)

            plt.tight_layout()
            plt.show()

        # ====================================================
        # DYNAMIC INSIGHT
        # ====================================================

        if cust_up > 0 and spend_up > 0:
            driver = "both higher customer traffic and higher spending per customer"

        elif cust_up > 0:
            driver = "primarily higher customer traffic"

        elif spend_up > 0:
            driver = "primarily higher spending per customer"

        else:
            driver = "neither customer traffic nor spending intensity"

        display(HTML(
            f"""
            <div style="
                margin-top:17px;
                padding:19px 22px;
                background:#f2eee5;
                border-left:4px solid {GOLD};
                border-radius:0 8px 8px 0;
                color:#4c4538;
                font-family:'Segoe UI',Arial,sans-serif;
                font-size:13.5px;
                line-height:1.85;
            ">

                <div style="
                    color:{NAVY};
                    font-size:10px;
                    font-weight:700;
                    letter-spacing:1.5px;
                    margin-bottom:7px;
                ">
                    DYNAMIC PROMOTION INTERPRETATION
                </div>

                Promotional store-days generate
                <strong>{sales_up:+.1f}%</strong> average sales uplift
                relative to comparable non-promotional observations.

                Customer traffic changes by
                <strong>{cust_up:+.1f}%</strong>, while sales per customer
                changes by <strong>{spend_up:+.1f}%</strong>.

                The observed network-level promotional effect is therefore
                associated with <strong>{driver}</strong>.

                <br><br>

                Promotional effectiveness is not uniform across the store
                network. Approximately <strong>{positive:.1f}%</strong> of
                stores with sufficient observations record positive sales
                uplift, while the median store-level response is
                <strong>{median_uplift:+.1f}%</strong>.

                This variation supports selective rather than uniform
                promotional deployment.
            </div>
            """
        ))


# ============================================================
# INTERACTIVITY
# ============================================================

for w in [
    promo_view,
    promo_store,
    promo_year,
    promo_day,
    promo_table
]:
    w.observe(render_promo, names="value")


display(
    HTML(
        f"""
        <div style="margin:5px 0 12px;font-family:'Segoe UI';">

            <div style="
                color:{NAVY};
                font-family:Georgia;
                font-size:23px;
                font-weight:600;
            ">
                Promotion Effectiveness Dashboard
            </div>

            <div style="
                color:{SLATE};
                font-size:12px;
                margin-top:4px;
                line-height:1.55;
            ">
                Measure sales uplift, customer acquisition,
                purchasing intensity and store-level promotional
                responsiveness.
            </div>

        </div>
        """
    )
)

display(
    widgets.HBox(
        [
            promo_view,
            promo_store,
            promo_year,
            promo_day
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap"
        )
    ),
    promo_table,
    promo_out
)

render_promo()

logger.info(
    "Promotion effectiveness dashboard initialized."
)
```

<h2><span style="color:lightblue"> BUSINESS INTERPRETATION<b></b></span>

Promotional effectiveness can arise through two different behavioural channels:
increased customer traffic and increased purchasing value per customer.
Separating these components provides a clearer explanation of why promotional
sales differ from non-promotional sales.


The store-level response distribution is equally important. A strong overall
average does not imply uniform effectiveness across the network. Stores with
persistent positive uplift represent stronger candidates for future promotional
investment, whereas weak-response stores may require different promotional
mechanics or reduced promotional intensity.


These results should be interpreted as observational promotional associations
rather than causal treatment effects, since promotions are not randomly
assigned across stores and dates. Nevertheless, the analysis provides a strong
basis for identifying commercially promising store segments and for developing
promotion-related features in the forecasting model.

<h2><span style="color:lightblue">Promotional Opportunity Intelligence<b></b></span>

A positive network-wide promotional effect does not imply that every store
should receive promotions with equal frequency. Effective deployment requires
identifying stores where promotions are associated with strong sales uplift,
additional customer traffic, higher purchasing value, and consistently positive
performance across time.


The following dashboard therefore ranks stores using a transparent
<strong>Promotion Opportunity Score</strong> combining sales uplift,
incremental sales, customer uplift, spending uplift, and monthly response
consistency. The score is an analytical prioritisation tool rather than a causal
estimate of promotional return.

```python
import plotly.express as px
import plotly.graph_objects as go

CREAM = "#F7F1E3"
LIGHT_GOLD = "#D9C27A"

opp = eda_df[
    eda_df["Open"].eq(1)
    & eda_df["Customers"].gt(0)
    & eda_df["StateHoliday"].eq("0")
].copy()

opp["SalesPerCustomer"] = opp["Sales"] / opp["Customers"]
opp["MonthKey"] = opp["Date"].dt.to_period("M")


# ============================================================
# CONTROLS
# ============================================================

opp_view = widgets.ToggleButtons(
    options=[("📊 Charts","Charts"), ("📋 Table","Table")],
    value="Charts"
)

opp_chart = widgets.ToggleButtons(
    options=[
        ("🧩 Opportunity Map","Treemap"),
        ("🎛 Multi-Factor","Parallel"),
        ("🔻 Decision Funnel","Funnel")
    ],
    value="Treemap"
)

opp_strategy = widgets.Dropdown(
    options=[
        "Balanced",
        "Growth Focus",
        "Traffic Focus",
        "Basket Value",
        "Consistency Focus"
    ],
    value="Balanced",
    description="Strategy:"
)

opp_store = widgets.Dropdown(
    options=["All"] + sorted(opp["StoreType"].dropna().unique()),
    value="All",
    description="Store:"
)

opp_year = widgets.Dropdown(
    options=["All"] + sorted(opp["Year"].unique()),
    value="All",
    description="Year:"
)

opp_priority = widgets.Dropdown(
    options=[
        "All",
        "High Priority",
        "Selective",
        "Low Priority"
    ],
    value="All",
    description="Priority:"
)

opp_sort = widgets.Dropdown(
    options={
        "Opportunity Score":"OpportunityScore",
        "Sales Uplift":"SalesUplift",
        "Incremental Sales":"IncrementalSales",
        "Consistency":"Consistency",
        "Customer Uplift":"CustomerUplift"
    },
    value="OpportunityScore",
    description="Rank:"
)

opp_min_days = widgets.IntSlider(
    value=10,
    min=5,
    max=30,
    step=5,
    description="Min Days:",
    continuous_update=False
)

opp_top = widgets.IntSlider(
    value=15,
    min=5,
    max=30,
    step=5,
    description="Top N:",
    continuous_update=False
)

opp_threshold = widgets.IntSlider(
    value=75,
    min=60,
    max=90,
    step=5,
    description="Priority:",
    continuous_update=False
)

opp_reset = widgets.Button(
    description="Reset Filters",
    icon="refresh"
)

opp_out = widgets.Output()


# ============================================================
# STRATEGY WEIGHTS
# ============================================================

strategies = {
    "Balanced":
        [.30, .25, .20, .15, .10],

    "Growth Focus":
        [.40, .30, .15, .10, .05],

    "Traffic Focus":
        [.20, .15, .15, .40, .10],

    "Basket Value":
        [.20, .15, .15, .10, .40],

    "Consistency Focus":
        [.20, .15, .45, .10, .10]
}

score_columns = [
    "SalesUplift",
    "IncrementalSales",
    "Consistency",
    "CustomerUplift",
    "SpendUplift"
]


# ============================================================
# DASHBOARD
# ============================================================

def render_opportunity(*_):

    with opp_out:
        clear_output(wait=True)

        d = opp.copy()

        if opp_year.value != "All":
            d = d[d["Year"].eq(opp_year.value)]

        if opp_store.value != "All":
            d = d[d["StoreType"].eq(opp_store.value)]

        if d["Promo"].nunique() < 2:
            print("Both promotional and non-promotional observations are required.")
            return


        # ====================================================
        # STORE PERFORMANCE
        # ====================================================

        a = (
            d.groupby(["Store","Promo"])
            .agg(
                Days=("Sales","size"),
                AvgSales=("Sales","mean"),
                AvgCustomers=("Customers","mean"),
                TotalSales=("Sales","sum"),
                TotalCustomers=("Customers","sum")
            )
            .reset_index()
        )

        a["SPC"] = (
            a["TotalSales"] /
            a["TotalCustomers"]
        )

        w = a.pivot(
            index="Store",
            columns="Promo",
            values=[
                "Days",
                "AvgSales",
                "AvgCustomers",
                "SPC"
            ]
        )

        w.columns = [
            f"{x}_{int(y)}"
            for x,y in w.columns
        ]

        w = (
            w.dropna()
            .reset_index()
        )

        w = w[
            (w["Days_0"] >= opp_min_days.value)
            & (w["Days_1"] >= opp_min_days.value)
        ].copy()

        if w.empty:
            print("No stores meet the selected evidence threshold.")
            return


        w["SalesUplift"] = (
            w["AvgSales_1"] /
            w["AvgSales_0"] - 1
        ) * 100

        w["CustomerUplift"] = (
            w["AvgCustomers_1"] /
            w["AvgCustomers_0"] - 1
        ) * 100

        w["SpendUplift"] = (
            w["SPC_1"] /
            w["SPC_0"] - 1
        ) * 100

        w["IncrementalSales"] = (
            w["AvgSales_1"] -
            w["AvgSales_0"]
        )


        # ====================================================
        # RESPONSE CONSISTENCY
        # ====================================================

        monthly = (
            d.groupby(
                ["Store","MonthKey","Promo"]
            )
            .agg(
                Days=("Sales","size"),
                AvgSales=("Sales","mean")
            )
            .reset_index()
        )

        monthly = monthly.pivot(
            index=["Store","MonthKey"],
            columns="Promo",
            values=["Days","AvgSales"]
        )

        monthly.columns = [
            f"{x}_{int(y)}"
            for x,y in monthly.columns
        ]

        monthly = (
            monthly
            .dropna()
            .reset_index()
        )

        monthly = monthly[
            (monthly["Days_0"] >= 2)
            & (monthly["Days_1"] >= 2)
        ]

        consistency = (
            monthly.assign(
                Positive=
                monthly["AvgSales_1"]
                > monthly["AvgSales_0"]
            )
            .groupby("Store")["Positive"]
            .mean()
            .mul(100)
        )

        w["Consistency"] = (
            w["Store"]
            .map(consistency)
            .fillna(0)
        )

        store_map = (
            d[["Store","StoreType"]]
            .drop_duplicates("Store")
            .set_index("Store")["StoreType"]
        )

        w["StoreType"] = (
            w["Store"].map(store_map)
        )


        # ====================================================
        # DYNAMIC OPPORTUNITY SCORE
        # ====================================================

        weights = strategies[
            opp_strategy.value
        ]

        ranks = (
            w[score_columns]
            .rank(pct=True)
        )

        w["OpportunityScore"] = (
            sum(
                ranks[c] * weight
                for c, weight
                in zip(score_columns, weights)
            ) * 100
        )

        selective_threshold = max(
            45,
            opp_threshold.value - 25
        )

        w["Priority"] = "Low Priority"

        w.loc[
            (w["OpportunityScore"] >= selective_threshold)
            & (w["SalesUplift"] > 0),
            "Priority"
        ] = "Selective"

        w.loc[
            (w["OpportunityScore"] >= opp_threshold.value)
            & (w["SalesUplift"] > 0)
            & (w["Consistency"] >= 60),
            "Priority"
        ] = "High Priority"

        w["StoreLabel"] = (
            "Store "
            + w["Store"].astype(int).astype(str)
        )


        # ====================================================
        # DISPLAY FILTER
        # ====================================================

        display_df = w.copy()

        if opp_priority.value != "All":
            display_df = display_df[
                display_df["Priority"]
                .eq(opp_priority.value)
            ]

        if display_df.empty:
            print("No stores match the selected priority filter.")
            return

        ranked = (
            display_df
            .sort_values(
                opp_sort.value,
                ascending=False
            )
            .reset_index(drop=True)
        )

        top = ranked.head(
            opp_top.value
        )


        # ====================================================
        # KPI METRICS
        # ====================================================

        high = w[
            w["Priority"].eq("High Priority")
        ]

        positive = w[
            w["SalesUplift"] > 0
        ]

        consistent = w[
            w["Consistency"] >= 80
        ]

        top_store = (
            w.sort_values(
                "OpportunityScore",
                ascending=False
            )
            .iloc[0]
        )

        uplift_10 = (
            w["SalesUplift"] >= 10
        ).sum()

        positive_incremental = (
            w["IncrementalSales"]
            .clip(lower=0)
        )

        concentration = (
            positive_incremental
            .nlargest(min(10, len(w)))
            .sum()
            / positive_incremental.sum()
            * 100
            if positive_incremental.sum() > 0
            else 0
        )


        # ====================================================
        # KPI CARDS
        # ====================================================

        cards = [

            (
                "ELIGIBLE STORES",
                f"{len(w):,}",
                "Evidence threshold passed"
            ),

            (
                "HIGH PRIORITY",
                f"{len(high):,}",
                f"{len(high)/len(w)*100:.1f}% of eligible"
            ),

            (
                "TOP STORE",
                f"{int(top_store['Store'])}",
                f"Score {top_store['OpportunityScore']:.1f}"
            ),

            (
                "MEDIAN UPLIFT",
                f"{w['SalesUplift'].median():+.1f}%",
                "Promo sales response"
            ),

            (
                "POSITIVE STORES",
                f"{len(positive)/len(w)*100:.1f}%",
                "Sales uplift > 0"
            ),

            (
                "≥10% UPLIFT",
                f"{uplift_10:,}",
                "Strong sales response"
            ),

            (
                "≥80% CONSISTENT",
                f"{len(consistent):,}",
                "Repeatable monthly response"
            ),

            (
                "TOP-10 CONCENTRATION",
                f"{concentration:.1f}%",
                "Positive opportunity share"
            )
        ]


        display(
            HTML(
                "<div style='"
                "display:grid;"
                "grid-template-columns:repeat(4,minmax(0,1fr));"
                "gap:11px;"
                "margin:8px 0 18px;'>"

                + "".join(
                    f"""
                    <div style="
                        background:#ffffff;
                        border:1px solid #ddd7cc;
                        border-top:4px solid {GOLD};
                        border-radius:10px;
                        padding:14px 15px;
                        min-height:94px;
                    ">

                    <div style="
                        font-size:8.5px;
                        color:{SLATE};
                        font-weight:700;
                        letter-spacing:1.1px;
                    ">
                        {a}
                    </div>

                    <div style="
                        margin-top:7px;
                        color:{NAVY};
                        font-family:Georgia;
                        font-size:20px;
                        font-weight:600;
                    ">
                        {b}
                    </div>

                    <div style="
                        margin-top:5px;
                        color:#7b8389;
                        font-size:9.5px;
                    ">
                        {c}
                    </div>

                    </div>
                    """
                    for a,b,c in cards
                )

                + "</div>"
            )
        )


        # ====================================================
        # STRATEGY INFORMATION
        # ====================================================

        weight_text = " · ".join(
            f"{name} {weight:.0%}"
            for name, weight in zip(
                [
                    "Sales",
                    "Incremental",
                    "Consistency",
                    "Traffic",
                    "Basket"
                ],
                weights
            )
        )

        display(
            HTML(
                f"""
                <div style="
                    padding:11px 15px;
                    margin-bottom:15px;
                    background:#edf3f6;
                    border-left:4px solid {NAVY};
                    color:{SLATE};
                    font-family:'Segoe UI';
                    font-size:11.5px;
                ">
                    <strong style="color:{NAVY};">
                        {opp_strategy.value}
                    </strong>
                    &nbsp; | &nbsp;
                    {weight_text}
                    &nbsp; | &nbsp;
                    High-priority score ≥
                    <strong>{opp_threshold.value}</strong>
                </div>
                """
            )
        )


        # ====================================================
        # TABLE VIEW
        # ====================================================

        if opp_view.value == "Table":

            table = top[
                [
                    "Store",
                    "StoreType",
                    "Priority",
                    "OpportunityScore",
                    "SalesUplift",
                    "IncrementalSales",
                    "CustomerUplift",
                    "SpendUplift",
                    "Consistency",
                    "Days_0",
                    "Days_1"
                ]
            ].copy()

            display(
                table.style

                .format({
                    "OpportunityScore":"{:.1f}",
                    "SalesUplift":"{:+.2f}%",
                    "IncrementalSales":"{:,.2f}",
                    "CustomerUplift":"{:+.2f}%",
                    "SpendUplift":"{:+.2f}%",
                    "Consistency":"{:.1f}%",
                    "Days_0":"{:,.0f}",
                    "Days_1":"{:,.0f}"
                })

                .background_gradient(
                    subset=["OpportunityScore"],
                    cmap=sns.light_palette(
                        GOLD,
                        as_cmap=True
                    )
                )

                .set_properties(**{
                    "text-align":"center",
                    "padding":"8px"
                })

                .set_table_styles([{
                    "selector":"th",
                    "props":[
                        ("background-color",NAVY),
                        ("color",WHITE),
                        ("font-weight","bold"),
                        ("text-align","center")
                    ]
                }])
            )


        # ====================================================
        # CHART VIEW
        # ====================================================

        else:

            # ------------------------------------------------
            # TREEMAP
            # ------------------------------------------------

            if opp_chart.value == "Treemap":

                fig = px.treemap(
                    display_df,
                    path=[
                        "StoreType",
                        "Priority",
                        "StoreLabel"
                    ],
                    values="OpportunityScore",
                    color="OpportunityScore",

                    color_continuous_scale=[
                        CREAM,
                        LIGHT_GOLD,
                        GOLD,
                        NAVY
                    ],

                    hover_data={
                        "OpportunityScore":":.1f",
                        "SalesUplift":":+.1f",
                        "IncrementalSales":":,.0f",
                        "CustomerUplift":":+.1f",
                        "SpendUplift":":+.1f",
                        "Consistency":":.1f"
                    }
                )

                fig.update_traces(
                    textinfo="label+value"
                )

                fig.update_layout(
                    title={
                        "text":
                            "Promotion Opportunity Portfolio"
                            "<br><sup>Store Type → Priority → Individual Store</sup>",
                        "x":0.01
                    },

                    height=620,
                    margin=dict(
                        l=15,r=15,t=70,b=15
                    ),

                    paper_bgcolor=WHITE,
                    font=dict(color=NAVY),

                    coloraxis_colorbar=dict(
                        title="Score"
                    )
                )

                fig.show()


            # ------------------------------------------------
            # PARALLEL COORDINATES
            # ------------------------------------------------

            elif opp_chart.value == "Parallel":

                p = top.copy()

                fig = go.Figure(
                    data=go.Parcoords(

                        line=dict(
                            color=p["OpportunityScore"],
                            colorscale=[
                                [0, CREAM],
                                [.5, GOLD],
                                [1, NAVY]
                            ],
                            showscale=True,
                            colorbar=dict(
                                title="Score"
                            )
                        ),

                        dimensions=[

                            dict(
                                label="Store",
                                values=p["Store"]
                            ),

                            dict(
                                label="Sales Uplift %",
                                values=p["SalesUplift"]
                            ),

                            dict(
                                label="Incremental Sales",
                                values=p["IncrementalSales"]
                            ),

                            dict(
                                label="Customer Uplift %",
                                values=p["CustomerUplift"]
                            ),

                            dict(
                                label="Spend Uplift %",
                                values=p["SpendUplift"]
                            ),

                            dict(
                                label="Consistency %",
                                values=p["Consistency"]
                            ),

                            dict(
                                label="Opportunity Score",
                                values=p["OpportunityScore"]
                            )
                        ]
                    )
                )

                fig.update_layout(
                    title={
                        "text":
                            "Multi-Factor Promotion Candidate Profile"
                            "<br><sup>Drag across any axis to interactively isolate stores</sup>",
                        "x":0.01
                    },

                    height=620,

                    margin=dict(
                        l=70,r=80,t=90,b=45
                    ),

                    paper_bgcolor=WHITE,
                    font=dict(
                        color=NAVY,
                        size=11
                    )
                )

                fig.show()


            # ------------------------------------------------
            # DECISION FUNNEL
            # ------------------------------------------------

            else:

                positive_sales = (
                    w["SalesUplift"] > 0
                )

                stable = (
                    positive_sales
                    & (w["Consistency"] >= 60)
                )

                high_mask = (
                    w["Priority"]
                    .eq("High Priority")
                )

                funnel = pd.DataFrame({
                    "Stage":[
                        "Eligible Stores",
                        "Positive Sales Uplift",
                        "Consistent Positive Response",
                        "High-Priority Candidates"
                    ],

                    "Stores":[
                        len(w),
                        positive_sales.sum(),
                        stable.sum(),
                        high_mask.sum()
                    ]
                })

                fig = go.Figure(
                    go.Funnel(
                        y=funnel["Stage"],
                        x=funnel["Stores"],

                        textinfo=
                            "value+percent initial",

                        marker=dict(
                            color=[
                                NAVY,
                                BLUE,
                                GOLD,
                                LIGHT_GOLD
                            ]
                        )
                    )
                )

                fig.update_layout(
                    title={
                        "text":
                            "Promotion Candidate Decision Funnel"
                            "<br><sup>From sufficient evidence to final promotional priority</sup>",
                        "x":0.01
                    },

                    height=560,
                    paper_bgcolor=WHITE,
                    font=dict(color=NAVY)
                )

                fig.show()


        # ====================================================
        # TOP RECOMMENDATION CARDS
        # ====================================================

        recommendation_cards = ""

        for _, r in top.head(3).iterrows():

            recommendation_cards += f"""
            <div style="
                background:#ffffff;
                border:1px solid #ddd7cc;
                border-radius:9px;
                padding:15px 17px;
            ">

                <div style="
                    color:{NAVY};
                    font-family:Georgia;
                    font-size:18px;
                    font-weight:600;
                ">
                    Store {int(r['Store'])}
                </div>

                <div style="
                    color:{GOLD};
                    font-size:11px;
                    font-weight:700;
                    margin:5px 0 8px;
                ">
                    {r['Priority']} · Score {r['OpportunityScore']:.1f}
                </div>

                <div style="
                    color:{SLATE};
                    font-size:11px;
                    line-height:1.7;
                ">
                    Sales uplift:
                    <strong>{r['SalesUplift']:+.1f}%</strong><br>

                    Customer uplift:
                    <strong>{r['CustomerUplift']:+.1f}%</strong><br>

                    Spend uplift:
                    <strong>{r['SpendUplift']:+.1f}%</strong><br>

                    Consistency:
                    <strong>{r['Consistency']:.1f}%</strong>
                </div>

            </div>
            """

        display(
            HTML(
                f"""
                <div style="
                    margin-top:18px;
                    font-family:'Segoe UI';
                ">

                    <div style="
                        color:{NAVY};
                        font-size:10px;
                        font-weight:700;
                        letter-spacing:1.5px;
                        margin-bottom:9px;
                    ">
                        TOP PROMOTIONAL CANDIDATES
                    </div>

                    <div style="
                        display:grid;
                        grid-template-columns:repeat(3,1fr);
                        gap:11px;
                    ">
                        {recommendation_cards}
                    </div>

                </div>
                """
            )
        )


        # ====================================================
        # DYNAMIC INTERPRETATION
        # ====================================================

        best = top_store

        display(
            HTML(
                f"""
                <div style="
                    margin-top:17px;
                    padding:19px 22px;
                    background:#f2eee5;
                    border-left:4px solid {GOLD};
                    border-radius:0 8px 8px 0;
                    color:#4c4538;
                    font-family:'Segoe UI';
                    font-size:13.5px;
                    line-height:1.85;
                ">

                    <strong style="color:{NAVY};">
                        DYNAMIC DEPLOYMENT RECOMMENDATION
                    </strong><br>

                    Under the
                    <strong>{opp_strategy.value}</strong>
                    strategy,
                    <strong>{len(high)}</strong> of
                    <strong>{len(w)}</strong> eligible stores are currently
                    classified as high-priority promotional candidates.

                    <br><br>

                    Store <strong>{int(best['Store'])}</strong>
                    ranks highest overall with a score of
                    <strong>{best['OpportunityScore']:.1f}</strong>,
                    sales uplift of
                    <strong>{best['SalesUplift']:+.1f}%</strong>,
                    customer uplift of
                    <strong>{best['CustomerUplift']:+.1f}%</strong>,
                    and monthly response consistency of
                    <strong>{best['Consistency']:.1f}%</strong>.

                    <br><br>

                    The top 10 stores account for approximately
                    <strong>{concentration:.1f}%</strong>
                    of positive incremental-sales opportunity,
                    indicating the degree to which promotional value is
                    concentrated within the network.

                    Changing the strategy selector recalculates the ranking,
                    allowing promotional deployment to prioritise
                    <strong>growth, traffic acquisition, basket value,
                    consistency, or a balanced combination</strong>.
                </div>
                """
            )
        )


# ============================================================
# RESET
# ============================================================

def reset_opportunity(_):

    opp_strategy.value = "Balanced"
    opp_store.value = "All"
    opp_year.value = "All"
    opp_priority.value = "All"
    opp_sort.value = "OpportunityScore"
    opp_min_days.value = 10
    opp_top.value = 15
    opp_threshold.value = 75
    opp_chart.value = "Treemap"
    opp_view.value = "Charts"


opp_reset.on_click(
    reset_opportunity
)


# ============================================================
# INTERACTIVITY
# ============================================================

for control in [
    opp_view,
    opp_chart,
    opp_strategy,
    opp_store,
    opp_year,
    opp_priority,
    opp_sort,
    opp_min_days,
    opp_top,
    opp_threshold
]:
    control.observe(
        render_opportunity,
        names="value"
    )


# ============================================================
# DISPLAY
# ============================================================

display(
    HTML(
        f"""
        <div style="
            margin:5px 0 13px;
            font-family:'Segoe UI';
        ">

            <div style="
                color:{NAVY};
                font-family:Georgia;
                font-size:24px;
                font-weight:600;
            ">
                Promotion Opportunity Intelligence Dashboard
            </div>

            <div style="
                color:{SLATE};
                font-size:12px;
                margin-top:5px;
                line-height:1.6;
            ">
                Interactively evaluate promotional candidates,
                change commercial strategy, test evidence thresholds,
                explore store-level drivers and generate a prioritised
                promotion shortlist.
            </div>

        </div>
        """
    )
)

display(
    widgets.HBox(
        [
            opp_view,
            opp_strategy,
            opp_store,
            opp_year,
            opp_priority
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap",
            margin="0 0 9px 0"
        )
    ),

    widgets.HBox(
        [
            opp_sort,
            opp_min_days,
            opp_top,
            opp_threshold,
            opp_reset
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap",
            margin="0 0 9px 0"
        )
    ),

    opp_chart,

    opp_out
)

render_opportunity()

logger.info(
    "Advanced promotion opportunity dashboard initialized."
)
```

<h2><span style="color:lightblue">BUSINESS INTERPRETATION<b></b></span>




Store-level promotional performance should be evaluated using both the magnitude
and the consistency of the observed response. Stores displaying strong sales
uplift together with repeatable positive monthly performance represent more
credible candidates for promotional prioritisation than stores whose apparent
uplift is based on limited or inconsistent historical evidence.



Customer and spending responses provide further diagnostic value. A promotion
that increases customer traffic may be effective for acquisition or visit
generation, whereas increased sales per customer suggests stronger purchasing
intensity. Stores exhibiting improvement across both dimensions provide the
most comprehensive promotional response.

The ranking should be interpreted as an <strong>opportunity shortlist rather
than a causal deployment rule</strong>. Historical promotions were not randomly
assigned, so factors such as seasonality, store characteristics and local demand
may contribute to the observed differences. Nevertheless, the framework
provides a practical evidence-based approach for identifying stores that warrant
further promotional testing.



<h2><span style="color:lightblue">Store Operations Intelligence<b></b></span>

Store availability directly determines when customer demand can be converted
into realised sales. This section examines recurring operating patterns across
the trading week, identifies stores that remain consistently open throughout
the standard Monday-to-Friday period, and evaluates whether their weekend sales
and customer behaviour differ from stores with less consistent weekday
availability.



<strong style="color:#102e91;">Data limitation:</strong>
The available data contains a daily <code>Open</code> indicator but no
hour-level opening or closing timestamps. Consequently, intraday behaviour
immediately after opening or before closing cannot be measured reliably.
The analysis therefore focuses on daily operating availability, weekday
coverage, weekend opening patterns, customer traffic, and realised sales.

```python
ops = eda_df[
    eda_df["StateHoliday"].eq("0")
].copy()

ops["SalesPerCustomer"] = np.where(
    ops["Customers"] > 0,
    ops["Sales"] / ops["Customers"],
    np.nan
)


# ============================================================
# CONTROLS
# ============================================================

ops_view = widgets.ToggleButtons(
    options=[
        ("📊 Charts", "Charts"),
        ("📋 Table", "Table")
    ],
    value="Charts"
)

ops_chart = widgets.ToggleButtons(
    options=[
        ("🔀 Operating Network", "Sankey"),
        ("📦 Weekend Performance", "Box"),
        ("🗓 Weekly Profile", "Matrix")
    ],
    value="Sankey"
)

ops_metric = widgets.Dropdown(
    options={
        "Sales": "Sales",
        "Customers": "Customers",
        "Sales / Customer": "SalesPerCustomer"
    },
    value="Sales",
    description="Metric:"
)

ops_year = widgets.Dropdown(
    options=["All"] + sorted(
        ops["Year"].dropna().unique().tolist()
    ),
    value="All",
    description="Year:"
)

ops_store = widgets.Dropdown(
    options=["All"] + sorted(
        ops["StoreType"].dropna().unique().tolist()
    ),
    value="All",
    description="Store:"
)

ops_weekend = widgets.Dropdown(
    options=[
        ("Saturday + Sunday", "Both"),
        ("Saturday", "Saturday"),
        ("Sunday", "Sunday")
    ],
    value="Both",
    description="Weekend:"
)

ops_threshold = widgets.IntSlider(
    value=95,
    min=80,
    max=100,
    step=5,
    description="Weekday %:",
    continuous_update=False
)

ops_table = widgets.ToggleButtons(
    options=[
        ("Store Classification", "Stores"),
        ("Day Profile", "Days")
    ],
    value="Stores"
)

ops_reset = widgets.Button(
    description="Reset Filters",
    icon="refresh"
)

ops_out = widgets.Output()


# ============================================================
# DASHBOARD
# ============================================================

def render_operations(*_):

    with ops_out:

        clear_output(wait=True)

        d = ops.copy()

        if ops_year.value != "All":
            d = d[d["Year"].eq(ops_year.value)]

        if ops_store.value != "All":
            d = d[d["StoreType"].eq(ops_store.value)]

        if d.empty:
            display(
                HTML(
                    "<b>No observations available for the selected filters.</b>"
                )
            )
            return


        # ====================================================
        # METRIC SETTINGS
        # ============================================================

        metric = ops_metric.value

        metric_label = {
            "Sales": "Sales",
            "Customers": "Customers",
            "SalesPerCustomer": "Sales / Customer"
        }[metric]

        performance_col = {
            "Sales": "AvgSales",
            "Customers": "AvgCustomers",
            "SalesPerCustomer": "SalesPerCustomer"
        }[metric]


        # ====================================================
        # STORE × DAY OPERATING PROFILE
        # ============================================================

        profile = (
            d.groupby(
                ["Store", "DayOfWeek"]
            )
            .agg(
                OpenRate=("Open", "mean"),
                AvgSales=("Sales", "mean"),
                AvgCustomers=("Customers", "mean"),
                SalesPerCustomer=("SalesPerCustomer", "mean")
            )
            .reset_index()
        )

        profile["OpenRate"] *= 100

        open_matrix = (
            profile.pivot(
                index="Store",
                columns="DayOfWeek",
                values="OpenRate"
            )
            .reindex(columns=range(1, 8))
            .fillna(0)
        )

        weekday_rate = (
            open_matrix[
                [1, 2, 3, 4, 5]
            ]
            .mean(axis=1)
        )

        weekday_min = (
            open_matrix[
                [1, 2, 3, 4, 5]
            ]
            .min(axis=1)
        )


        # ====================================================
        # STORE CLASSIFICATION
        # ============================================================

        classification = pd.DataFrame({
            "Store": open_matrix.index,
            "WeekdayOpenRate": weekday_rate.values,
            "SaturdayOpenRate": open_matrix[6].values,
            "SundayOpenRate": open_matrix[7].values
        }).reset_index(drop=True)

        classification["AllWeekdaysOpen"] = (
            weekday_min.values
            >= ops_threshold.value
        )

        classification["WeekendOpen"] = (
            classification[
                [
                    "SaturdayOpenRate",
                    "SundayOpenRate"
                ]
            ]
            .max(axis=1)
            > 0
        )

        type_map = (
            d[
                ["Store", "StoreType"]
            ]
            .drop_duplicates("Store")
            .set_index("Store")["StoreType"]
        )

        classification["StoreType"] = (
            classification["Store"]
            .map(type_map)
        )

        classification["OperatingGroup"] = np.where(
            classification["AllWeekdaysOpen"],
            "All Weekdays Open",
            "Variable Weekday Opening"
        )

        classification["WeekendGroup"] = np.where(
            classification["WeekendOpen"],
            "Weekend Open",
            "Weekend Closed"
        )

        store_group = (
            classification
            .set_index("Store")["OperatingGroup"]
        )

        d["OperatingGroup"] = (
            d["Store"]
            .map(store_group)
        )


        # ====================================================
        # WEEKDAY / WEEKEND DATA
        # ============================================================

        weekday = d[
            d["DayOfWeek"].isin(
                [1, 2, 3, 4, 5]
            )
            & d["Open"].eq(1)
        ]

        if ops_weekend.value == "Saturday":

            weekend = d[
                d["DayOfWeek"].eq(6)
                & d["Open"].eq(1)
            ]

        elif ops_weekend.value == "Sunday":

            weekend = d[
                d["DayOfWeek"].eq(7)
                & d["Open"].eq(1)
            ]

        else:

            weekend = d[
                d["DayOfWeek"].isin([6, 7])
                & d["Open"].eq(1)
            ]


        weekday_avg = weekday[metric].mean()
        weekend_avg = weekend[metric].mean()

        weekend_effect = (
            (weekend_avg / weekday_avg - 1) * 100
            if pd.notna(weekday_avg)
            and weekday_avg != 0
            and pd.notna(weekend_avg)
            else np.nan
        )


        # ====================================================
        # PERFORMANCE BY OPERATING GROUP
        # ============================================================

        group_perf = (
            weekend.groupby(
                "OperatingGroup"
            )
            .agg(
                StoreDays=("Sales", "size"),
                AvgSales=("Sales", "mean"),
                AvgCustomers=("Customers", "mean"),
                SalesPerCustomer=(
                    "SalesPerCustomer",
                    "mean"
                )
            )
            .reset_index()
        )

        group_map = (
            group_perf
            .set_index("OperatingGroup")
        )

        stable_weekend = (
            group_map.loc[
                "All Weekdays Open",
                performance_col
            ]
            if "All Weekdays Open"
            in group_map.index
            else np.nan
        )

        variable_weekend = (
            group_map.loc[
                "Variable Weekday Opening",
                performance_col
            ]
            if "Variable Weekday Opening"
            in group_map.index
            else np.nan
        )

        group_effect = (
            (
                stable_weekend
                / variable_weekend
                - 1
            ) * 100
            if pd.notna(stable_weekend)
            and pd.notna(variable_weekend)
            and variable_weekend != 0
            else np.nan
        )


        # ====================================================
        # DAY PERFORMANCE
        # ============================================================

        day_perf = (
            d[
                d["Open"].eq(1)
            ]
            .groupby("DayOfWeek")
            .agg(
                StoreDays=("Sales", "size"),
                OpenStores=("Store", "nunique"),
                AvgSales=("Sales", "mean"),
                AvgCustomers=("Customers", "mean"),
                SalesPerCustomer=(
                    "SalesPerCustomer",
                    "mean"
                )
            )
            .reset_index()
        )

        day_names = {
            1: "Monday",
            2: "Tuesday",
            3: "Wednesday",
            4: "Thursday",
            5: "Friday",
            6: "Saturday",
            7: "Sunday"
        }

        day_perf["Day"] = (
            day_perf["DayOfWeek"]
            .map(day_names)
        )

        best_day_row = day_perf.loc[
            day_perf[
                performance_col
            ].idxmax()
        ]


        # ====================================================
        # KPI CALCULATIONS
        # ============================================================

        all_weekday_n = int(
            classification[
                "AllWeekdaysOpen"
            ].sum()
        )

        all_weekday_share = (
            all_weekday_n
            / len(classification)
            * 100
        )

        weekend_open_n = int(
            classification[
                "WeekendOpen"
            ].sum()
        )

        weekend_open_share = (
            weekend_open_n
            / len(classification)
            * 100
        )


        # ====================================================
        # KPI CARDS
        # ============================================================

        cards = [

            (
                "ALL-WEEKDAY STORES",
                f"{all_weekday_n:,}",
                f"{all_weekday_share:.1f}% of stores"
            ),

            (
                "WEEKEND-OPEN STORES",
                f"{weekend_open_n:,}",
                f"{weekend_open_share:.1f}% of stores"
            ),

            (
                "WEEKDAY LEVEL",
                f"{weekday_avg:,.1f}",
                metric_label
            ),

            (
                "WEEKEND LEVEL",
                (
                    f"{weekend_avg:,.1f}"
                    if pd.notna(weekend_avg)
                    else "—"
                ),
                metric_label
            ),

            (
                "WEEKEND EFFECT",
                (
                    f"{weekend_effect:+.1f}%"
                    if pd.notna(weekend_effect)
                    else "—"
                ),
                "Weekend vs weekday"
            ),

            (
                "CONSISTENCY EFFECT",
                (
                    f"{group_effect:+.1f}%"
                    if pd.notna(group_effect)
                    else "—"
                ),
                "Stable vs variable stores"
            ),

            (
                "STRONGEST DAY",
                str(
                    best_day_row["Day"]
                ),
                f"{metric_label} peak"
            ),

            (
                "STORES ANALYSED",
                f"{len(classification):,}",
                f"Threshold ≥ {ops_threshold.value}%"
            )
        ]


        # ====================================================
        # KPI DISPLAY
        # ============================================================

        display(
            HTML(
                "<div style='"
                "display:grid;"
                "grid-template-columns:repeat(4,minmax(0,1fr));"
                "gap:11px;"
                "margin:8px 0 19px;'>"

                + "".join(
                    f"""
                    <div style="
                        background:#ffffff;
                        border:1px solid #ddd7cc;
                        border-top:4px solid {GOLD};
                        border-radius:10px;
                        padding:14px 15px;
                        min-height:96px;
                        box-shadow:0 2px 6px rgba(12,36,57,.05);
                    ">

                        <div style="
                            color:{SLATE};
                            font-size:8.5px;
                            font-weight:700;
                            letter-spacing:1.2px;
                        ">
                            {title}
                        </div>

                        <div style="
                            color:{NAVY};
                            font-family:Georgia;
                            font-size:20px;
                            font-weight:600;
                            margin-top:7px;
                        ">
                            {value}
                        </div>

                        <div style="
                            color:#7b8389;
                            font-size:9.5px;
                            margin-top:5px;
                        ">
                            {subtitle}
                        </div>

                    </div>
                    """
                    for title, value, subtitle
                    in cards
                )

                + "</div>"
            )
        )


        # ====================================================
        # CURRENT VIEW SUMMARY
        # ============================================================

        display(
            HTML(
                f"""
                <div style="
                    padding:11px 15px;
                    margin-bottom:16px;
                    background:#edf3f6;
                    border-left:4px solid {NAVY};
                    color:{SLATE};
                    font-family:'Segoe UI';
                    font-size:11.5px;
                    line-height:1.6;
                ">

                    <strong style="color:{NAVY};">
                        Current analytical view
                    </strong>

                    &nbsp; · &nbsp;

                    Metric:
                    <strong>{metric_label}</strong>

                    &nbsp; · &nbsp;

                    Weekend:
                    <strong>{ops_weekend.label}</strong>

                    &nbsp; · &nbsp;

                    Weekday threshold:
                    <strong>{ops_threshold.value}%</strong>

                    &nbsp; · &nbsp;

                    Year:
                    <strong>{ops_year.value}</strong>

                </div>
                """
            )
        )


        # ====================================================
        # TABLE VIEW — CORRECTED
        # ============================================================

        if ops_view.value == "Table":

            if ops_table.value == "Stores":

                # IMPORTANT:
                # Sort BEFORE removing helper columns
                table = (
                    classification
                    .sort_values(
                        [
                            "AllWeekdaysOpen",
                            "WeekdayOpenRate",
                            "WeekendOpen"
                        ],
                        ascending=[
                            False,
                            False,
                            False
                        ]
                    )
                    [
                        [
                            "Store",
                            "StoreType",
                            "OperatingGroup",
                            "WeekdayOpenRate",
                            "SaturdayOpenRate",
                            "SundayOpenRate",
                            "WeekendGroup"
                        ]
                    ]
                    .reset_index(drop=True)
                )

                display(
                    table.style

                    .format({
                        "WeekdayOpenRate": "{:.1f}%",
                        "SaturdayOpenRate": "{:.1f}%",
                        "SundayOpenRate": "{:.1f}%"
                    })

                    .background_gradient(
                        subset=[
                            "WeekdayOpenRate",
                            "SaturdayOpenRate",
                            "SundayOpenRate"
                        ],
                        cmap=sns.light_palette(
                            GOLD,
                            as_cmap=True
                        )
                    )

                    .set_properties(**{
                        "text-align": "center",
                        "padding": "8px"
                    })

                    .set_table_styles([
                        {
                            "selector": "th",
                            "props": [
                                (
                                    "background-color",
                                    NAVY
                                ),
                                (
                                    "color",
                                    WHITE
                                ),
                                (
                                    "font-weight",
                                    "bold"
                                ),
                                (
                                    "text-align",
                                    "center"
                                ),
                                (
                                    "padding",
                                    "9px"
                                )
                            ]
                        },

                        {
                            "selector": "td",
                            "props": [
                                (
                                    "border-bottom",
                                    "1px solid #e6e1d8"
                                )
                            ]
                        }
                    ])
                )


            else:

                day_table = (
                    day_perf[
                        [
                            "Day",
                            "StoreDays",
                            "OpenStores",
                            "AvgSales",
                            "AvgCustomers",
                            "SalesPerCustomer"
                        ]
                    ]
                    .copy()
                )

                display(
                    day_table.style

                    .format({
                        "StoreDays": "{:,.0f}",
                        "OpenStores": "{:,.0f}",
                        "AvgSales": "{:,.2f}",
                        "AvgCustomers": "{:,.2f}",
                        "SalesPerCustomer": "{:.2f}"
                    })

                    .background_gradient(
                        subset=[
                            "AvgSales",
                            "AvgCustomers",
                            "SalesPerCustomer"
                        ],
                        cmap=sns.light_palette(
                            GOLD,
                            as_cmap=True
                        )
                    )

                    .set_properties(**{
                        "text-align": "center",
                        "padding": "8px"
                    })

                    .set_table_styles([
                        {
                            "selector": "th",
                            "props": [
                                (
                                    "background-color",
                                    NAVY
                                ),
                                (
                                    "color",
                                    WHITE
                                ),
                                (
                                    "font-weight",
                                    "bold"
                                ),
                                (
                                    "text-align",
                                    "center"
                                ),
                                (
                                    "padding",
                                    "9px"
                                )
                            ]
                        }
                    ])
                )


        # ====================================================
        # CHART VIEW
        # ============================================================

        else:

            # =================================================
            # OPERATING NETWORK — SANKEY
            # =================================================

            if ops_chart.value == "Sankey":

                flow = (
                    classification
                    .groupby(
                        [
                            "OperatingGroup",
                            "WeekendGroup"
                        ]
                    )
                    .size()
                    .reset_index(
                        name="Stores"
                    )
                )

                labels = [
                    "All Weekdays Open",
                    "Variable Weekday Opening",
                    "Weekend Open",
                    "Weekend Closed"
                ]

                index_map = {
                    name: i
                    for i, name
                    in enumerate(labels)
                }

                fig = go.Figure(
                    go.Sankey(

                        arrangement="snap",

                        node=dict(
                            label=labels,
                            pad=25,
                            thickness=24,
                            line=dict(
                                width=.6
                            )
                        ),

                        link=dict(

                            source=[
                                index_map[x]
                                for x
                                in flow[
                                    "OperatingGroup"
                                ]
                            ],

                            target=[
                                index_map[x]
                                for x
                                in flow[
                                    "WeekendGroup"
                                ]
                            ],

                            value=flow["Stores"],

                            hovertemplate=(
                                "%{source.label}"
                                " → "
                                "%{target.label}"
                                "<br>Stores: %{value:,}"
                                "<extra></extra>"
                            )
                        )
                    )
                )

                fig.update_layout(

                    title={
                        "text":
                            "From Weekday Reliability to Weekend Availability"
                            "<br><sup>"
                            "How store operating consistency translates "
                            "into weekend access"
                            "</sup>",
                        "x": .02
                    },

                    height=520,

                    margin=dict(
                        l=25,
                        r=25,
                        t=85,
                        b=20
                    ),

                    paper_bgcolor=WHITE,

                    font=dict(
                        color=NAVY,
                        size=12
                    )
                )

                fig.show()


            # =================================================
            # WEEKEND PERFORMANCE DISTRIBUTION
            # =================================================

            elif ops_chart.value == "Box":

                if weekend.empty:

                    display(
                        HTML(
                            "<b>No open weekend observations "
                            "are available for the selected filters.</b>"
                        )
                    )
                    return

                fig = px.box(

                    weekend,

                    x="OperatingGroup",

                    y=metric,

                    points="outliers",

                    hover_data={
                        "Store": True,
                        "StoreType": True,
                        "DayOfWeek": True,
                        "Promo": True
                    }
                )

                fig.update_layout(

                    title={
                        "text":
                            "Does Weekday Operating Consistency Translate "
                            f"Into Stronger Weekend {metric_label}?"
                            "<br><sup>"
                            "Distribution across stores operating during "
                            "the selected weekend period"
                            "</sup>",
                        "x": .02
                    },

                    xaxis_title=(
                        "Weekday Operating Pattern"
                    ),

                    yaxis_title=metric_label,

                    height=560,

                    paper_bgcolor=WHITE,

                    plot_bgcolor=WHITE,

                    font=dict(
                        color=NAVY
                    )
                )

                fig.update_xaxes(
                    showgrid=False
                )

                fig.update_yaxes(
                    gridcolor="#ece8df"
                )

                fig.show()


            # =================================================
            # WEEKLY AVAILABILITY MATRIX
            # =================================================

            else:

                matrix = (
                    classification
                    .set_index("Store")
                    .join(open_matrix)
                    .sort_values(
                        [
                            "AllWeekdaysOpen",
                            "WeekdayOpenRate"
                        ],
                        ascending=[
                            False,
                            False
                        ]
                    )
                    .head(100)
                )

                fig = go.Figure(
                    go.Heatmap(

                        z=matrix[
                            [
                                1, 2, 3, 4,
                                5, 6, 7
                            ]
                        ].values,

                        x=[
                            "Monday",
                            "Tuesday",
                            "Wednesday",
                            "Thursday",
                            "Friday",
                            "Saturday",
                            "Sunday"
                        ],

                        y=[
                            f"Store {int(x)}"
                            for x
                            in matrix.index
                        ],

                        colorscale=[
                            [0, CREAM],
                            [.50, LIGHT_GOLD],
                            [1, NAVY]
                        ],

                        zmin=0,

                        zmax=100,

                        colorbar=dict(
                            title="Open %"
                        ),

                        hovertemplate=(
                            "%{y}"
                            "<br>%{x}"
                            "<br>Opening rate: %{z:.1f}%"
                            "<extra></extra>"
                        )
                    )
                )

                fig.update_layout(

                    title={
                        "text":
                            "Weekly Store Availability Signature"
                            "<br><sup>"
                            "Historical opening reliability across "
                            "the seven-day trading week"
                            "</sup>",
                        "x": .02
                    },

                    height=720,

                    xaxis_title="Trading Day",

                    yaxis_title="Store",

                    paper_bgcolor=WHITE,

                    font=dict(
                        color=NAVY
                    )
                )

                fig.show()


        # ====================================================
        # DYNAMIC INTERPRETATION
        # ============================================================

        if pd.notna(weekend_effect):

            weekend_text = (
                f"<strong>{abs(weekend_effect):.1f}% "
                f"{'higher' if weekend_effect >= 0 else 'lower'}</strong>"
            )

        else:

            weekend_text = (
                "<strong>not estimable</strong>"
            )


        if pd.notna(group_effect):

            consistency_text = (
                "Stores classified as consistently open "
                "across all weekdays record "
                f"<strong>{abs(group_effect):.1f}% "
                f"{'higher' if group_effect >= 0 else 'lower'}</strong> "
                f"weekend {metric_label.lower()} than stores "
                "with variable weekday opening."
            )

        else:

            consistency_text = (
                "The selected filters do not provide sufficient "
                "observations to compare both weekday operating groups."
            )


        display(
            HTML(
                f"""
                <div style="
                    margin-top:18px;
                    padding:19px 22px;
                    background:#f2eee5;
                    border-left:4px solid {GOLD};
                    border-radius:0 8px 8px 0;
                    color:#4c4538;
                    font-family:'Segoe UI';
                    font-size:13.5px;
                    line-height:1.85;
                ">

                    <div style="
                        color:{NAVY};
                        font-size:10px;
                        font-weight:700;
                        letter-spacing:1.5px;
                        margin-bottom:7px;
                    ">
                        DYNAMIC OPERATING PERFORMANCE INSIGHT
                    </div>

                    At a weekday-opening threshold of
                    <strong>{ops_threshold.value}%</strong>,
                    <strong>{all_weekday_n}</strong>
                    stores qualify as consistently open
                    Monday through Friday, representing
                    <strong>{all_weekday_share:.1f}%</strong>
                    of the analysed store network.

                    <br><br>

                    Weekend
                    {metric_label.lower()}
                    is {weekend_text}
                    than the corresponding weekday level.

                    <strong>
                        {best_day_row["Day"]}
                    </strong>
                    records the strongest average
                    {metric_label.lower()}
                    among open-store observations.

                    <br><br>

                    {consistency_text}

                    <br><br>

                    These findings describe historical
                    operating associations. Store format,
                    location, promotions, competition and
                    local demand may jointly influence both
                    operating behaviour and commercial
                    performance.

                </div>
                """
            )
        )


# ============================================================
# RESET FILTERS
# ============================================================

def reset_operations(_):

    ops_view.value = "Charts"
    ops_chart.value = "Sankey"
    ops_metric.value = "Sales"
    ops_year.value = "All"
    ops_store.value = "All"
    ops_weekend.value = "Both"
    ops_threshold.value = 95
    ops_table.value = "Stores"


ops_reset.on_click(
    reset_operations
)


# ============================================================
# INTERACTIVITY
# ============================================================

for control in [
    ops_view,
    ops_chart,
    ops_metric,
    ops_year,
    ops_store,
    ops_weekend,
    ops_threshold,
    ops_table
]:

    control.observe(
        render_operations,
        names="value"
    )


# ============================================================
# PROFESSIONAL DASHBOARD HEADING
# ============================================================

display(
    HTML(
        f"""
        <div style="
            margin:6px 0 14px;
            padding:20px 22px;
            background:linear-gradient(
                135deg,
                #0c2439,
                #173e5c
            );
            border-left:5px solid {GOLD};
            border-radius:10px;
            font-family:'Segoe UI',Arial,sans-serif;
        ">

            <div style="
                color:{LIGHT_GOLD};
                font-size:10px;
                font-weight:700;
                letter-spacing:2px;
                text-transform:uppercase;
                margin-bottom:6px;
            ">
                Store Operations · Interactive Decision Intelligence
            </div>

            <div style="
                color:{WHITE};
                font-family:Georgia,'Times New Roman',serif;
                font-size:25px;
                font-weight:500;
                margin-bottom:7px;
            ">
                Store Availability & Weekend Performance Intelligence
            </div>

            <div style="
                color:#d8e2e8;
                font-size:12.5px;
                line-height:1.65;
                max-width:1100px;
            ">
                Evaluate weekday operating reliability,
                weekend accessibility, customer traffic and
                commercial performance to understand how
                store availability relates to realised retail demand.
            </div>

        </div>
        """
    )
)


# ============================================================
# DISPLAY CONTROLS
# ============================================================

display(

    widgets.HBox(
        [
            ops_view,
            ops_metric,
            ops_year,
            ops_store,
            ops_weekend
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap",
            margin="0 0 8px 0"
        )
    ),

    widgets.HBox(
        [
            ops_threshold,
            ops_table,
            ops_reset
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap",
            margin="0 0 9px 0"
        )
    ),

    ops_chart,

    ops_out
)


render_operations()

logger.info(
    "Store operations intelligence dashboard initialized."
)
```

<h2><span style="color:lightblue">BUSINESS INTERPRETATION<b></b></span>

The analysis distinguishes between customer demand and the opportunity to
capture that demand. Stores that are closed cannot realise sales even when local
customer demand exists, making operating status an essential variable in both
exploratory analysis and subsequent forecasting.



The weekday-consistency classification provides a practical method for
identifying stores with stable Monday-to-Friday operating patterns. Comparing
their weekend performance with stores displaying more variable weekday
availability helps determine whether operating consistency is associated with
differences in weekend customer traffic, sales, or purchasing intensity.



These relationships should remain descriptive rather than causal. Store type,
location, local demand, competition, and management decisions may simultaneously
influence both opening patterns and sales performance.

Because the source data contains no hourly opening or closing timestamps,
intraday customer behaviour around the precise time of opening and closing
cannot be evaluated from this dataset. The daily operating analysis represents
the maximum defensible interpretation supported by the available information.

<h2><span style="color:lightblue">Competition Intelligence<b></b></span>

Competitive pressure may differ substantially depending on how close the
nearest competitor is to a Rossmann store. This analysis evaluates whether
competitor proximity is associated with <strong>sales, customer traffic and
sales per customer</strong>, while also examining how the relationship varies
across store types, promotional conditions and years.


Because competition distance is fundamentally a <strong>store-level
characteristic</strong>, the main relationship statistics are calculated using
store-level average performance rather than treating every daily observation as
an independent competitor-distance measurement.

Analytical limitation:
No variable identifies whether a Rossmann store or its competitor is located in
a city centre. Therefore, the city-centre-specific question in the assignment
cannot be answered directly without additional geographic information.

```python
# Dashboard colour palette
NAVY = "#17324D"
BLUE = "#2F6690"
GOLD = "#D4A72C"
LIGHT_GOLD = "#E8D39B"
CREAM = "#F7F3EA"
WHITE = "#FFFFFF"
SLATE = "#536471"

comp = eda_df[
    eda_df["Open"].eq(1)
    & eda_df["Customers"].gt(0)
    & eda_df["CompetitionDistance"].notna()
    & eda_df["CompetitionDistance"].gt(0)
].copy()

missing_col = next(
    (c for c in [
        "CompetitionDistanceMissing",
        "CompetitionDistance_Missing",
        "CompetitionDistance_missing"
    ] if c in comp.columns),
    None
)

if missing_col:
    comp = comp[~comp[missing_col].astype(bool)]

comp["CompetitionDistanceKm"] = comp["CompetitionDistance"] / 1000
comp["SalesPerCustomer"] = comp["Sales"] / comp["Customers"]


# ============================================================
# CONTROLS
# ============================================================

comp_view = widgets.ToggleButtons(
    options=[("📊 Charts", "Charts"), ("📋 Table", "Table")],
    value="Charts"
)

comp_chart = widgets.ToggleButtons(
    options=[
        ("🎯 Store Landscape", "Scatter"),
        ("📈 Distance Response", "Curve")
    ],
    value="Scatter"
)

comp_metric = widgets.Dropdown(
    options={
        "Sales": "AvgSales",
        "Customers": "AvgCustomers",
        "Sales / Customer": "SalesPerCustomer"
    },
    value="AvgSales",
    description="Metric:"
)

comp_year = widgets.Dropdown(
    options=["All"] + sorted(comp["Year"].dropna().unique()),
    value="All",
    description="Year:"
)

comp_store = widgets.Dropdown(
    options=["All"] + sorted(comp["StoreType"].dropna().unique()),
    value="All",
    description="Store:"
)

comp_promo = widgets.Dropdown(
    options=[
        ("All Days", "All"),
        ("Promo Days", 1),
        ("Non-Promo Days", 0)
    ],
    value="All",
    description="Promo:"
)

comp_min_days = widgets.IntSlider(
    value=30,
    min=10,
    max=90,
    step=10,
    description="Min Days:",
    continuous_update=False
)

comp_reset = widgets.Button(
    description="Reset Filters",
    icon="refresh"
)

comp_out = widgets.Output()


# ============================================================
# DASHBOARD
# ============================================================

def render_competition(*_):

    with comp_out:
        clear_output(wait=True)

        d = comp.copy()

        if comp_year.value != "All":
            d = d[d["Year"].eq(comp_year.value)]

        if comp_store.value != "All":
            d = d[d["StoreType"].eq(comp_store.value)]

        if comp_promo.value != "All":
            d = d[d["Promo"].eq(comp_promo.value)]


        # ----------------------------------------------------
        # STORE-LEVEL PERFORMANCE
        # ----------------------------------------------------

        stores = (
            d.groupby(
                ["Store", "StoreType", "CompetitionDistanceKm"],
                observed=True
            )
            .agg(
                StoreDays=("Sales", "size"),
                AvgSales=("Sales", "mean"),
                AvgCustomers=("Customers", "mean"),
                TotalSales=("Sales", "sum"),
                TotalCustomers=("Customers", "sum")
            )
            .reset_index()
        )

        stores = stores[
            stores["StoreDays"] >= comp_min_days.value
        ].copy()

        if len(stores) < 5:
            display(HTML(
                f"""
                <div style="
                    padding:15px;
                    background:{CREAM};
                    color:{SLATE};
                    border-left:4px solid {GOLD};
                ">
                    Insufficient stores for the selected filters.
                </div>
                """
            ))
            return

        stores["SalesPerCustomer"] = (
            stores["TotalSales"] / stores["TotalCustomers"]
        )

        metric = comp_metric.value
        metric_label = comp_metric.label


        # ----------------------------------------------------
        # DISTANCE BANDS
        # ----------------------------------------------------

        stores["DistanceBand"] = pd.cut(
            stores["CompetitionDistanceKm"],
            bins=[0, .5, 1, 2, 5, 10, np.inf],
            labels=[
                "< 0.5 km",
                "0.5–1 km",
                "1–2 km",
                "2–5 km",
                "5–10 km",
                "10+ km"
            ],
            right=False,
            include_lowest=True
        )

        bands = (
            stores.groupby(
                "DistanceBand",
                observed=True
            )
            .agg(
                Stores=("Store", "size"),
                MedianDistance=("CompetitionDistanceKm", "median"),
                AvgSales=("AvgSales", "mean"),
                AvgCustomers=("AvgCustomers", "mean"),
                SalesPerCustomer=("SalesPerCustomer", "mean")
            )
            .reset_index()
            .sort_values("MedianDistance")
        )


        # ----------------------------------------------------
        # RELATIONSHIP METRICS
        # ----------------------------------------------------

        pearson = stores["CompetitionDistanceKm"].corr(
            stores[metric]
        )

        spearman = stores["CompetitionDistanceKm"].corr(
            stores[metric],
            method="spearman"
        )

        best = bands.loc[bands[metric].idxmax()]

        nearest = bands.iloc[0]
        farthest = bands.iloc[-1]

        far_effect = (
            farthest[metric] / nearest[metric] - 1
        ) * 100

        close_share = (
            stores["CompetitionDistanceKm"].le(1).mean()
            * 100
        )


        # ====================================================
        # KPI CARDS
        # ============================================================

        cards = [
            (
                "STORES ANALYSED",
                f"{len(stores):,}",
                f"≥ {comp_min_days.value} trading days"
            ),
            (
                "MEDIAN DISTANCE",
                f"{stores['CompetitionDistanceKm'].median():.2f} km",
                "Nearest competitor"
            ),
            (
                "COMPETITOR ≤ 1 KM",
                f"{close_share:.1f}%",
                "Share of stores"
            ),
            (
                "BEST DISTANCE BAND",
                str(best["DistanceBand"]),
                metric_label
            ),
            (
                "PEARSON r",
                f"{pearson:+.3f}",
                "Linear relationship"
            ),
            (
                "SPEARMAN ρ",
                f"{spearman:+.3f}",
                "Rank relationship"
            ),
            (
                "FAR vs NEAR",
                f"{far_effect:+.1f}%",
                metric_label
            ),
            (
                "MAX DISTANCE",
                f"{stores['CompetitionDistanceKm'].max():.1f} km",
                "Observed range"
            )
        ]

        display(HTML(
            "<div style='"
            "display:grid;"
            "grid-template-columns:repeat(4,minmax(0,1fr));"
            "gap:10px;"
            "margin:8px 0 18px;'>"

            + "".join(
                f"""
                <div style="
                    background:{WHITE};
                    border:1px solid {LIGHT_GOLD};
                    border-top:4px solid {GOLD};
                    border-radius:9px;
                    padding:13px 15px;
                    min-height:94px;
                ">

                    <div style="
                        color:{SLATE};
                        font-size:8.5px;
                        font-weight:700;
                        letter-spacing:1px;
                    ">
                        {title}
                    </div>

                    <div style="
                        color:{NAVY};
                        font-family:Georgia;
                        font-size:20px;
                        font-weight:600;
                        margin-top:7px;
                    ">
                        {value}
                    </div>

                    <div style="
                        color:{BLUE};
                        font-size:9.5px;
                        margin-top:5px;
                    ">
                        {subtitle}
                    </div>

                </div>
                """
                for title, value, subtitle in cards
            )

            + "</div>"
        ))


        # ====================================================
        # CURRENT VIEW
        # ============================================================

        display(HTML(
            f"""
            <div style="
                padding:11px 15px;
                margin-bottom:15px;
                background:{CREAM};
                border-left:4px solid {GOLD};
                color:{SLATE};
                font-size:11.5px;
            ">

                <strong style="color:{NAVY};">
                    Current View
                </strong>

                &nbsp; · &nbsp;

                Metric:
                <strong>{metric_label}</strong>

                &nbsp; · &nbsp;

                Store Type:
                <strong>{comp_store.value}</strong>

                &nbsp; · &nbsp;

                Promo:
                <strong>{comp_promo.label}</strong>

                &nbsp; · &nbsp;

                Year:
                <strong>{comp_year.value}</strong>

            </div>
            """
        ))


        # ====================================================
        # TABLE
        # ============================================================

        if comp_view.value == "Table":

            table = bands[
                [
                    "DistanceBand",
                    "Stores",
                    "MedianDistance",
                    "AvgSales",
                    "AvgCustomers",
                    "SalesPerCustomer"
                ]
            ].reset_index(drop=True)

            display(
                table.style
                .format({
                    "Stores": "{:,.0f}",
                    "MedianDistance": "{:.2f} km",
                    "AvgSales": "{:,.2f}",
                    "AvgCustomers": "{:,.2f}",
                    "SalesPerCustomer": "{:.2f}"
                })
                .set_properties(**{
                    "text-align": "center",
                    "padding": "8px",
                    "color": SLATE
                })
                .set_table_styles([
                    {
                        "selector": "th",
                        "props": [
                            ("background-color", NAVY),
                            ("color", WHITE),
                            ("font-weight", "bold"),
                            ("text-align", "center"),
                            ("padding", "9px")
                        ]
                    },
                    {
                        "selector": "td",
                        "props": [
                            ("border-bottom", f"1px solid {CREAM}")
                        ]
                    }
                ])
            )


        # ====================================================
        # STORE LANDSCAPE
        # ============================================================

        elif comp_chart.value == "Scatter":

            fig = px.scatter(
                stores,
                x="CompetitionDistanceKm",
                y=metric,
                size="StoreDays",
                color="StoreType",

                color_discrete_sequence=[
                    NAVY,
                    BLUE,
                    GOLD,
                    LIGHT_GOLD
                ],

                hover_data={
                    "Store": True,
                    "StoreType": True,
                    "CompetitionDistanceKm": ":.2f",
                    "StoreDays": True,
                    "AvgSales": ":,.0f",
                    "AvgCustomers": ":,.0f",
                    "SalesPerCustomer": ":.2f"
                },

                size_max=32
            )

            fig.update_layout(
                title={
                    "text":
                        f"Competitor Proximity vs {metric_label}"
                        "<br><sup>"
                        "Each bubble represents one store · "
                        "Bubble size represents available trading history"
                        "</sup>",
                    "x": .02
                },

                xaxis_title="Nearest Competitor Distance (km)",
                yaxis_title=metric_label,

                height=570,

                paper_bgcolor=WHITE,
                plot_bgcolor=WHITE,

                font=dict(
                    color=SLATE
                ),

                legend_title="Store Type"
            )

            fig.update_xaxes(
                gridcolor=CREAM,
                linecolor=SLATE
            )

            fig.update_yaxes(
                gridcolor=CREAM,
                linecolor=SLATE
            )

            fig.show()


        # ====================================================
        # DISTANCE RESPONSE CURVE
        # ============================================================

        else:

            q = min(
                10,
                stores["CompetitionDistanceKm"].nunique()
            )

            curve = stores.copy()

            curve["Segment"] = pd.qcut(
                curve["CompetitionDistanceKm"],
                q=q,
                duplicates="drop"
            )

            curve = (
                curve.groupby(
                    "Segment",
                    observed=True
                )
                .agg(
                    Distance=("CompetitionDistanceKm", "median"),
                    Median=(metric, "median"),
                    Q25=(metric, lambda x: x.quantile(.25)),
                    Q75=(metric, lambda x: x.quantile(.75)),
                    Stores=("Store", "size")
                )
                .reset_index()
                .sort_values("Distance")
            )

            fig = go.Figure()

            fig.add_trace(
                go.Scatter(
                    x=curve["Distance"],
                    y=curve["Q75"],
                    mode="lines",
                    line=dict(
                        width=0,
                        color=LIGHT_GOLD
                    ),
                    showlegend=False,
                    hoverinfo="skip"
                )
            )

            fig.add_trace(
                go.Scatter(
                    x=curve["Distance"],
                    y=curve["Q25"],
                    mode="lines",
                    fill="tonexty",
                    fillcolor=CREAM,
                    line=dict(
                        color=LIGHT_GOLD
                    ),
                    name="Middle 50%",
                    hoverinfo="skip"
                )
            )

            fig.add_trace(
                go.Scatter(
                    x=curve["Distance"],
                    y=curve["Median"],
                    mode="lines+markers",

                    line=dict(
                        color=NAVY,
                        width=3
                    ),

                    marker=dict(
                        color=GOLD,
                        size=9,
                        line=dict(
                            color=BLUE,
                            width=1
                        )
                    ),

                    name=f"Median {metric_label}",

                    customdata=curve[["Stores"]],

                    hovertemplate=(
                        "Distance: %{x:.2f} km"
                        "<br>Median: %{y:,.2f}"
                        "<br>Stores: %{customdata[0]}"
                        "<extra></extra>"
                    )
                )
            )

            fig.update_layout(
                title={
                    "text":
                        "Competition Distance Response Curve"
                        "<br><sup>"
                        f"Median {metric_label.lower()} with "
                        "middle 50% store variation"
                        "</sup>",
                    "x": .02
                },

                xaxis_title="Nearest Competitor Distance (km)",
                yaxis_title=metric_label,

                height=570,

                hovermode="x unified",

                paper_bgcolor=WHITE,
                plot_bgcolor=WHITE,

                font=dict(
                    color=SLATE
                )
            )

            fig.update_xaxes(
                gridcolor=CREAM,
                linecolor=SLATE
            )

            fig.update_yaxes(
                gridcolor=CREAM,
                linecolor=SLATE
            )

            fig.show()


        # ====================================================
        # DYNAMIC INSIGHT
        # ============================================================

        strength = (
            "strong"
            if abs(spearman) >= .60
            else "moderate"
            if abs(spearman) >= .30
            else "weak"
        )

        if spearman > 0:
            direction = (
                "performance tends to increase as "
                "competitor distance increases"
            )

        elif spearman < 0:
            direction = (
                "higher performance tends to occur "
                "where competitors are closer"
            )

        else:
            direction = "no directional pattern is observed"

        display(HTML(
            f"""
            <div style="
                margin-top:17px;
                padding:18px 21px;
                background:{CREAM};
                border-left:4px solid {GOLD};
                border-radius:0 8px 8px 0;
                color:{SLATE};
                line-height:1.8;
                font-size:13px;
            ">

                <div style="
                    color:{NAVY};
                    font-size:10px;
                    font-weight:700;
                    letter-spacing:1.4px;
                    margin-bottom:6px;
                ">
                    DYNAMIC COMPETITION INSIGHT
                </div>

                Across <strong>{len(stores)}</strong> eligible stores,
                competitor distance shows a
                <strong>{strength}</strong> relationship with
                {metric_label.lower()}.

                Spearman correlation is
                <strong>{spearman:+.3f}</strong>, indicating that
                <strong>{direction}</strong>.

                <br><br>

                The strongest performance is observed in the
                <strong>{best['DistanceBand']}</strong> band.

                Stores in the farthest available competition band record
                <strong>
                    {abs(far_effect):.1f}%
                    {'higher' if far_effect >= 0 else 'lower'}
                </strong>
                {metric_label.lower()} than stores in the nearest band.

                <br><br>

                The relationship is observational rather than causal;
                competitor proximity may also reflect differences in
                location, commercial density, store type and local demand.

            </div>
            """
        ))


# ============================================================
# RESET + INTERACTIVITY
# ============================================================

def reset_competition(_):
    comp_view.value = "Charts"
    comp_chart.value = "Scatter"
    comp_metric.value = "AvgSales"
    comp_year.value = "All"
    comp_store.value = "All"
    comp_promo.value = "All"
    comp_min_days.value = 30


comp_reset.on_click(reset_competition)

for control in [
    comp_view,
    comp_chart,
    comp_metric,
    comp_year,
    comp_store,
    comp_promo,
    comp_min_days
]:
    control.observe(
        render_competition,
        names="value"
    )


# ============================================================
# DASHBOARD HEADING
# ============================================================

display(HTML(
    f"""
    <div style="
        padding:20px 23px;
        margin:6px 0 13px;
        background:linear-gradient(
            135deg,
            {NAVY},
            {BLUE}
        );
        border-left:5px solid {GOLD};
        border-radius:10px;
        font-family:'Segoe UI',Arial,sans-serif;
    ">

        <div style="
            color:{LIGHT_GOLD};
            font-size:10px;
            font-weight:700;
            letter-spacing:2px;
            margin-bottom:6px;
        ">
            COMPETITIVE LANDSCAPE · INTERACTIVE MARKET INTELLIGENCE
        </div>

        <div style="
            color:{WHITE};
            font-family:Georgia,'Times New Roman',serif;
            font-size:25px;
            font-weight:500;
            margin-bottom:7px;
        ">
            Competitor Proximity & Store Performance Intelligence
        </div>

        <div style="
            color:{CREAM};
            font-size:12.5px;
            line-height:1.6;
        ">
            Explore how distance to the nearest competitor relates to
            sales, customer traffic and purchasing value across the
            store network.
        </div>

    </div>
    """
))


# ============================================================
# DISPLAY
# ============================================================

display(
    widgets.HBox(
        [
            comp_view,
            comp_metric,
            comp_year,
            comp_store,
            comp_promo
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap"
        )
    ),

    widgets.HBox(
        [
            comp_min_days,
            comp_reset
        ],
        layout=widgets.Layout(
            gap="10px",
            margin="8px 0"
        )
    ),

    comp_chart,
    comp_out
)

render_competition()

logger.info(
    "Competition distance intelligence dashboard initialized."
)
```

<h2><span style="color:lightblue">BUSINESS INTERPRETATION<b></b></span>

The competition-distance analysis evaluates whether store performance changes
systematically as the nearest competitor becomes more distant. Using
store-level averages is particularly important because
<strong>CompetitionDistance is a store-level characteristic</strong>; treating
every daily observation as an independent distance measurement would
artificially exaggerate the amount of evidence available.

<h2><span style="color:lightblue">COMPETITOR ENTRY INTELLIGENCE<b></b></span>

This event-based analysis compares store performance before and after the
recorded opening period of the nearest competitor. Sales, customer traffic and
sales per customer are examined separately to determine whether competitive
entry is associated with changes in demand volume or purchasing intensity.



<strong style="color:#0C2492;">Important:</strong>
Competition opening information is available only at month-and-year level.
Therefore, the analysis treats the first day of the recorded month as an
approximate event anchor and excludes the event month from the direct
before-versus-after comparison.

```python
entry = eda_df[
    eda_df["Open"].eq(1) &
    eda_df["Customers"].gt(0)
].copy()

entry["SalesPerCustomer"] = entry["Sales"] / entry["Customers"]

entry["CompetitionOpenDate"] = pd.to_datetime(
    {
        "year": pd.to_numeric(
            entry["CompetitionOpenSinceYear"],
            errors="coerce"
        ),
        "month": pd.to_numeric(
            entry["CompetitionOpenSinceMonth"],
            errors="coerce"
        ),
        "day": 1
    },
    errors="coerce"
)

entry = entry[
    entry["CompetitionOpenDate"].notna()
].copy()

entry["RelativeMonth"] = (
    (entry["Date"].dt.year - entry["CompetitionOpenDate"].dt.year) * 12
    + entry["Date"].dt.month
    - entry["CompetitionOpenDate"].dt.month
)


# ============================================================
# CONTROLS
# ============================================================

entry_view = widgets.ToggleButtons(
    options=[("📊 Charts","Charts"),("📋 Table","Table")],
    value="Charts"
)

entry_chart = widgets.ToggleButtons(
    options=[
        ("📈 Event Response","Trend"),
        ("🎯 Store Impact","Impact")
    ],
    value="Trend"
)

entry_metric = widgets.Dropdown(
    options={
        "Sales":"Sales",
        "Customers":"Customers",
        "Sales / Customer":"SalesPerCustomer"
    },
    value="Sales",
    description="Metric:"
)

entry_store = widgets.Dropdown(
    options=["All"] + sorted(
        entry["StoreType"].dropna().unique()
    ),
    value="All",
    description="Store:"
)

entry_promo = widgets.Dropdown(
    options=[
        ("All Days","All"),
        ("Promo Days",1),
        ("Non-Promo Days",0)
    ],
    value="All",
    description="Promo:"
)

entry_window = widgets.Dropdown(
    options={
        "± 3 Months":3,
        "± 6 Months":6,
        "± 12 Months":12
    },
    value=6,
    description="Window:"
)

entry_min = widgets.IntSlider(
    value=20,
    min=10,
    max=60,
    step=10,
    description="Min Days:",
    continuous_update=False
)

entry_reset = widgets.Button(
    description="Reset",
    icon="refresh"
)

entry_out = widgets.Output()


# ============================================================
# DASHBOARD
# ============================================================

def render_entry(*_):

    with entry_out:
        clear_output(wait=True)

        d = entry.copy()

        if entry_store.value != "All":
            d = d[d["StoreType"].eq(entry_store.value)]

        if entry_promo.value != "All":
            d = d[d["Promo"].eq(entry_promo.value)]

        m = entry_metric.value
        label = entry_metric.label
        window = entry_window.value

        d = d[
            d["RelativeMonth"].between(-window, window)
        ].copy()

        d["Period"] = "Event Month"
        d.loc[d["RelativeMonth"] < 0, "Period"] = "Before"
        d.loc[d["RelativeMonth"] > 0, "Period"] = "After"


        # ----------------------------------------------------
        # ELIGIBLE STORES
        # ----------------------------------------------------

        evidence = (
            d[d["Period"].isin(["Before","After"])]
            .groupby(["Store","Period"])
            .size()
            .unstack(fill_value=0)
        )

        before_n = evidence.get(
            "Before",
            pd.Series(0, index=evidence.index)
        )

        after_n = evidence.get(
            "After",
            pd.Series(0, index=evidence.index)
        )

        eligible = evidence[
            (before_n >= entry_min.value) &
            (after_n >= entry_min.value)
        ].index

        d = d[d["Store"].isin(eligible)]

        if d["Store"].nunique() < 3:
            display(HTML(
                f"""
                <div style="
                    padding:14px;
                    background:{CREAM};
                    border-left:4px solid {GOLD};
                    color:{SLATE};
                ">
                    Insufficient stores meet the selected
                    pre/post evidence requirement.
                </div>
                """
            ))
            return


        # ----------------------------------------------------
        # STORE PRE / POST EFFECT
        # ----------------------------------------------------

        effect = (
            d[d["Period"].isin(["Before","After"])]
            .groupby(["Store","StoreType","Period"])[m]
            .mean()
            .unstack()
            .dropna()
            .reset_index()
        )

        effect["ChangePct"] = (
            effect["After"] / effect["Before"] - 1
        ) * 100

        pre = effect["Before"].mean()
        post = effect["After"].mean()

        overall = (post / pre - 1) * 100
        median_effect = effect["ChangePct"].median()
        positive = effect["ChangePct"].gt(0).mean() * 100

        strongest = effect.loc[
            effect["ChangePct"].idxmax()
        ]

        weakest = effect.loc[
            effect["ChangePct"].idxmin()
        ]


        # ----------------------------------------------------
        # RELATIVE-MONTH CURVE
        # ----------------------------------------------------

        curve = (
            d.groupby(["Store","RelativeMonth"])[m]
            .mean()
            .reset_index()
            .groupby("RelativeMonth")[m]
            .agg(
                Median="median",
                Q25=lambda x:x.quantile(.25),
                Q75=lambda x:x.quantile(.75),
                Stores="size"
            )
            .reset_index()
        )


        # ====================================================
        # KPIs
        # ============================================================

        cards = [
            ("ELIGIBLE STORES",f"{len(effect):,}","Pre + post evidence"),
            ("BEFORE",f"{pre:,.1f}",label),
            ("AFTER",f"{post:,.1f}",label),
            ("POST CHANGE",f"{overall:+.1f}%","After vs before"),
            ("POSITIVE STORES",f"{positive:.1f}%","Improved"),
            ("MEDIAN EFFECT",f"{median_effect:+.1f}%","Store-level"),
            ("MOST RESILIENT",f"Store {int(strongest['Store'])}",f"{strongest['ChangePct']:+.1f}%"),
            ("LARGEST DECLINE",f"Store {int(weakest['Store'])}",f"{weakest['ChangePct']:+.1f}%")
        ]

        display(HTML(
            "<div style='display:grid;grid-template-columns:repeat(4,1fr);"
            "gap:10px;margin:8px 0 18px'>"
            + "".join(
                f"""
                <div style="
                    background:{WHITE};
                    border:1px solid {LIGHT_GOLD};
                    border-top:4px solid {GOLD};
                    border-radius:9px;
                    padding:13px 15px;
                ">
                    <div style="
                        color:{SLATE};
                        font-size:8px;
                        font-weight:700;
                        letter-spacing:1px;
                    ">{a}</div>

                    <div style="
                        color:{NAVY};
                        font:600 20px Georgia;
                        margin-top:6px;
                    ">{b}</div>

                    <div style="
                        color:{BLUE};
                        font-size:9px;
                        margin-top:4px;
                    ">{c}</div>
                </div>
                """
                for a,b,c in cards
            )
            + "</div>"
        ))


        # ====================================================
        # TABLE
        # ============================================================

        if entry_view.value == "Table":

            table = (
                effect
                .sort_values(
                    "ChangePct",
                    ascending=False
                )
                .reset_index(drop=True)
            )

            display(
                table.style
                .format({
                    "Before":"{:,.2f}",
                    "After":"{:,.2f}",
                    "ChangePct":"{:+.2f}%"
                })
                .set_properties(**{
                    "text-align":"center",
                    "padding":"8px",
                    "color":SLATE
                })
                .set_table_styles([{
                    "selector":"th",
                    "props":[
                        ("background-color",NAVY),
                        ("color",WHITE),
                        ("font-weight","bold"),
                        ("text-align","center")
                    ]
                }])
            )


        # ====================================================
        # EVENT RESPONSE
        # ============================================================

        elif entry_chart.value == "Trend":

            fig = go.Figure()

            fig.add_trace(go.Scatter(
                x=curve["RelativeMonth"],
                y=curve["Q75"],
                mode="lines",
                line=dict(width=0),
                showlegend=False,
                hoverinfo="skip"
            ))

            fig.add_trace(go.Scatter(
                x=curve["RelativeMonth"],
                y=curve["Q25"],
                fill="tonexty",
                mode="lines",
                fillcolor=CREAM,
                line=dict(color=LIGHT_GOLD),
                name="Middle 50%",
                hoverinfo="skip"
            ))

            fig.add_trace(go.Scatter(
                x=curve["RelativeMonth"],
                y=curve["Median"],
                mode="lines+markers",
                line=dict(
                    color=NAVY,
                    width=3
                ),
                marker=dict(
                    color=GOLD,
                    size=8,
                    line=dict(
                        color=BLUE,
                        width=1
                    )
                ),
                name=f"Median {label}",
                customdata=curve[["Stores"]],
                hovertemplate=(
                    "Relative month: %{x}"
                    "<br>Median: %{y:,.2f}"
                    "<br>Stores: %{customdata[0]}"
                    "<extra></extra>"
                )
            ))

            fig.add_vline(
                x=0,
                line_dash="dash",
                line_color=GOLD
            )

            fig.update_layout(
                title={
                    "text":
                    "Performance Around Nearby Competitor Entry"
                    "<br><sup>"
                    "Month 0 represents the recorded competitor opening month"
                    "</sup>",
                    "x":.02
                },
                xaxis_title="Months Relative to Competitor Opening",
                yaxis_title=label,
                hovermode="x unified",
                height=570,
                paper_bgcolor=WHITE,
                plot_bgcolor=WHITE,
                font=dict(color=SLATE)
            )

            fig.update_xaxes(gridcolor=CREAM)
            fig.update_yaxes(gridcolor=CREAM)

            fig.show()


        # ====================================================
        # STORE IMPACT
        # ============================================================

        else:

            effect["Response"] = np.where(
                effect["ChangePct"] >= 0,
                "Improved",
                "Declined"
            )

            fig = px.scatter(
                effect,
                x="Before",
                y="After",
                size=effect["ChangePct"].abs() + 2,
                color="Response",

                color_discrete_map={
                    "Improved":BLUE,
                    "Declined":GOLD
                },

                hover_data={
                    "Store":True,
                    "StoreType":True,
                    "ChangePct":":+.1f"
                },

                size_max=30
            )

            low = min(
                effect["Before"].min(),
                effect["After"].min()
            )

            high = max(
                effect["Before"].max(),
                effect["After"].max()
            )

            fig.add_shape(
                type="line",
                x0=low,
                y0=low,
                x1=high,
                y1=high,
                line=dict(
                    color=NAVY,
                    dash="dash"
                )
            )

            fig.update_layout(
                title={
                    "text":
                    "Store-Level Response to Competitor Entry"
                    "<br><sup>"
                    "Stores above the diagonal performed better after entry"
                    "</sup>",
                    "x":.02
                },
                xaxis_title=f"Before · {label}",
                yaxis_title=f"After · {label}",
                height=570,
                paper_bgcolor=WHITE,
                plot_bgcolor=WHITE,
                font=dict(color=SLATE)
            )

            fig.update_xaxes(gridcolor=CREAM)
            fig.update_yaxes(gridcolor=CREAM)

            fig.show()


        # ====================================================
        # DYNAMIC INSIGHT
        # ============================================================

        display(HTML(
            f"""
            <div style="
                margin-top:17px;
                padding:18px 21px;
                background:{CREAM};
                border-left:4px solid {GOLD};
                border-radius:0 8px 8px 0;
                color:{SLATE};
                line-height:1.8;
                font-size:13px;
            ">

                <div style="
                    color:{NAVY};
                    font-size:10px;
                    font-weight:700;
                    letter-spacing:1.4px;
                    margin-bottom:6px;
                ">
                    DYNAMIC COMPETITOR ENTRY INSIGHT
                </div>

                Across <strong>{len(effect)}</strong> eligible stores,
                average {label.lower()}
                <strong>
                    {'increased' if overall >= 0 else 'declined'}
                    by {abs(overall):.1f}%
                </strong>
                following the recorded competitor opening period.

                <br><br>

                The median store-level effect is
                <strong>{median_effect:+.1f}%</strong>, while
                <strong>{positive:.1f}%</strong> of stores record better
                performance after competitor entry.

                The result is observational rather than causal and may also
                reflect seasonality, promotions and wider market conditions.

            </div>
            """
        ))


# ============================================================
# RESET + INTERACTIVITY
# ============================================================

def reset_entry(_):
    entry_view.value = "Charts"
    entry_chart.value = "Trend"
    entry_metric.value = "Sales"
    entry_store.value = "All"
    entry_promo.value = "All"
    entry_window.value = 6
    entry_min.value = 20


entry_reset.on_click(reset_entry)

for w in [
    entry_view,
    entry_chart,
    entry_metric,
    entry_store,
    entry_promo,
    entry_window,
    entry_min
]:
    w.observe(render_entry, names="value")


# ============================================================
# DASHBOARD HEADING
# ============================================================

display(HTML(
    f"""
    <div style="
        padding:20px 23px;
        margin:6px 0 13px;
        background:linear-gradient(135deg,{NAVY},{BLUE});
        border-left:5px solid {GOLD};
        border-radius:10px;
    ">

        <div style="
            color:{LIGHT_GOLD};
            font-size:10px;
            font-weight:700;
            letter-spacing:2px;
        ">
            COMPETITOR ENTRY · INTERACTIVE EVENT INTELLIGENCE
        </div>

        <div style="
            color:{WHITE};
            font:500 25px Georgia;
            margin:6px 0;
        ">
            Market Entry Impact & Store Resilience
        </div>

        <div style="
            color:{CREAM};
            font-size:12.5px;
        ">
            Compare performance before and after nearby competitor entry
            and identify stores displaying stronger competitive resilience.
        </div>

    </div>
    """
))


display(
    widgets.HBox(
        [
            entry_view,
            entry_metric,
            entry_store,
            entry_promo,
            entry_window
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap"
        )
    ),

    widgets.HBox(
        [entry_min, entry_reset],
        layout=widgets.Layout(
            gap="10px",
            margin="8px 0"
        )
    ),

    entry_chart,
    entry_out
)

render_entry()

logger.info(
    "Competitor entry event dashboard initialized."
)
```

<h2><span style="color:lightblue">BUSINESS INTERPRETATION<b></b></span>

The event analysis compares store performance before and after the recorded
opening period of the nearest competitor. Evaluating
<strong>Sales, Customers and Sales per Customer</strong> separately helps
distinguish whether any post-entry change is associated mainly with customer
traffic, purchasing intensity, or a combination of both.



The store-level comparison is especially important because competitive entry
does not necessarily affect every location in the same way. Some stores may
experience weaker sales or customer traffic after competitor entry, while others
may remain stable or even improve. Such differences indicate varying levels of
<strong>competitive resilience</strong> across the Rossmann network.

<h2><span style="color:lightblue">COMPETITION DATA AVAILABILITY<b></b></span>

The assignment asks whether stores with initially missing competition distance
later receive valid distance information. This requires checking whether
<code>CompetitionDistance</code> changes over time for the same store.



In the supplied data structure, competition distance is a
<strong>store-level attribute</strong> rather than a daily measurement.
Consequently, after store information is merged with the daily sales data,
the same distance value is repeated across dates for each store. A genuine
missing-to-known transition therefore cannot be reconstructed unless historical
versions of the store metadata are available.

```python
missing_col = next(
    (c for c in [
        "CompetitionDistanceMissing",
        "CompetitionDistance_Missing",
        "CompetitionDistance_missing"
    ] if c in eda_df.columns),
    None
)

cols = ["Store", "Date", "StoreType", "CompetitionDistance"]
if missing_col:
    cols.append(missing_col)

audit = eda_df[cols].sort_values(["Store", "Date"]).copy()

audit["OriginalMissing"] = (
    audit[missing_col].fillna(0).astype(bool)
    if missing_col
    else audit["CompetitionDistance"].isna()
)


# ============================================================
# STORE-LEVEL AUDIT
# ============================================================

store_audit = (
    audit.groupby("Store", observed=True)
    .agg(
        StoreType=("StoreType", "first"),
        CompetitionDistance=("CompetitionDistance", "last"),
        FirstMissing=("OriginalMissing", "first"),
        LastMissing=("OriginalMissing", "last"),
        DistanceValues=("CompetitionDistance",
                        lambda x: x.nunique(dropna=False))
    )
    .reset_index()
)

store_audit["CompetitionDistanceKm"] = (
    store_audit["CompetitionDistance"] / 1000
)

store_audit["DistanceChanged"] = (
    store_audit["DistanceValues"] > 1
)

store_audit["MissingToKnown"] = (
    store_audit["FirstMissing"] &
    ~store_audit["LastMissing"]
)

store_audit["AvailabilityStatus"] = np.where(
    store_audit["FirstMissing"],
    "Originally Missing / Imputed",
    "Originally Observed"
)

store_audit["Consistency"] = np.where(
    store_audit["DistanceChanged"],
    "Distance Changed",
    "Static Distance"
)

store_audit["Transition"] = np.where(
    store_audit["MissingToKnown"],
    "Missing → Known",
    "No Transition"
)

store_audit["DistanceBand"] = pd.cut(
    store_audit["CompetitionDistanceKm"],
    bins=[0, .5, 1, 2, 5, 10, np.inf],
    labels=[
        "<0.5 km", "0.5–1 km", "1–2 km",
        "2–5 km", "5–10 km", "10+ km"
    ],
    right=False
)


# ============================================================
# CONTROLS
# ============================================================

audit_view = widgets.ToggleButtons(
    options=[
        ("📊 Intelligence", "Charts"),
        ("📋 Store Detail", "Table")
    ],
    value="Charts"
)

audit_chart = widgets.ToggleButtons(
    options=[
        ("🧩 Availability Architecture", "Icicle"),
        ("🔀 Data Lineage Flow", "Parcats")
    ],
    value="Icicle"
)

audit_store = widgets.Dropdown(
    options=["All"] + sorted(
        store_audit["StoreType"].dropna().unique()
    ),
    value="All",
    description="Store:"
)

audit_status = widgets.Dropdown(
    options=["All"] + sorted(
        store_audit["AvailabilityStatus"].unique()
    ),
    value="All",
    description="Status:"
)

audit_reset = widgets.Button(
    description="Reset",
    icon="refresh"
)

audit_out = widgets.Output()


# ============================================================
# DASHBOARD
# ============================================================

def render_audit(*_):

    with audit_out:
        clear_output(wait=True)

        d = store_audit.copy()

        if audit_store.value != "All":
            d = d[d["StoreType"].eq(audit_store.value)]

        if audit_status.value != "All":
            d = d[
                d["AvailabilityStatus"].eq(
                    audit_status.value
                )
            ]

        if d.empty:
            display(HTML(
                f"""
                <div style="
                    padding:15px;
                    background:{CREAM};
                    border-left:4px solid {GOLD};
                    color:{SLATE};
                ">
                    No stores match the selected audit filters.
                </div>
                """
            ))
            return


        # ====================================================
        # KPI CALCULATIONS
        # ============================================================

        total = len(d)
        missing = int(d["FirstMissing"].sum())
        observed = total - missing
        transitions = int(d["MissingToKnown"].sum())
        changed = int(d["DistanceChanged"].sum())

        missing_share = (
            missing / total * 100
            if total else 0
        )

        observed_share = (
            observed / total * 100
            if total else 0
        )

        static_share = (
            (~d["DistanceChanged"]).mean() * 100
        )

        usable_share = (
            d["CompetitionDistance"].notna().mean() * 100
        )

        median_distance = (
            d["CompetitionDistanceKm"].median()
        )

        nearest_share = (
            d["CompetitionDistanceKm"].le(1).mean() * 100
        )

        missing_by_type = (
            d.groupby("StoreType", observed=True)["FirstMissing"]
            .mean()
            .mul(100)
        )

        highest_missing_type = (
            missing_by_type.idxmax()
            if not missing_by_type.empty
            else "—"
        )

        highest_missing_rate = (
            missing_by_type.max()
            if not missing_by_type.empty
            else np.nan
        )


        # ====================================================
        # KPI CARDS
        # ============================================================

        cards = [
            (
                "STORES AUDITED",
                f"{total:,}",
                "Unique stores"
            ),
            (
                "ORIGINALLY OBSERVED",
                f"{observed:,}",
                f"{observed_share:.1f}% of stores"
            ),
            (
                "ORIGINALLY MISSING",
                f"{missing:,}",
                f"{missing_share:.1f}% of stores"
            ),
            (
                "USABLE AFTER PREP",
                f"{usable_share:.1f}%",
                "Distance available"
            ),
            (
                "STATIC DISTANCE",
                f"{static_share:.1f}%",
                "No historical change"
            ),
            (
                "MISSING → KNOWN",
                f"{transitions:,}",
                "Observed transitions"
            ),
            (
                "MEDIAN DISTANCE",
                (
                    f"{median_distance:.2f} km"
                    if pd.notna(median_distance)
                    else "—"
                ),
                "Nearest competitor"
            ),
            (
                "COMPETITOR ≤ 1 KM",
                f"{nearest_share:.1f}%",
                "Store share"
            )
        ]

        display(HTML(
            "<div style='"
            "display:grid;"
            "grid-template-columns:repeat(4,minmax(0,1fr));"
            "gap:11px;"
            "margin:8px 0 18px;'>"

            + "".join(
                f"""
                <div style="
                    background:linear-gradient(
                        145deg,
                        {WHITE},
                        {CREAM}
                    );
                    border:1px solid {LIGHT_GOLD};
                    border-top:4px solid {GOLD};
                    border-radius:11px;
                    padding:14px 16px;
                    min-height:96px;
                    box-shadow:0 3px 8px rgba(12,36,57,.06);
                ">

                    <div style="
                        color:{SLATE};
                        font-size:8.5px;
                        font-weight:700;
                        letter-spacing:1.15px;
                    ">
                        {title}
                    </div>

                    <div style="
                        color:{NAVY};
                        font:600 21px Georgia;
                        margin-top:7px;
                    ">
                        {value}
                    </div>

                    <div style="
                        color:{BLUE};
                        font-size:9.5px;
                        margin-top:5px;
                    ">
                        {subtitle}
                    </div>

                </div>
                """
                for title, value, subtitle in cards
            )

            + "</div>"
        ))


        # ====================================================
        # EXECUTIVE CONTEXT STRIP
        # ============================================================

        display(HTML(
            f"""
            <div style="
                display:grid;
                grid-template-columns:repeat(3,1fr);
                gap:8px;
                margin-bottom:17px;
            ">

                <div style="
                    background:{NAVY};
                    padding:11px 14px;
                    border-radius:7px;
                    color:{WHITE};
                ">
                    <span style="
                        color:{LIGHT_GOLD};
                        font-size:9px;
                        font-weight:700;
                    ">
                        ORIGINAL DATA QUALITY
                    </span><br>
                    <span style="font-size:12px;">
                        {missing_share:.1f}% originally missing
                    </span>
                </div>

                <div style="
                    background:{BLUE};
                    padding:11px 14px;
                    border-radius:7px;
                    color:{WHITE};
                ">
                    <span style="
                        color:{LIGHT_GOLD};
                        font-size:9px;
                        font-weight:700;
                    ">
                        MOST AFFECTED STORE TYPE
                    </span><br>
                    <span style="font-size:12px;">
                        {highest_missing_type}
                        {
                            f" · {highest_missing_rate:.1f}% missing"
                            if pd.notna(highest_missing_rate)
                            else ""
                        }
                    </span>
                </div>

                <div style="
                    background:{GOLD};
                    padding:11px 14px;
                    border-radius:7px;
                    color:{NAVY};
                ">
                    <span style="
                        font-size:9px;
                        font-weight:700;
                    ">
                        HISTORICAL EVIDENCE
                    </span><br>
                    <span style="font-size:12px;">
                        {transitions:,} missing → known transitions
                    </span>
                </div>

            </div>
            """
        ))


        # ====================================================
        # TABLE VIEW
        # ============================================================

        if audit_view.value == "Table":

            table = d[
                [
                    "Store",
                    "StoreType",
                    "AvailabilityStatus",
                    "CompetitionDistanceKm",
                    "DistanceBand",
                    "Consistency",
                    "Transition"
                ]
            ].sort_values(
                [
                    "AvailabilityStatus",
                    "CompetitionDistanceKm"
                ]
            ).reset_index(drop=True)

            display(
                table.style
                .format({
                    "CompetitionDistanceKm":
                    lambda x:
                    f"{x:.2f} km"
                    if pd.notna(x)
                    else "Missing"
                })
                .set_properties(**{
                    "text-align": "center",
                    "padding": "8px",
                    "color": SLATE
                })
                .set_table_styles([
                    {
                        "selector": "th",
                        "props": [
                            ("background-color", NAVY),
                            ("color", WHITE),
                            ("font-weight", "bold"),
                            ("text-align", "center"),
                            ("padding", "9px")
                        ]
                    },
                    {
                        "selector": "tr:nth-child(even)",
                        "props": [
                            ("background-color", CREAM)
                        ]
                    }
                ])
            )


        # ====================================================
        # UNIQUE CHART 1 — ICICLE
        # ============================================================

        elif audit_chart.value == "Icicle":

            tree = (
                d.groupby(
                    [
                        "StoreType",
                        "AvailabilityStatus",
                        "Consistency",
                        "DistanceBand"
                    ],
                    observed=True
                )
                .size()
                .reset_index(name="Stores")
            )

            tree["AvailabilityCode"] = np.where(
                tree["AvailabilityStatus"]
                .eq("Originally Observed"),
                0,
                1
            )

            fig = px.icicle(
                tree,

                path=[
                    px.Constant("Rossmann Network"),
                    "StoreType",
                    "AvailabilityStatus",
                    "Consistency",
                    "DistanceBand"
                ],

                values="Stores",

                color="AvailabilityCode",

                color_continuous_scale=[
                    [0.00, NAVY],
                    [0.35, BLUE],
                    [0.70, LIGHT_GOLD],
                    [1.00, GOLD]
                ]
            )

            fig.update_traces(
                textinfo="label+value+percent root",
                textfont=dict(
                    size=12
                ),
                hovertemplate=(
                    "<b>%{label}</b>"
                    "<br>Stores: %{value:,}"
                    "<br>Share of network: %{percentRoot:.1%}"
                    "<br>Share of parent: %{percentParent:.1%}"
                    "<extra></extra>"
                ),
                marker=dict(
                    line=dict(
                        color=WHITE,
                        width=1.2
                    )
                )
            )

            fig.update_layout(
                title={
                    "text":
                        "Competition Data Availability Architecture"
                        "<br><sup>"
                        "Network → Store Type → Original Availability → "
                        "Consistency → Competition Distance Band"
                        "</sup>",
                    "x": .02
                },

                height=650,

                paper_bgcolor=WHITE,

                font=dict(
                    color=SLATE,
                    size=12
                ),

                coloraxis_showscale=False,

                margin=dict(
                    l=20,
                    r=20,
                    t=95,
                    b=20
                )
            )

            fig.show()


        # ====================================================
        # UNIQUE CHART 2 — PARALLEL CATEGORIES
        # ============================================================

        else:

            flow = d.copy()

            flow["Availability"] = np.where(
                flow["FirstMissing"],
                "Originally Missing",
                "Originally Observed"
            )

            flow["LineCode"] = np.where(
                flow["FirstMissing"],
                1,
                0
            )

            fig = go.Figure(
                go.Parcats(
                    dimensions=[
                        dict(
                            label="Store Type",
                            values=flow["StoreType"]
                        ),
                        dict(
                            label="Original Availability",
                            values=flow["Availability"]
                        ),
                        dict(
                            label="Distance Band",
                            values=flow["DistanceBand"]
                            .astype(str)
                        ),
                        dict(
                            label="Distance Behaviour",
                            values=flow["Consistency"]
                        ),
                        dict(
                            label="Historical Transition",
                            values=flow["Transition"]
                        )
                    ],

                    line=dict(
                        color=flow["LineCode"],
                        colorscale=[
                            [0.00, NAVY],
                            [0.45, BLUE],
                            [0.55, LIGHT_GOLD],
                            [1.00, GOLD]
                        ],
                        shape="hspline"
                    ),

                    labelfont=dict(
                        size=13,
                        color=NAVY
                    ),

                    tickfont=dict(
                        size=11,
                        color=SLATE
                    ),

                    hoveron="color",

                    arrangement="freeform"
                )
            )

            fig.update_layout(
                title={
                    "text":
                        "Competition Distance Data Lineage"
                        "<br><sup>"
                        "Tracing store type, original availability, "
                        "distance band, consistency and historical transition"
                        "</sup>",
                    "x": .02
                },

                height=630,

                paper_bgcolor=WHITE,

                font=dict(
                    color=SLATE
                ),

                margin=dict(
                    l=45,
                    r=45,
                    t=95,
                    b=45
                )
            )

            fig.show()


        # ====================================================
        # BUSINESS / DATA QUALITY INTERPRETATION
        # ============================================================

        if transitions > 0:
            transition_text = (
                f"<strong>{transitions:,}</strong> stores demonstrate "
                "an observable missing-to-known transition."
            )
        else:
            transition_text = (
                "No genuine <strong>missing → known</strong> transition "
                "is observable in the available historical records."
            )

        if static_share >= 95:
            stability_text = (
                "CompetitionDistance behaves overwhelmingly as a "
                "<strong>static store characteristic</strong>."
            )
        else:
            stability_text = (
                "A minority of stores show more than one available "
                "competition-distance value and require individual review."
            )

        display(HTML(
            f"""
            <div style="
                margin-top:18px;
                padding:20px 22px;
                background:linear-gradient(
                    135deg,
                    {CREAM},
                    {WHITE}
                );
                border-left:5px solid {GOLD};
                border-radius:0 9px 9px 0;
                color:{SLATE};
                line-height:1.85;
                font-size:13px;
            ">

                <div style="
                    color:{NAVY};
                    font-size:10px;
                    font-weight:700;
                    letter-spacing:1.5px;
                    margin-bottom:8px;
                ">
                    DYNAMIC COMPETITION DATA QUALITY INSIGHT
                </div>

                Of <strong>{total:,}</strong> stores in the current view,
                <strong>{missing:,}</strong>
                (<strong>{missing_share:.1f}%</strong>) were originally
                missing CompetitionDistance, compared with
                <strong>{observed_share:.1f}%</strong> whose distance was
                originally observed.

                <br><br>

                The highest original missingness is currently associated
                with <strong>Store Type {highest_missing_type}</strong>
                {
                    f" at <strong>{highest_missing_rate:.1f}%</strong>"
                    if pd.notna(highest_missing_rate)
                    else ""
                }.

                <strong>{static_share:.1f}%</strong> of stores maintain
                a single distance value across the available history.
                {stability_text}

                <br><br>

                {transition_text}

                This means the dataset provides substantially stronger
                evidence for analysing <strong>static competitive
                proximity</strong> than for reconstructing historical
                changes in when competition-distance information became
                available.

            </div>
            """
        ))


# ============================================================
# RESET + INTERACTIVITY
# ============================================================

def reset_audit(_):
    audit_view.value = "Charts"
    audit_chart.value = "Icicle"
    audit_store.value = "All"
    audit_status.value = "All"


audit_reset.on_click(reset_audit)

for control in [
    audit_view,
    audit_chart,
    audit_store,
    audit_status
]:
    control.observe(
        render_audit,
        names="value"
    )


# ============================================================
# PROFESSIONAL HEADING
# ============================================================

display(HTML(
    f"""
    <div style="
        padding:21px 24px;
        margin:6px 0 14px;
        background:
            linear-gradient(
                115deg,
                {NAVY} 0%,
                {BLUE} 72%,
                {NAVY} 100%
            );
        border-bottom:4px solid {GOLD};
        border-radius:11px;
        box-shadow:0 4px 12px rgba(12,36,57,.12);
    ">

        <div style="
            color:{LIGHT_GOLD};
            font-size:10px;
            font-weight:700;
            letter-spacing:2.1px;
            margin-bottom:6px;
        ">
            DATA QUALITY · COMPETITIVE MARKET INTELLIGENCE
        </div>

        <div style="
            color:{WHITE};
            font:500 26px Georgia;
            margin-bottom:7px;
        ">
            Competition Distance Availability & Integrity Audit
        </div>

        <div style="
            color:{CREAM};
            font-size:12.5px;
            line-height:1.65;
            max-width:1050px;
        ">
            Evaluate original distance availability, imputation exposure,
            store-type concentration, data consistency and the presence of
            historical missing-to-known transitions across the Rossmann network.
        </div>

    </div>
    """
))


# ============================================================
# DISPLAY
# ============================================================

display(
    widgets.HBox(
        [
            audit_view,
            audit_store,
            audit_status,
            audit_reset
        ],
        layout=widgets.Layout(
            gap="10px",
            flex_flow="row wrap",
            margin="0 0 8px 0"
        )
    ),

    audit_chart,

    audit_out
)

render_audit()

logger.info(
    "Competition availability and integrity dashboard initialized."
)
```

<h2><span style="color:lightblue">INTERPRETATION<b></b></span>

CompetitionDistance is represented as a store-level characteristic rather than
a time-varying daily variable. As a result, the same available value is repeated
across the historical observations of each store after the store metadata is
merged with daily sales records.



Therefore, a store whose distance was genuinely unknown at one historical point
and recorded later cannot be identified unless multiple dated versions of the
store metadata are available. Any missing distance observed in the supplied
store information should instead be treated transparently as missing data.



The effect of changing competitive conditions is better captured through the
available <strong>CompetitionOpenSinceMonth</strong> and
<strong>CompetitionOpenSinceYear</strong> variables, which were analysed in
the preceding competitor-entry event study.

```python
task1_df = eda_df.copy()

task1_df["SalesPerCustomer"] = np.where(
    task1_df["Customers"].gt(0),
    task1_df["Sales"] / task1_df["Customers"],
    np.nan
)

task1_df["Weekend"] = (
    task1_df["IsWeekend"]
    if "IsWeekend" in task1_df.columns
    else task1_df["DayOfWeek"].isin([6, 7]).astype(int)
)

task1_df["Holiday"] = ~task1_df["StateHoliday"].astype(str).str.lower().eq("0")
task1_df["MonthNo"] = task1_df["Date"].dt.month
task1_df["Month"] = task1_df["Date"].dt.month_name().str[:3]

task1_df["AssortmentName"] = (
    task1_df["Assortment"].astype(str).str.lower()
    .map({"a":"Basic", "b":"Extra", "c":"Extended"})
)

comp_missing_col = next(
    (c for c in [
        "CompetitionDistanceMissing",
        "CompetitionDistance_Missing",
        "CompetitionDistance_missing"
    ] if c in task1_df.columns),
    None
)

if {
    "CompetitionOpenSinceYear",
    "CompetitionOpenSinceMonth"
}.issubset(task1_df.columns):

    task1_df["CompetitionOpenDate"] = pd.to_datetime(
        {
            "year": pd.to_numeric(
                task1_df["CompetitionOpenSinceYear"],
                errors="coerce"
            ),
            "month": pd.to_numeric(
                task1_df["CompetitionOpenSinceMonth"],
                errors="coerce"
            ),
            "day": 1
        },
        errors="coerce"
    )

    task1_df["RelativeCompMonth"] = (
        (task1_df["Date"].dt.year -
         task1_df["CompetitionOpenDate"].dt.year) * 12
        + task1_df["Date"].dt.month
        - task1_df["CompetitionOpenDate"].dt.month
    )

train_promo_rate = task1_df["Promo"].mean() * 100

test_promo_rate = (
    test_clean["Promo"].mean() * 100
    if "test_clean" in globals() and "Promo" in test_clean.columns
    else np.nan
)

promo_gap = (
    abs(train_promo_rate - test_promo_rate)
    if pd.notna(test_promo_rate)
    else np.nan
)

def pct_effect(a, b):
    return (
        (a / b - 1) * 100
        if pd.notna(a) and pd.notna(b) and b != 0
        else np.nan
    )

def score_status(x):
    if pd.isna(x): return "Not Available"
    if x >= 90: return "Excellent"
    if x >= 80: return "Strong"
    if x >= 70: return "Good"
    if x >= 60: return "Moderate"
    return "Requires Attention"

def task1_metrics(d):

    o = d[
        d["Open"].eq(1) &
        d["Customers"].gt(0)
    ].copy()

    if o.empty:
        return None

    avg_sales = o["Sales"].mean()
    avg_customers = o["Customers"].mean()
    spc = (
        o["Sales"].sum() /
        o["Customers"].sum()
    )

    corr = o["Sales"].corr(o["Customers"])

    sales_cv = (
        o["Sales"].std() /
        avg_sales * 100
        if avg_sales else np.nan
    )

    promo_effect = pct_effect(
        o.loc[o["Promo"].eq(1), "Sales"].mean(),
        o.loc[o["Promo"].eq(0), "Sales"].mean()
    )

    weekend_effect = pct_effect(
        o.loc[o["Weekend"].eq(1), "Sales"].mean(),
        o.loc[o["Weekend"].eq(0), "Sales"].mean()
    )

    holiday_effect = pct_effect(
        o.loc[o["Holiday"], "Sales"].mean(),
        o.loc[~o["Holiday"], "Sales"].mean()
    )

    school_effect = (
        pct_effect(
            o.loc[o["SchoolHoliday"].eq(1), "Sales"].mean(),
            o.loc[o["SchoolHoliday"].eq(0), "Sales"].mean()
        )
        if "SchoolHoliday" in o.columns
        else np.nan
    )

    monthly = (
        o.groupby(["MonthNo", "Month"], observed=True)["Sales"]
        .mean()
        .reset_index()
        .sort_values("MonthNo")
    )

    peak = (
        monthly.loc[monthly["Sales"].idxmax()]
        if not monthly.empty
        else None
    )

    assortment = (
        o.dropna(subset=["AssortmentName"])
        .groupby("AssortmentName", observed=True)["Sales"]
        .mean()
    )

    comp = (
        o.dropna(subset=["CompetitionDistance"])
        .groupby(
            ["Store", "StoreType", "CompetitionDistance"],
            observed=True
        )
        .agg(
            AvgSales=("Sales", "mean"),
            AvgCustomers=("Customers", "mean")
        )
        .reset_index()
    )

    near_sales = (
        comp.loc[
            comp["CompetitionDistance"].le(1000),
            "AvgSales"
        ].mean()
    )

    far_sales = (
        comp.loc[
            comp["CompetitionDistance"].gt(5000),
            "AvgSales"
        ].mean()
    )

    entry_effect = np.nan

    if "RelativeCompMonth" in o.columns:

        event = o[
            o["RelativeCompMonth"].between(-6, 6) &
            o["RelativeCompMonth"].ne(0)
        ].copy()

        if not event.empty:

            event["Period"] = np.where(
                event["RelativeCompMonth"].lt(0),
                "Before",
                "After"
            )

            counts = (
                event.groupby(["Store", "Period"])
                .size()
                .unstack(fill_value=0)
            )

            if {"Before","After"}.issubset(counts.columns):

                eligible = counts[
                    counts["Before"].ge(20) &
                    counts["After"].ge(20)
                ].index

                impact = (
                    event[event["Store"].isin(eligible)]
                    .groupby(["Store","Period"])["Sales"]
                    .mean()
                    .unstack()
                    .dropna()
                )

                if not impact.empty:

                    entry_effect = (
                        (
                            impact["After"] /
                            impact["Before"] - 1
                        ) * 100
                    ).median()

    core = [
        c for c in [
            "Sales","Customers","Open","Promo",
            "StoreType","Assortment",
            "CompetitionDistance"
        ]
        if c in d.columns
    ]

    completeness = (
        1 -
        d[core].isna().mean().mean()
    ) * 100

    competition_observed = (
        (
            1 -
            d.groupby("Store")[comp_missing_col]
            .first()
            .fillna(0)
            .astype(bool)
            .mean()
        ) * 100
        if comp_missing_col
        else
        d.groupby("Store")["CompetitionDistance"]
        .first()
        .notna()
        .mean() * 100
    )

    return {

        "TotalSales": o["Sales"].sum(),
        "Sales": avg_sales,
        "MedianSales": o["Sales"].median(),
        "P90Sales": o["Sales"].quantile(.90),

        "TotalCustomers": o["Customers"].sum(),
        "Customers": avg_customers,
        "SPC": spc,
        "Corr": corr,
        "SalesCV": sales_cv,

        "PromoRate": o["Promo"].mean() * 100,
        "PromoEffect": promo_effect,

        "OpenRate": d["Open"].mean() * 100,
        "WeekendEffect": weekend_effect,

        "HolidayEffect": holiday_effect,
        "SchoolEffect": school_effect,

        "PeakMonth": peak["Month"] if peak is not None else "—",

        "PeakEffect": (
            pct_effect(peak["Sales"], avg_sales)
            if peak is not None
            else np.nan
        ),

        "TopAssortment": (
            assortment.idxmax()
            if not assortment.empty
            else "—"
        ),

        "AssortmentEffect": (
            pct_effect(assortment.max(), assortment.min())
            if len(assortment) > 1
            else np.nan
        ),

        "CompetitionDistance": (
            comp["CompetitionDistance"].median() / 1000
            if not comp.empty
            else np.nan
        ),

        "CompetitionEffect": pct_effect(
            far_sales,
            near_sales
        ),

        "Within1km": (
            comp["CompetitionDistance"].le(1000).mean() * 100
            if not comp.empty
            else np.nan
        ),

        "EntryEffect": entry_effect,

        "Completeness": completeness,
        "CompetitionObserved": competition_observed,

        "Stores": o["Store"].nunique(),
        "OpenDays": len(o),

        "Monthly": monthly,
        "CompStore": comp
    }

network = task1_metrics(task1_df)

def card_grid(cards):

    return HTML(
        "<div style='"
        "display:grid;"
        "grid-template-columns:repeat(auto-fit,minmax(175px,1fr));"
        "gap:10px;margin:8px 0 17px;'>"

        + "".join(
            f"""
            <div style="
                background:linear-gradient(145deg,{WHITE},{CREAM});
                border:1px solid {LIGHT_GOLD};
                border-top:4px solid {GOLD};
                border-radius:10px;
                padding:13px 15px;
                min-height:91px;
            ">

                <div style="
                    color:{SLATE};
                    font-size:8px;
                    font-weight:700;
                    letter-spacing:1px;
                ">
                    {title}
                </div>

                <div style="
                    color:{NAVY};
                    font:600 19px Georgia;
                    margin-top:6px;
                ">
                    {value}
                </div>

                <div style="
                    color:{BLUE};
                    font-size:9px;
                    margin-top:5px;
                    line-height:1.35;
                ">
                    {note}
                </div>

            </div>
            """
            for title,value,note in cards
        )

        + "</div>"
    )

def panel_grid(panels):

    return HTML(
        "<div style='"
        "display:grid;"
        "grid-template-columns:repeat(auto-fit,minmax(260px,1fr));"
        "gap:11px;margin-top:19px;'>"

        + "".join(
            f"""
            <div style="
                background:{WHITE};
                border:1px solid {LIGHT_GOLD};
                border-top:4px solid {accent};
                border-radius:10px;
                padding:18px 19px;
                min-height:210px;
            ">

                <div style="
                    color:{NAVY};
                    font-size:10px;
                    font-weight:700;
                    letter-spacing:1.3px;
                    margin-bottom:11px;
                ">
                    {title}
                </div>

                <div style="
                    color:{SLATE};
                    font-size:12.3px;
                    line-height:1.78;
                ">
                    {text}
                </div>

            </div>
            """
            for title,accent,text in panels
        )

        + "</div>"
    )

def clean_layout(fig, title, x="", y="", height=460):

    fig.update_layout(
        title=dict(
            text=title,
            x=.02,
            font=dict(
                color=NAVY,
                size=17
            )
        ),
        xaxis_title=x,
        yaxis_title=y,
        height=height,
        paper_bgcolor=WHITE,
        plot_bgcolor=WHITE,
        font=dict(color=SLATE),
        margin=dict(
            l=60,
            r=35,
            t=75,
            b=65
        )
    )

    fig.update_xaxes(
        gridcolor=CREAM,
        automargin=True
    )

    fig.update_yaxes(
        gridcolor=CREAM,
        automargin=True
    )

task1_view = widgets.ToggleButtons(
    options=[
        ("★ Executive Scorecard","Scorecard"),
        ("◉ Performance Profile","Profile"),
        ("↔ Driver Intelligence","Drivers"),
        ("◒ Calendar Pulse","Calendar"),
        ("◇ Store Strategy","Strategy"),
        ("◎ Competition","Competition"),
        ("▦ Evidence Register","Evidence")
    ],
    value="Scorecard"
)

task1_year = widgets.Dropdown(
    options=["All"] + sorted(
        task1_df["Year"].dropna().unique().tolist()
    ),
    value="All",
    description="Year:"
)

task1_store = widgets.Dropdown(
    options=["All"] + sorted(
        task1_df["StoreType"].dropna().unique().tolist()
    ),
    value="All",
    description="Store:"
)

task1_promo = widgets.Dropdown(
    options=[
        ("All Trading Days","All"),
        ("Promo Days",1),
        ("Non-Promo Days",0)
    ],
    value="All",
    description="Promo:"
)

task1_day = widgets.Dropdown(
    options=[
        ("All Days","All"),
        ("Weekdays",0),
        ("Weekends",1)
    ],
    value="All",
    description="Day:"
)

task1_reset = widgets.Button(
    description="Reset Filters",
    icon="refresh"
)

task1_out = widgets.Output()

def render_task1(*_):

    with task1_out:

        clear_output(wait=True)

        d = task1_df.copy()

        for col,val in {
            "Year":task1_year.value,
            "StoreType":task1_store.value,
            "Promo":task1_promo.value,
            "Weekend":task1_day.value
        }.items():

            if val != "All":
                d = d[d[col].eq(val)]

        m = task1_metrics(d)

        if not m:
            display(HTML(
                f"""
                <div style="
                    padding:15px;
                    background:{CREAM};
                    border-left:4px solid {GOLD};
                    color:{SLATE};
                ">
                    No valid observations are available for the selected scope.
                </div>
                """
            ))
            return

        drivers = {
            "Promotion":m["PromoEffect"],
            "Weekend Trading":m["WeekendEffect"],
            "State Holiday":m["HolidayEffect"],
            "School Holiday":m["SchoolEffect"],
            "Seasonal Peak":m["PeakEffect"],
            "Assortment":m["AssortmentEffect"],
            "Competition Distance":m["CompetitionEffect"],
            "Competitor Entry":m["EntryEffect"]
        }

        drivers = {
            k:v for k,v in drivers.items()
            if pd.notna(v)
        }

        ranked = sorted(
            drivers.items(),
            key=lambda x:abs(x[1]),
            reverse=True
        )

        strongest,strongest_value = (
            ranked[0]
            if ranked
            else ("—",np.nan)
        )

        display(HTML(f"""
        <div style="
            display:grid;
            grid-template-columns:repeat(auto-fit,minmax(190px,1fr));
            gap:9px;
            margin-bottom:15px;
        ">
            <div style="
                background:{NAVY};
                padding:13px 15px;
                border-radius:8px;
            ">
                <small style="
                    color:{LIGHT_GOLD};
                    font-weight:700;
                    letter-spacing:1px;
                ">
                    ANALYTICAL SCOPE
                </small>

                <div style="color:{WHITE};margin-top:5px;">
                    {m['Stores']:,} stores ·
                    {m['OpenDays']:,} open store-days
                </div>
            </div>
            <div style="
                background:{BLUE};
                padding:13px 15px;
                border-radius:8px;
            ">
                <small style="
                    color:{LIGHT_GOLD};
                    font-weight:700;
                    letter-spacing:1px;
                ">
                    STRONGEST OBSERVED SIGNAL
                </small>

                <div style="color:{WHITE};margin-top:5px;">
                    {strongest}
                    {
                        f" · {strongest_value:+.1f}%"
                        if pd.notna(strongest_value)
                        else ""
                    }
                </div>
            </div>

            <div style="
                background:{CREAM};
                border-left:4px solid {GOLD};
                padding:13px 15px;
                border-radius:8px;
            ">
                <small style="
                    color:{SLATE};
                    font-weight:700;
                ">
                    DATA COMPLETENESS
                </small>

                <div style="
                    color:{NAVY};
                    font-weight:600;
                    margin-top:5px;
                ">
                    {m['Completeness']:.1f}%
                </div>
            </div>

            <div style="
                background:{GOLD};
                padding:13px 15px;
                border-radius:8px;
            ">
                <small style="
                    color:{NAVY};
                    font-weight:700;
                ">
                    TRAIN–TEST PROMO GAP
                </small>

                <div style="
                    color:{NAVY};
                    font-weight:600;
                    margin-top:5px;
                ">
                    {
                        f"{promo_gap:.2f} pp"
                        if pd.notna(promo_gap)
                        else "N/A"
                    }
                </div>
            </div>

        </div>
        """))

        cards = [

            ("TOTAL SALES",
             f"{m['TotalSales']:,.0f}",
             "Selected analytical scope"),

            ("AVG DAILY SALES",
             f"{m['Sales']:,.0f}",
             "Per open store-day"),

            ("MEDIAN SALES",
             f"{m['MedianSales']:,.0f}",
             "Typical store-day"),

            ("P90 SALES",
             f"{m['P90Sales']:,.0f}",
             "High-demand threshold"),


            ("TOTAL CUSTOMERS",
             f"{m['TotalCustomers']:,.0f}",
             "Observed customer visits"),

            ("AVG CUSTOMERS",
             f"{m['Customers']:,.0f}",
             "Per open store-day"),

            ("SALES / CUSTOMER",
             f"{m['SPC']:.2f}",
             "Purchasing intensity"),

            ("SALES–CUSTOMER r",
             f"{m['Corr']:.3f}",
             "Demand association"),


            ("PROMO EXPOSURE",
             f"{m['PromoRate']:.1f}%",
             "Open days on promotion"),

            (
                "PROMOTION EFFECT",
                f"{m['PromoEffect']:+.1f}%"
                if pd.notna(m["PromoEffect"])
                else "N/A",
                "Promo vs non-promo sales"
            ),

            ("OPEN RATE",
             f"{m['OpenRate']:.1f}%",
             f"{100-m['OpenRate']:.1f}% closed"),

            (
                "WEEKEND EFFECT",
                f"{m['WeekendEffect']:+.1f}%"
                if pd.notna(m["WeekendEffect"])
                else "N/A",
                "Weekend vs weekday"
            ),

            (
                "SEASONAL PEAK",
                m["PeakMonth"],
                (
                    f"{m['PeakEffect']:+.1f}% vs average"
                    if pd.notna(m["PeakEffect"])
                    else "Highest-sales month"
                )
            ),

            (
                "TOP ASSORTMENT",
                m["TopAssortment"],
                (
                    f"{m['AssortmentEffect']:+.1f}% best–lowest"
                    if pd.notna(m["AssortmentEffect"])
                    else "Sales leader"
                )
            ),

            (
                "MEDIAN COMPETITION",
                f"{m['CompetitionDistance']:.2f} km"
                if pd.notna(m["CompetitionDistance"])
                else "N/A",
                (
                    f"{m['Within1km']:.1f}% within 1 km"
                    if pd.notna(m["Within1km"])
                    else ""
                )
            ),

            (
                "COMPETITOR ENTRY",
                f"{m['EntryEffect']:+.1f}%"
                if pd.notna(m["EntryEffect"])
                else "N/A",
                "Median post-entry sales effect"
            )
        ]

        display(card_grid(cards))

        volatility = (
            "High"
            if m["SalesCV"] >= 50
            else "Moderate"
            if m["SalesCV"] >= 30
            else "Low"
        )

        demand_structure = (
            "Strong Traffic Link"
            if abs(m["Corr"]) >= .70
            else "Moderate Traffic Link"
            if abs(m["Corr"]) >= .40
            else "Mixed Demand Drivers"
        )

        secondary = [

            (
                "SALES VOLATILITY",
                f"{m['SalesCV']:.1f}% CV",
                volatility
            ),

            (
                "DEMAND STRUCTURE",
                f"r = {m['Corr']:.3f}",
                demand_structure
            ),

            (
                "STATE HOLIDAY EFFECT",
                f"{m['HolidayEffect']:+.1f}%"
                if pd.notna(m["HolidayEffect"])
                else "N/A",
                "Holiday vs normal"
            ),

            (
                "SCHOOL HOLIDAY EFFECT",
                f"{m['SchoolEffect']:+.1f}%"
                if pd.notna(m["SchoolEffect"])
                else "N/A",
                "Calendar effect"
            ),

            (
                "DATA QUALITY",
                f"{m['Completeness']:.1f}%",
                score_status(m["Completeness"])
            ),

            (
                "COMPETITION COVERAGE",
                f"{m['CompetitionObserved']:.1f}%",
                "Originally observed"
            ),

            (
                "COMPETITION ≤1 KM",
                f"{m['Within1km']:.1f}%"
                if pd.notna(m["Within1km"])
                else "N/A",
                "Store proximity"
            ),

            (
                "PROMO ALIGNMENT",
                f"{promo_gap:.2f} pp"
                if pd.notna(promo_gap)
                else "N/A",
                "Train–test difference"
            )
        ]

        display(HTML(
            "<div style='"
            "display:grid;"
            "grid-template-columns:repeat(auto-fit,minmax(190px,1fr));"
            "gap:8px;margin-bottom:18px;'>"

            + "".join(
                f"""
                <div style="
                    background:{WHITE};
                    border-left:4px solid {accent};
                    padding:11px 13px;
                    border-radius:6px;
                ">

                    <div style="
                        color:{SLATE};
                        font-size:8px;
                        font-weight:700;
                        letter-spacing:1px;
                    ">
                        {title}
                    </div>

                    <div style="
                        color:{NAVY};
                        font-size:14px;
                        font-weight:600;
                        margin-top:4px;
                    ">
                        {value}
                    </div>

                    <div style="
                        color:{BLUE};
                        font-size:9px;
                        margin-top:3px;
                    ">
                        {status}
                    </div>

                </div>
                """
                for (title,value,status),accent
                in zip(
                    secondary,
                    [
                        NAVY,BLUE,GOLD,LIGHT_GOLD,
                        NAVY,BLUE,GOLD,LIGHT_GOLD
                    ]
                )
            )

            + "</div>"
        ))

        if task1_view.value == "Scorecard":

            demand_score = np.clip(
                abs(m["Corr"]) * 100,
                0,100
            )

            operation_score = np.clip(
                m["OpenRate"],
                0,100
            )

            quality_score = np.clip(
                m["Completeness"],
                0,100
            )

            competition_score = np.clip(
                m["CompetitionObserved"],
                0,100
            )

            promo_score = (
                np.clip(
                    100 - promo_gap,
                    0,100
                )
                if pd.notna(promo_gap)
                else np.nan
            )

            score_values = [
                demand_score,
                operation_score,
                quality_score,
                competition_score,
                promo_score
            ]

            overall_score = np.mean(
                [
                    x for x in score_values
                    if pd.notna(x)
                ]
            )

            scores = [
                (
                    "OVERALL EDA READINESS",
                    overall_score,
                    "Composite analytical readiness"
                ),
                (
                    "DEMAND SIGNAL",
                    demand_score,
                    "Sales–customer relationship"
                ),
                (
                    "OPERATING COVERAGE",
                    operation_score,
                    "Store availability evidence"
                ),
                (
                    "DATA READINESS",
                    quality_score,
                    "Core analytical completeness"
                ),
                (
                    "COMPETITION COVERAGE",
                    competition_score,
                    "Originally observed distance"
                ),
                (
                    "PROMO ALIGNMENT",
                    promo_score,
                    "Train–test consistency"
                )
            ]

            display(HTML(
                "<div style='"
                "display:grid;"
                "grid-template-columns:repeat(auto-fit,minmax(220px,1fr));"
                "gap:11px;margin:10px 0 20px;'>"

                + "".join(
                    f"""
                    <div style="
                        background:{WHITE};
                        border:1px solid {LIGHT_GOLD};
                        border-radius:10px;
                        padding:16px 17px;
                        min-height:132px;
                    ">

                        <div style="
                            display:flex;
                            justify-content:space-between;
                            gap:10px;
                            flex-wrap:wrap;
                        ">

                            <span style="
                                color:{SLATE};
                                font-size:8px;
                                font-weight:700;
                                letter-spacing:1px;
                            ">
                                {title}
                            </span>

                            <span style="
                                color:{GOLD};
                                font-size:9px;
                                font-weight:700;
                            ">
                                {score_status(value)}
                            </span>

                        </div>

                        <div style="
                            color:{NAVY};
                            font:600 25px Georgia;
                            margin:8px 0 9px;
                        ">
                            {
                                f"{value:.1f}"
                                if pd.notna(value)
                                else "N/A"
                            }

                            <span style="
                                font-size:10px;
                                color:{SLATE};
                                font-family:'Segoe UI';
                            ">
                                / 100
                            </span>
                        </div>

                        <div style="
                            width:100%;
                            height:7px;
                            background:{CREAM};
                            border-radius:10px;
                            overflow:hidden;
                        ">

                            <div style="
                                width:{
                                    min(value,100)
                                    if pd.notna(value)
                                    else 0
                                }%;
                                height:100%;
                                background:linear-gradient(
                                    90deg,
                                    {BLUE},
                                    {GOLD}
                                );
                            ">
                            </div>

                        </div>

                        <div style="
                            color:{BLUE};
                            font-size:9px;
                            margin-top:9px;
                            line-height:1.4;
                        ">
                            {note}
                        </div>

                    </div>
                    """
                    for title,value,note in scores
                )

                + "</div>"
            ))

            score_df = pd.DataFrame(
                scores[1:],
                columns=[
                    "Dimension",
                    "Score",
                    "Description"
                ]
            ).dropna(
                subset=["Score"]
            ).sort_values("Score")

            fig = go.Figure(
                go.Bar(
                    x=score_df["Score"],
                    y=score_df["Dimension"],
                    orientation="h",

                    marker=dict(
                        color=NAVY,
                        line=dict(
                            color=BLUE,
                            width=.7
                        )
                    ),

                    customdata=score_df[
                        ["Description"]
                    ],

                    hovertemplate=(
                        "<b>%{y}</b>"
                        "<br>Score: %{x:.1f}/100"
                        "<br>%{customdata[0]}"
                        "<extra></extra>"
                    )
                )
            )

            fig.add_vline(
                x=80,
                line_dash="dash",
                line_color=GOLD
            )

            clean_layout(
                fig,
                "Executive Analytical Readiness Scorecard",
                "Readiness Score / 100",
                "",
                430
            )

            fig.update_xaxes(
                range=[0,105],
                gridcolor=CREAM
            )

            fig.show()

            strongest_score = (
                score_df.loc[
                    score_df["Score"].idxmax()
                ]
                if not score_df.empty
                else None
            )

            weakest_score = (
                score_df.loc[
                    score_df["Score"].idxmin()
                ]
                if not score_df.empty
                else None
            )

            display(HTML(f"""
            <div style="
                display:grid;
                grid-template-columns:repeat(auto-fit,minmax(250px,1fr));
                gap:10px;
                margin-top:10px;
            ">

                <div style="
                    background:{CREAM};
                    border-left:4px solid {NAVY};
                    padding:14px 16px;
                    border-radius:7px;
                ">

                    <strong style="color:{NAVY};font-size:9px;">
                        STRONGEST READINESS DIMENSION
                    </strong>

                    <div style="
                        color:{SLATE};
                        font-size:12px;
                        line-height:1.6;
                        margin-top:6px;
                    ">
                        {
                            f"{strongest_score['Dimension']} records "
                            f"the strongest score at "
                            f"<strong>{strongest_score['Score']:.1f}/100</strong>."
                            if strongest_score is not None
                            else "Not available."
                        }
                    </div>
                </div>

                <div style="
                    background:{CREAM};
                    border-left:4px solid {GOLD};
                    padding:14px 16px;
                    border-radius:7px;
                ">

                    <strong style="color:{NAVY};font-size:9px;">
                        PRIORITY IMPROVEMENT AREA
                    </strong>

                    <div style="
                        color:{SLATE};
                        font-size:12px;
                        line-height:1.6;
                        margin-top:6px;
                    ">
                        {
                            f"{weakest_score['Dimension']} is the "
                            f"lowest-scoring dimension at "
                            f"<strong>{weakest_score['Score']:.1f}/100</strong>."
                            if weakest_score is not None
                            else "Not available."
                        }
                    </div>
                </div>


                <div style="
                    background:{NAVY};
                    padding:14px 16px;
                    border-radius:7px;
                ">

                    <strong style="
                        color:{LIGHT_GOLD};
                        font-size:9px;
                    ">
                        OVERALL ANALYTICAL READINESS
                    </strong>

                    <div style="
                        color:{CREAM};
                        font-size:12px;
                        line-height:1.6;
                        margin-top:6px;
                    ">
                        Overall readiness is
                        <strong>{overall_score:.1f}/100</strong>,
                        classified as
                        <strong>{score_status(overall_score)}</strong>.
                    </div>

                </div>

            </div>
            """))

        elif task1_view.value == "Profile":

            profile = pd.DataFrame({

                "Metric":[
                    "Average Sales",
                    "Customer Traffic",
                    "Sales / Customer",
                    "Open Coverage",
                    "Promo Exposure",
                    "Competition Observed"
                ],

                "Current":[
                    m["Sales"],
                    m["Customers"],
                    m["SPC"],
                    m["OpenRate"],
                    m["PromoRate"],
                    m["CompetitionObserved"]
                ],

                "Network":[
                    network["Sales"],
                    network["Customers"],
                    network["SPC"],
                    network["OpenRate"],
                    network["PromoRate"],
                    network["CompetitionObserved"]
                ]
            })

            profile["Index"] = (
                profile["Current"] /
                profile["Network"] * 100
            )

            profile = profile.sort_values(
                "Index"
            )

            fig = go.Figure()

            for _,r in profile.iterrows():

                fig.add_trace(
                    go.Scatter(
                        x=[100,r["Index"]],
                        y=[
                            r["Metric"],
                            r["Metric"]
                        ],
                        mode="lines",
                        line=dict(
                            color=LIGHT_GOLD,
                            width=5
                        ),
                        showlegend=False,
                        hoverinfo="skip"
                    )
                )

            fig.add_trace(
                go.Scatter(
                    x=[100]*len(profile),
                    y=profile["Metric"],
                    mode="markers",
                    marker=dict(
                        size=11,
                        color=GOLD,
                        symbol="diamond"
                    ),
                    name="Network"
                )
            )

            fig.add_trace(
                go.Scatter(
                    x=profile["Index"],
                    y=profile["Metric"],
                    mode="markers",
                    marker=dict(
                        size=16,
                        color=NAVY,
                        line=dict(
                            color=BLUE,
                            width=1
                        )
                    ),
                    customdata=profile[
                        ["Current","Network"]
                    ],
                    name="Current scope",

                    hovertemplate=(
                        "<b>%{y}</b>"
                        "<br>Index: %{x:.1f}"
                        "<br>Current: %{customdata[0]:,.2f}"
                        "<br>Network: %{customdata[1]:,.2f}"
                        "<extra></extra>"
                    )
                )
            )

            fig.add_vline(
                x=100,
                line_dash="dash",
                line_color=GOLD
            )

            clean_layout(
                fig,
                "Executive Performance Profile",
                "Network Benchmark Index · 100 = Network Average",
                "",
                470
            )

            fig.update_layout(
                legend=dict(
                    orientation="h",
                    y=-.17
                )
            )

            fig.show()

        elif task1_view.value == "Drivers":

            signals = pd.DataFrame(
                ranked,
                columns=["Driver","Effect"]
            ).sort_values("Effect")

            if signals.empty:

                display(HTML(
                    f"""
                    <div style="
                        padding:15px;
                        background:{CREAM};
                        border-left:4px solid {GOLD};
                        color:{SLATE};
                    ">
                        No comparable driver effects are available
                        under the selected filters.
                    </div>
                    """
                ))

            else:

                fig = go.Figure(
                    go.Bar(
                        x=signals["Effect"],
                        y=signals["Driver"],
                        orientation="h",

                        marker=dict(
                            color=[
                                NAVY if x >= 0 else GOLD
                                for x in signals["Effect"]
                            ],
                            line=dict(
                                color=BLUE,
                                width=.7
                            )
                        ),

                        customdata=np.abs(
                            signals[["Effect"]].values
                        ),

                        hovertemplate=(
                            "<b>%{y}</b>"
                            "<br>Observed difference: %{x:+.1f}%"
                            "<br>Magnitude: %{customdata[0]:.1f}%"
                            "<extra></extra>"
                        )
                    )
                )

                fig.add_vline(
                    x=0,
                    line_color=SLATE,
                    line_width=1
                )

                clean_layout(
                    fig,
                    "Business Driver Priority Intelligence",
                    "Observed Relative Difference (%)",
                    "",
                    max(
                        460,
                        len(signals)*55 + 150
                    )
                )

                fig.show()

        elif task1_view.value == "Calendar":

            current = m["Monthly"].rename(
                columns={"Sales":"Current"}
            )

            base = network["Monthly"].rename(
                columns={"Sales":"Network"}
            )

            seasonal = current.merge(
                base[
                    ["MonthNo","Network"]
                ],
                on="MonthNo",
                how="left"
            )

            seasonal["Index"] = (
                seasonal["Current"] /
                seasonal["Network"] * 100
            )

            fig = go.Figure()

            fig.add_trace(
                go.Scatter(
                    x=seasonal["Month"],
                    y=seasonal["Index"],
                    mode="lines+markers",

                    line=dict(
                        color=NAVY,
                        width=3
                    ),

                    marker=dict(
                        size=9,
                        color=GOLD,
                        line=dict(
                            color=BLUE,
                            width=1
                        )
                    ),

                    customdata=seasonal[
                        ["Current","Network"]
                    ],

                    hovertemplate=(
                        "<b>%{x}</b>"
                        "<br>Performance index: %{y:.1f}"
                        "<br>Current sales: %{customdata[0]:,.0f}"
                        "<br>Network sales: %{customdata[1]:,.0f}"
                        "<extra></extra>"
                    )
                )
            )

            fig.add_hline(
                y=100,
                line_dash="dash",
                line_color=GOLD
            )

            clean_layout(
                fig,
                "Calendar & Seasonal Performance Pulse",
                "Month",
                "Sales Index · Network Month = 100",
                460
            )

            fig.show()

        elif task1_view.value == "Strategy":

            strategy = (
                d[
                    d["Open"].eq(1) &
                    d["Customers"].gt(0)
                ]
                .dropna(subset=["AssortmentName"])
                .groupby(
                    ["StoreType","AssortmentName"],
                    observed=True
                )
                .agg(
                    AvgSales=("Sales","mean"),
                    AvgCustomers=("Customers","mean"),
                    TotalSales=("Sales","sum"),
                    TotalCustomers=("Customers","sum"),
                    Stores=("Store","nunique")
                )
                .reset_index()
            )

            strategy["SalesPerCustomer"] = (
                strategy["TotalSales"] /
                strategy["TotalCustomers"]
            )

            fig = px.scatter(
                strategy,
                x="AssortmentName",
                y="StoreType",
                size="AvgSales",
                color="SalesPerCustomer",

                color_continuous_scale=[
                    [0,CREAM],
                    [.5,LIGHT_GOLD],
                    [1,NAVY]
                ],

                size_max=55,

                hover_data={
                    "Stores":True,
                    "AvgSales":":,.0f",
                    "AvgCustomers":":,.0f",
                    "SalesPerCustomer":":.2f"
                }
            )

            clean_layout(
                fig,
                "Store Format & Assortment Strategic Positioning",
                "Assortment Strategy",
                "Store Type",
                480
            )

            fig.update_layout(
                coloraxis_colorbar=dict(
                    title="Sales / Customer"
                )
            )

            fig.show()

        elif task1_view.value == "Competition":

            comp = m["CompStore"].copy()

            comp = comp[
                comp["CompetitionDistance"].gt(0)
            ]

            if comp.empty:

                display(HTML(
                    f"""
                    <div style="
                        padding:15px;
                        background:{CREAM};
                        border-left:4px solid {GOLD};
                        color:{SLATE};
                    ">
                        Competition-distance evidence is unavailable
                        for the current scope.
                    </div>
                    """
                ))

            else:

                comp["DistanceKm"] = (
                    comp["CompetitionDistance"] /
                    1000
                )

                q = min(
                    8,
                    comp["DistanceKm"].nunique()
                )

                response = pd.DataFrame()

                if q > 1:

                    comp["Band"] = pd.qcut(
                        comp["DistanceKm"],
                        q=q,
                        duplicates="drop"
                    )

                    response = (
                        comp.groupby(
                            "Band",
                            observed=True
                        )
                        .agg(
                            Distance=(
                                "DistanceKm",
                                "median"
                            ),
                            Sales=(
                                "AvgSales",
                                "median"
                            ),
                            Customers=(
                                "AvgCustomers",
                                "median"
                            ),
                            Stores=(
                                "Store",
                                "size"
                            )
                        )
                        .reset_index()
                        .sort_values("Distance")
                    )

                fig = go.Figure()

                fig.add_trace(
                    go.Scatter(
                        x=comp["DistanceKm"],
                        y=comp["AvgSales"],
                        mode="markers",

                        marker=dict(
                            color=LIGHT_GOLD,
                            size=6,
                            opacity=.28
                        ),

                        text=comp["Store"],

                        hovertemplate=(
                            "Store %{text}"
                            "<br>Distance: %{x:.2f} km"
                            "<br>Average sales: %{y:,.0f}"
                            "<extra></extra>"
                        ),

                        name="Stores"
                    )
                )

                if not response.empty:

                    fig.add_trace(
                        go.Scatter(
                            x=response["Distance"],
                            y=response["Sales"],
                            mode="lines+markers",

                            line=dict(
                                color=NAVY,
                                width=3
                            ),

                            marker=dict(
                                color=GOLD,
                                size=10,
                                line=dict(
                                    color=BLUE,
                                    width=1
                                )
                            ),

                            customdata=response[
                                ["Stores","Customers"]
                            ],

                            hovertemplate=(
                                "Median distance: %{x:.2f} km"
                                "<br>Median sales: %{y:,.0f}"
                                "<br>Stores: %{customdata[0]}"
                                "<br>Median customers: %{customdata[1]:,.0f}"
                                "<extra></extra>"
                            ),

                            name="Distance Response"
                        )
                    )

                clean_layout(
                    fig,
                    "Competitive Proximity & Store Performance",
                    "Nearest Competitor Distance (km)",
                    "Average Store Sales",
                    490
                )

                fig.update_layout(
                    legend=dict(
                        orientation="h",
                        y=-.17
                    )
                )

                fig.show()

        else:

            evidence = pd.DataFrame({

                "Task 1 Dimension":[
                    "Data Completeness",
                    "Sales Distribution",
                    "Customer Demand",
                    "Sales per Customer",
                    "Train–Test Promotion Alignment",
                    "Promotion Effectiveness",
                    "Promotion Deployment",
                    "Operating Availability",
                    "Weekend Trading",
                    "State Holidays",
                    "School Holidays",
                    "Seasonality",
                    "Assortment Strategy",
                    "Competition Distance",
                    "Competitor Entry",
                    "Competition Availability",
                    "Forecasting Readiness"
                ],

                "Executive Evidence":[

                    f"Core-field completeness = "
                    f"{m['Completeness']:.1f}%",

                    f"Median sales = {m['MedianSales']:,.0f}; "
                    f"P90 = {m['P90Sales']:,.0f}; "
                    f"CV = {m['SalesCV']:.1f}%",

                    f"Sales–customer correlation = "
                    f"{m['Corr']:.3f}",

                    f"Average sales/customer = "
                    f"{m['SPC']:.2f}",

                    (
                        f"Train–test promo-rate gap = "
                        f"{promo_gap:.2f} percentage points"
                        if pd.notna(promo_gap)
                        else "Test comparison unavailable"
                    ),

                    (
                        f"Promo vs non-promo sales = "
                        f"{m['PromoEffect']:+.1f}%"
                        if pd.notna(m["PromoEffect"])
                        else "Comparison unavailable"
                    ),

                    "Store-level promotional opportunity "
                    "and targeting analysed",

                    f"Operating coverage = "
                    f"{m['OpenRate']:.1f}%",

                    (
                        f"Weekend vs weekday sales = "
                        f"{m['WeekendEffect']:+.1f}%"
                        if pd.notna(m["WeekendEffect"])
                        else "Comparison unavailable"
                    ),

                    (
                        f"State-holiday sales difference = "
                        f"{m['HolidayEffect']:+.1f}%"
                        if pd.notna(m["HolidayEffect"])
                        else "Insufficient comparison"
                    ),

                    (
                        f"School-holiday sales difference = "
                        f"{m['SchoolEffect']:+.1f}%"
                        if pd.notna(m["SchoolEffect"])
                        else "Insufficient comparison"
                    ),

                    (
                        f"Peak month = {m['PeakMonth']}; "
                        f"{m['PeakEffect']:+.1f}% vs average"
                        if pd.notna(m["PeakEffect"])
                        else f"Peak month = {m['PeakMonth']}"
                    ),

                    (
                        f"Leading assortment = "
                        f"{m['TopAssortment']}; "
                        f"best–lowest gap = "
                        f"{m['AssortmentEffect']:+.1f}%"
                        if pd.notna(m["AssortmentEffect"])
                        else
                        f"Leading assortment = "
                        f"{m['TopAssortment']}"
                    ),

                    (
                        f"Median competitor distance = "
                        f"{m['CompetitionDistance']:.2f} km"
                        if pd.notna(m["CompetitionDistance"])
                        else "Unavailable"
                    ),

                    (
                        f"Median post-entry sales effect = "
                        f"{m['EntryEffect']:+.1f}%"
                        if pd.notna(m["EntryEffect"])
                        else "Insufficient matched evidence"
                    ),

                    f"Originally observed CompetitionDistance = "
                    f"{m['CompetitionObserved']:.1f}%",

                    f"{m['Stores']:,} stores and "
                    f"{m['OpenDays']:,} open store-days "
                    "available for behavioural analysis"
                ],

                "Management Relevance":[
                    "Data reliability",
                    "Demand variability",
                    "Traffic-led demand",
                    "Customer value",
                    "Train–test consistency",
                    "Campaign effectiveness",
                    "Promotion targeting",
                    "Capacity planning",
                    "Weekend resource allocation",
                    "Holiday preparedness",
                    "Calendar planning",
                    "Inventory and staffing",
                    "Product-range strategy",
                    "Competitive market context",
                    "Competitive resilience",
                    "Data-quality governance",
                    "Forecast development"
                ]
            })

            display(
                evidence.style
                .set_properties(**{
                    "text-align":"left",
                    "padding":"9px",
                    "color":SLATE,
                    "font-size":"11.5px"
                })
                .set_table_styles([
                    {
                        "selector":"th",
                        "props":[
                            ("background-color",NAVY),
                            ("color",WHITE),
                            ("font-weight","bold"),
                            ("padding","10px")
                        ]
                    },
                    {
                        "selector":"tr:nth-child(even)",
                        "props":[
                            ("background-color",CREAM)
                        ]
                    }
                ])
            )
        top3 = ranked[:3]
        top_text = (
            ", ".join(
                f"<strong>{k}</strong> ({v:+.1f}%)"
                for k,v in top3
            )
            if top3
            else "No comparable driver effects are available"
        )

        demand_text = (
            "The strong association between customer traffic and sales "
            "indicates that customer volume is a major component of "
            "realised store demand."
            if abs(m["Corr"]) >= .70
            else
            "Sales variation reflects both customer traffic and "
            "purchasing intensity, indicating a more diversified "
            "demand structure."
        )

        promo_text = (
            "Promotional activity demonstrates a positive aggregate "
            "sales association; nevertheless, deployment should remain "
            "store-specific because response is heterogeneous."
            if pd.notna(m["PromoEffect"]) and m["PromoEffect"] > 0
            else
            "The aggregate promotional comparison does not support "
            "universal promotional expansion under the current scope."
        )

        weekend_text = (
            "Protect inventory availability and staffing capacity "
            "during stronger weekend trading periods."
            if pd.notna(m["WeekendEffect"]) and m["WeekendEffect"] > 0
            else
            "Evaluate weekend operating intensity at store level "
            "rather than increasing capacity uniformly."
        )

        panels = [

            (
                "KEY EXECUTIVE INSIGHTS",
                GOLD,
                f"""
                The most pronounced descriptive signals are
                {top_text}.<br><br>

                Sales and customer traffic correlate at
                <strong>{m['Corr']:.3f}</strong>, while average
                purchasing intensity is
                <strong>{m['SPC']:.2f}</strong> per customer.<br><br>

                <strong>{m['PeakMonth']}</strong> is the strongest
                seasonal month and
                <strong>{m['TopAssortment']}</strong> leads current
                assortment performance.
                """
            ),

            (
                "BUSINESS INTERPRETATION",
                BLUE,
                f"""
                {demand_text}<br><br>

                {promo_text}<br><br>

                Store performance reflects the interaction of customer
                behaviour, promotional exposure, operating availability,
                calendar effects, assortment configuration and competitive
                conditions. These relationships remain descriptive rather
                than causal.
                """
            ),

            (
                "MANAGEMENT RECOMMENDATIONS",
                NAVY,
                f"""
                • Prioritise promotional investment where store-level
                  incremental response is demonstrated.<br><br>

                • {weekend_text}<br><br>

                • Increase inventory and staffing preparedness ahead of
                  identified seasonal peaks.<br><br>

                • Benchmark stronger assortment configurations within
                  comparable store formats.
                """
            ),

            (
                "FORECASTING IMPLICATIONS",
                GOLD,
                """
                • Retain Open, Promo, day-of-week, holiday and seasonal
                  calendar variables.<br><br>

                • Preserve StoreType and Assortment as structural store
                  characteristics.<br><br>

                • Include CompetitionDistance and competitor-opening
                  timing as market-context variables.<br><br>

                • Preserve original missingness indicators alongside
                  imputed competition values.
                """
            )
        ]

        display(panel_grid(panels))

        display(HTML(f"""
        <div style="
            margin-top:18px;
            padding:20px 22px;
            background:linear-gradient(135deg,{NAVY},{BLUE});
            border-left:5px solid {GOLD};
            border-radius:9px;
        ">

            <div style="
                color:{LIGHT_GOLD};
                font-size:10px;
                font-weight:700;
                letter-spacing:1.5px;
                margin-bottom:7px;
            ">
                TASK 1 · EXECUTIVE DECISION CONCLUSION
            </div>

            <div style="
                color:{CREAM};
                font-size:13px;
                line-height:1.85;
            ">
                The completed exploratory analysis identifies a
                <strong>multi-dimensional retail demand environment</strong>
                in which Rossmann sales are jointly associated with customer
                traffic, purchasing intensity, promotional activity, store
                availability, weekend behaviour, holidays, seasonality,
                assortment strategy and competitive market conditions.

                Under the selected scope,
                <strong>{strongest}</strong>
                {
                    f" ({strongest_value:+.1f}%)"
                    if pd.notna(strongest_value)
                    else ""
                }
                represents the strongest comparable descriptive signal.

                The subsequent forecasting framework should therefore adopt
                a <strong>store-aware, calendar-aware, promotion-aware and
                competition-aware multi-factor specification</strong>, while
                retaining identified data-quality indicators alongside
                imputed information.
            </div>

        </div>
        """))

def reset_task1(_):

    task1_view.value = "Scorecard"
    task1_year.value = "All"
    task1_store.value = "All"
    task1_promo.value = "All"
    task1_day.value = "All"

task1_reset.on_click(reset_task1)
for control in [
    task1_view,
    task1_year,
    task1_store,
    task1_promo,
    task1_day
]:
    control.observe(
        render_task1,
        names="value"
    )

display(HTML(f"""
<div style="
    padding:22px 25px;
    margin:6px 0 14px;
    background:linear-gradient(
        115deg,
        {NAVY} 0%,
        {BLUE} 76%,
        {NAVY} 100%
    );
    border-bottom:4px solid {GOLD};
    border-radius:11px;
">

    <div style="
        color:{LIGHT_GOLD};
        font-size:10px;
        font-weight:700;
        letter-spacing:2px;
        margin-bottom:6px;
    ">
        TASK 1 · EXECUTIVE EDA & BUSINESS INTELLIGENCE
    </div>

    <div style="
        color:{WHITE};
        font:500 28px Georgia;
        margin-bottom:7px;
    ">
        Rossmann Customer, Store & Market Intelligence Hub
    </div>

    <div style="
        color:{CREAM};
        font-size:12.5px;
        line-height:1.65;
        max-width:1100px;
    ">
        Consolidated decision intelligence covering customer demand,
        purchasing behaviour, promotional effectiveness, store operations,
        holidays, seasonality, assortment strategy, competitive conditions,
        analytical readiness and forecasting implications.
    </div>

</div>
"""))

display(

    widgets.HBox(
        [
            task1_view,
            task1_year,
            task1_store
        ],
        layout=widgets.Layout(
            gap="9px",
            flex_flow="row wrap",
            margin="0 0 7px 0"
        )
    ),

    widgets.HBox(
        [
            task1_promo,
            task1_day,
            task1_reset
        ],
        layout=widgets.Layout(
            gap="9px",
            flex_flow="row wrap",
            margin="0 0 9px 0"
        )
    ),

    task1_out
)

render_task1()
logger.info("Final Task 1 executive EDA intelligence hub initialized.")
```

From Exploratory Evidence to Forecasting Readiness



The exploratory findings are now consolidated into modelling decisions by
identifying reliable predictive signals, potential leakage, structural data
